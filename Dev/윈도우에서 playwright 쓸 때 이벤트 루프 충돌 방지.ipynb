1. **Playwright의 제약 (Windows)**:
   - Windows에서 Playwright(Chromium)가 정상 작동하려면 Python의 비동기 정책이 반드시 **`WindowsProactorEventLoopPolicy`**로 설정되어 있어야 합니다.
2. **Jupyter Notebook / FastAPI 등에서의 충돌**:
   - 사용자가 **Jupyter Notebook(IPython)**이나 Tornado, FastAPI 같은 환경에서 코드를 실행하면, 이미 해당 프로그램들이 자체적인 이벤트 루프를 쥐고 있어 Playwright 실행 시 다음과 같은 에러를 뿜으며 뻗어버립니다:
     - `WinError 10014: A request to send or receive data was disallowed...`
     - `RuntimeError: This event loop is already running`
3. **해결책 (프로세스 격리)**:
   - 부모 프로세스가 주피터든 어떤 환경이든 상관없이, **새롭고 깨끗한 독립 파이썬 프로세스**를 띄워서 Proactor 정책을 세팅한 후 브라우저를 띄우면 **어떤 환경에서도 100% 충돌 없이 안전하게 PDF가 렌더링**됩니다.
