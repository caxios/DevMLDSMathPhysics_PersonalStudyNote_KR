```python
import scipy.stats as stats

df = 58
t_val = 2.00
p_val_one_tailed = 1 - stats.t.cdf(t_val, df=df)
p_val_two_tailed = 2 * (1 - stats.t.cdf(t_val, df=df))

print(f"One-tailed p-value: {p_val_one_tailed:.6f}")
print(f"Two-tailed p-value: {p_val_two_tailed:.6f}")


```

### 1. 직관적인 요약 및 결론

**유의확률** (p-value, $p$ 값) 은 **"귀무가설(효과나 차이가 없다는 기본 전제)이 참이라고 가정했을 때, 우리가 관측한 데이터만큼 극단적이거나 그보다 더 극단적인 표본 결과가 순전히 우연에 의해 나타날 확률"** 입니다.

* **직관적 의미**: $p$ 값은 관측된 표본 데이터가 귀무가설의 가정과 얼마나 부합하는지를 나타내는 지표입니다.
* **결론 및 판단 기준**:
* $p$ **값이 매우 작을 때** (통상적으로 설정한 기준치인 유의수준 $\alpha = 0.05$ 미만): "귀무가설이 참인 세상이라면 이런 극단적인 표본이 순전히 표본 오차(우연)로 관측될 가능성이 희박하다"는 것을 의미합니다. 따라서 귀무가설을 기각하고 대립가설(효과나 차이가 존재함)을 채택합니다.
* $p$ **값이 클 때**: "귀무가설이 참일 때도 충분히 우연히 관측될 수 있는 평범한 결과"라는 뜻이므로, 귀무가설을 기각할 통계적 근거를 확보하지 못했다고 판단합니다.

### 2. 가설검정의 수학적 정식화

유의확률을 엄밀하게 다루기 위해 통계적 가설검정 체계를 확률공간과 모수공간 상에서 정의합니다.

1. **확률모형 및 표본공간**:
확률공간 $(\Omega, \mathcal{F}, \mathbb{P})$ 과 관측 가능한 표본공간 $(\mathcal{X}, \mathcal{B}_{\mathcal{X}})$ 을 정의합니다. 관측 데이터는 확률벡터 $\mathbf{X} = (X_1, X_2, \dots, X_n)^\top : \Omega \to \mathcal{X}$ 이며, 실현된 표본 관측치를 $\mathbf{x}_{\text{obs}} \in \mathcal{X}$ 라 합니다.
2. **모수공간과 가설**:
분포족을 $\mathcal{P} = \{\mathbb{P}_\theta : \theta \in \Theta\}$ 라 할 때, 모수공간 $\Theta$ 를 서로소인 두 부분집합으로 분할합니다:

$$\Theta = \Theta_0 \cup \Theta_1, \quad \Theta_0 \cap \Theta_1 = \emptyset$$


* 귀무가설(Null Hypothesis): $H_0: \theta \in \Theta_0$
* 대립가설(Alternative Hypothesis): $H_1: \theta \in \Theta_1$


3. **검정통계량 (Test Statistic)**:
보렐 가측 함수 $T: (\mathcal{X}, \mathcal{B}_{\mathcal{X}}) \to (\mathbb{R}, \mathcal{B}_{\mathbb{R}})$ 를 정의합니다. 검정통계량 $T(\mathbf{X})$ 는 표본의 정보를 요약하는 실수값 확률변수이며, 실측값은 $t_{\text{obs}} = T(\mathbf{x}_{\text{obs}})$ 입니다. 통계량 $T$ 는 $H_1$ 이 지지될수록 큰(또는 극단적인) 값을 갖도록 설계됩니다.

### 3. 유의확률($p$-value)의 수학적 엄밀한 정의

#### (1) 단순 귀무가설에서의 정의

귀무가설이 단일 모수로 구성된 경우($H_0: \theta = \theta_0$), 대립가설의 형태에 따른 유의확률은 다음과 같이 정의됩니다.

* **우측 단측 검정** ($H_1: \theta > \theta_0$, $T$ 가 클수록 대립가설 지지):

$$p(\mathbf{x}_{\text{obs}}) = \mathbb{P}_{\theta_0}(T(\mathbf{X}) \ge T(\mathbf{x}_{\text{obs}})) = \int_{\{\mathbf{x} \in \mathcal{X} : T(\mathbf{x}) \ge t_{\text{obs}}\}} \mathrm{d}\mathbb{P}_{\theta_0}(\mathbf{x})$$


* **좌측 단측 검정** ($H_1: \theta < \theta_0$, $T$ 가 작을수록 대립가설 지지):

$$p(\mathbf{x}_{\text{obs}}) = \mathbb{P}_{\theta_0}(T(\mathbf{X}) \le T(\mathbf{x}_{\text{obs}}))$$


* **양측 검정** ($H_1: \theta \neq \theta_0$):
귀무분포가 대칭인 경우,

$$p(\mathbf{x}_{\text{obs}}) = \mathbb{P}_{\theta_0}(\vert{}T(\mathbf{X})\vert{} \ge \vert{}t_{\text{obs}}\vert{}) = 2 \min \left( \mathbb{P}_{\theta_0}(T(\mathbf{X}) \ge t_{\text{obs}}), \ \mathbb{P}_{\theta_0}(T(\mathbf{X}) \le t_{\text{obs}}) \right)$$



#### (2) 복합 귀무가설에서의 정의 (일반화)

귀무가설 집합 $\Theta_0$ 가 단일 원소가 아닌 복합 가설인 경우, 유의확률은 제1종 오류 확률을 보수적으로 상한(Supremum) 통제하도록 정의됩니다:


$$p(\mathbf{x}_{\text{obs}}) = \sup_{\theta \in \Theta_0} \mathbb{P}_\theta (T(\mathbf{X}) \ge T(\mathbf{x}_{\text{obs}}))$$

#### (3) 내포된 기각역 집합족을 통한 동치 정의

각 유의수준 $\alpha \in (0, 1)$ 에 대해 제1종 오류의 크기가 $\alpha$ 이하인 기각역(Rejection Region) 집합족 $\{W_\alpha\}_{\alpha \in (0, 1)}$ 이 내포적(Nested) 구조를 만족한다고 가정합니다:


$$\alpha_1 < \alpha_2 \implies W_{\alpha_1} \subseteq W_{\alpha_2}$$


이때 유의확률 $p(\mathbf{x}_{\text{obs}})$ 는 **"관측된 표본 $\mathbf{x}_{\text{obs}}$ 를 기각역에 포함시킬 수 있는 최소의 유의수준(Infimum of Significance Levels)"** 으로 엄밀하게 정의됩니다:


$$p(\mathbf{x}_{\text{obs}}) = \inf \{ \alpha \in (0, 1) : \mathbf{x}_{\text{obs}} \in W_\alpha \}$$

### 4. 확률변수로서의 유의확률과 분포적 성질

표본 관측치 $\mathbf{x}_{\text{obs}}$ 가 주어지면 $p(\mathbf{x}_{\text{obs}})$ 는 특정 실수값이 되지만, 표본 관측 전의 $\mathbf{X}$ 를 대입하면 유의확률은 확률변수 $P(\mathbf{X})$ 가 됩니다.

#### (1) 귀무가설 하에서의 균등분포 성질

$H_0: \theta = \theta_0$ 하에서 검정통계량 $T(\mathbf{X})$ 의 누적분포함수(CDF)를 $F_0(t) = \mathbb{P}_{\theta_0}(T(\mathbf{X}) \le t)$ 라 할 때, $F_0$ 가 연속이고 순증가(strictly increasing) 함수라고 가정합니다.

우측 검정에서의 유의확률 확률변수는 $P(\mathbf{X}) = 1 - F_0(T(\mathbf{X}))$ 입니다. 임의의 $u \in [0, 1]$ 에 대해 $P(\mathbf{X})$ 의 분포를 구하면:


$$\mathbb{P}_{\theta_0}(P(\mathbf{X}) \le u) = \mathbb{P}_{\theta_0}(1 - F_0(T(\mathbf{X})) \le u) = \mathbb{P}_{\theta_0}(F_0(T(\mathbf{X})) \ge 1 - u)$$

확률적분변환(Probability Integral Transform)에 의해 $F_0(T(\mathbf{X})) \overset{H_0}{\sim} \text{Uniform}(0, 1)$ 이므로,


$$\mathbb{P}_{\theta_0}(F_0(T(\mathbf{X})) \ge 1 - u) = 1 - (1 - u) = u$$

따라서 다음이 성립합니다:


$$P(\mathbf{X}) \overset{H_0}{\sim} \text{Uniform}(0, 1)$$

이 성질을 통해 기각 규칙 $\delta(\mathbf{X}) = \mathbf{1}_{\{P(\mathbf{X}) \le \alpha\}}$ 의 제1종 오류 확률을 계산하면:


$$\mathbb{P}_{\theta_0}(P(\mathbf{X}) \le \alpha) = \alpha$$


즉, $p \le \alpha$ 기준을 사용하는 검정법이 유의수준 $\alpha$ 를 정확하게 통제함을 수학적으로 증명할 수 있습니다.

#### (2) 대립가설 하에서의 거동

대립가설 $H_1: \theta = \theta_1$ 하에서 검정통계량이 확률적으로 더 큰 값을 취할 때($F_1(t) \le F_0(t)$), 유의확률의 누적분포함수는 다음과 같습니다:


$$\mathbb{P}_{\theta_1}(P(\mathbf{X}) \le \alpha) = \mathbb{P}_{\theta_1}(F_0(T(\mathbf{X})) \ge 1 - \alpha) = 1 - F_1(F_0^{-1}(1 - \alpha)) > \alpha$$


이 값은 검정력(Power, $1 - \beta$)에 해당하며, 대립가설 하에서 $P(\mathbf{X})$ 의 확률질량은 0 근처로 강하게 편향됩니다.

### 5. 주요 통계적 오개념의 수식적 교정

1. **$p$ 값은 귀무가설이 참일 확률이 아닙니다**:

$$p \ne \mathbb{P}(H_0 \mid \mathbf{X} = \mathbf{x}_{\text{obs}})$$



빈도주의 통계학에서 모수 $\theta$ 와 가설 $H_0$ 는 확률변수가 아닌 고정된 상태이므로 $\mathbb{P}(H_0)$ 자체가 정의되지 않습니다. 베이즈 정리를 적용할 경우:

$$\mathbb{P}(H_0 \mid \mathbf{X}) = \frac{\mathbb{P}(\mathbf{X} \mid H_0)\mathbb{P}(H_0)}{\mathbb{P}(\mathbf{X})}$$



사후확률은 사전확률 $\mathbb{P}(H_0)$ 에 의존하지만, 유의확률은 귀무가설을 참으로 가정한 상태에서 산출된 표본의 조건부 꼬리 확률 $\mathbb{P}_{\theta_0}(T \ge t_{\text{obs}})$ 입니다.
2. **$1 - p$ 는 대립가설이 참일 확률이 아닙니다**:
마찬가지로 $1 - p \ne \mathbb{P}(H_1 \mid \mathbf{X} = \mathbf{x}_{\text{obs}})$ 입니다.
3. **$p > \alpha$ 는 귀무가설이 참이라는 증거가 아닙니다**:
$p > \alpha$ 는 단지 현재 주어진 표본 크기와 노이즈 하에서 귀무가설을 기각할 충분한 통계적 증거를 찾지 못했다(Failure to Reject)는 의미일 뿐, $H_0$ 가 참임을 입증(Accept)한 것이 아닙니다.

### 6. 금융 통계에서의 응용 예시: 펀드 매니저의 젠센 알파(Jensen's Alpha) 검정

#### (1) 금융 모형 설정

자산가격결정모형(CAPM)에 기반하여 어떤 액티브 주식형 펀드의 초과수익률 회귀식을 수립합니다:


$$R_{p, t} - R_{f, t} = \alpha_p + \beta_p (R_{m, t} - R_{f, t}) + \epsilon_t, \quad t = 1, 2, \dots, n$$

* $R_{p, t}$: $t$ 시점 펀드 수익률
* $R_{m, t}$: $t$ 시점 벤치마크 시장 수익률
* $R_{f, t}$: 무위험 이자율
* $\epsilon_t \overset{\text{i.i.d.}}{\sim} \mathcal{N}(0, \sigma^2)$: 오차항
* $\alpha_p$: 시장 위험을 조정한 후 펀드 매니저가 창출한 고유 초과수익(젠센 알파)

#### (2) 가설 설정

투자자는 매니저가 운이 아닌 실제 초과 수익 창출 능력을 보유하고 있는지 검정하고자 합니다:

* 귀무가설 $H_0: \alpha_p \le 0$ (초과 수익 창출 능력이 없음)
* 대립가설 $H_1: \alpha_p > 0$ (유의미한 초과 수익 창출 능력이 존재함)

#### (3) 데이터 및 검정통계량 도출

5년간의 월별 데이터($n = 60$)를 수집하여 보통최소제곱법(OLS)으로 추정한 결과가 다음과 같다고 가정합니다:

* 월간 알파 추정치: $\hat{\alpha}_p = 0.40\%$ (연율화 약 $4.8\%$)
* 알파 추정량의 표준오차: $\text{SE}(\hat{\alpha}_p) = 0.20\%$

검정통계량 $T$ 는 다음과 같이 $t$ 통계량으로 구성됩니다:


$$t_{\text{obs}} = \frac{\hat{\alpha}_p - 0}{\text{SE}(\hat{\alpha}_p)} = \frac{0.40\%}{0.20\%} = 2.00$$

귀무가설의 경계($\alpha_p = 0$) 하에서 통계량 $T$ 는 자유도 $n - 2 = 58$ 인 스튜던트 $t$ 분포를 따릅니다:


$$T \overset{H_0}{\sim} t(58)$$

#### (4) 유의확률 계산

우측 단측 검정이므로, 자유도 $58$ 인 $t$ 분포에서 $t \ge 2.00$ 일 확률을 계산합니다:


$$p = \mathbb{P}(T \ge 2.00 \mid T \sim t(58)) = \int_{2.00}^{\infty} f_{t(58)}(u) \, \mathrm{d}u \approx 0.0251 \quad (2.51\%)$$

#### (5) 금융적 의미 및 의사결정

* **수치적 해석**: 만약 해당 매니저의 실제 역량이 시장 평균 이하($\alpha_p \le 0$)임에도 불구하고, 5년 동안 순전히 시장 노이즈와 무작위적인 가격 변동(운)에 의해 표본 알파가 이 정도로 높게($t \ge 2.00$) 관측될 확률은 약 $2.51\%$ 에 불과합니다.
* **최종 결론**: 통상적인 금융 연구의 유의수준 $\alpha_{\text{level}} = 0.05$ ($5\%$)를 기준으로 할 때,

$$p \approx 0.0251 < 0.05$$



이므로 귀무가설 $H_0$ 를 기각합니다. 따라서 "해당 매니저가 통계적으로 유의미한 초과수익 능력을 지니고 있다"는 대립가설을 통계적 유의성 하에서 채택하게 됩니다.
