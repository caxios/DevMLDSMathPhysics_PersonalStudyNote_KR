```python
import numpy as np

# Let's set up a clean 2D, 2-class example.
# Class 1: mu1 = [2, 2]
# Class 2: mu2 = [6, 4]

# For LDA:
# Common covariance Sigma
Sigma = np.array([[2, 1],
                  [1, 2]])
inv_Sigma = np.linalg.inv(Sigma)
det_Sigma = np.linalg.det(Sigma)

print("inv_Sigma:\n", inv_Sigma)
print("det_Sigma:", det_Sigma)

# mu1 = [2, 2]^T, mu2 = [6, 4]^T
mu1 = np.array([2, 2])
mu2 = np.array([6, 4])
pi1 = 0.5
pi2 = 0.5

# Test point x = [4, 3]^T or [3, 2]^T
x = np.array([3, 2])

# LDA weights:
# w_k = inv_Sigma @ mu_k
# c_k = -0.5 * mu_k.T @ inv_Sigma @ mu_k + np.log(pi_k)

w1 = inv_Sigma @ mu1
c1 = -0.5 * mu1.T @ inv_Sigma @ mu1 + np.log(pi1)
delta1_lda = x.T @ w1 + c1

w2 = inv_Sigma @ mu2
c2 = -0.5 * mu2.T @ inv_Sigma @ mu2 + np.log(pi2)
delta2_lda = x.T @ w2 + c2

print("LDA:")
print("w1:", w1, "c1:", c1, "delta1:", delta1_lda)
print("w2:", w2, "c2:", c2, "delta2:", delta2_lda)

# Difference delta1 - delta2:
# (w1 - w2)^T x + (c1 - c2) = 0
print("w1 - w2:", w1 - w2)
print("c1 - c2:", c1 - c2)

# For QDA:
# Sigma1 and Sigma2 are different
Sigma1 = np.array([[1, 0],
                   [0, 1]]) # identity
Sigma2 = np.array([[2, 1],
                   [1, 2]])

inv_Sigma1 = np.linalg.inv(Sigma1)
det_Sigma1 = np.linalg.det(Sigma1)

inv_Sigma2 = np.linalg.inv(Sigma2)
det_Sigma2 = np.linalg.det(Sigma2)

# QDA discriminant:
# delta_k(x) = -0.5 * log(det(Sigma_k)) - 0.5 * (x - mu_k)^T inv_Sigma_k (x - mu_k) + log(pi_k)
d1_qda = -0.5 * np.log(det_Sigma1) - 0.5 * (x - mu1).T @ inv_Sigma1 @ (x - mu1) + np.log(pi1)
d2_qda = -0.5 * np.log(det_Sigma2) - 0.5 * (x - mu2).T @ inv_Sigma2 @ (x - mu2) + np.log(pi2)

print("\nQDA:")
print("det_Sigma1:", det_Sigma1, "inv_Sigma1:\n", inv_Sigma1)
print("det_Sigma2:", det_Sigma2, "inv_Sigma2:\n", inv_Sigma2)
print("d1_qda:", d1_qda)
print("d2_qda:", d2_qda)


```

**선형 판별분석(LDA, Linear Discriminant Analysis)** 과 **이차 판별분석(QDA, Quadratic Discriminant Analysis)** 은 관측된 $p$ 차원 특징 벡터 $x$ 가 주어졌을 때, 해당 관측치가 어떤 범주 $Y \in \{1, 2, \dots, K\}$ 에 속하는지 분류하는 확률적 생성 모델(Generative Model)입니다.

---

## 1. 확률적 기저: 베이즈 정리와 정규분포 가정

분류의 기본 원리는 벡터 $x$ 가 관측되었을 때 클래스 $k$ 에 속할 **사후확률(Posterior Probability)** 을 최대화하는 $k$ 를 선택하는 것입니다. **베이즈 정리(Bayes' Theorem)** 에 따라 사후확률은 다음과 같이 정의됩니다.

$$P(Y=k \mid X=x) = \frac{P(X=x \mid Y=k) P(Y=k)}{P(X=x)} = \frac{f_k(x) \pi_k}{\sum_{l=1}^K f_l(x) \pi_l}$$

* $\pi_k = P(Y=k)$ : 클래스 $k$ 의 **사전확률(Prior Probability)**
* $f_k(x) = P(X=x \mid Y=k)$ : 클래스 $k$ 에서 $x$ 가 관측될 **조건부 확률밀도(Likelihood)**

분모인 $\sum_{l=1}^K f_l(x) \pi_l$ 는 모든 클래스 $k$ 에 대해 동일한 정규화 상수이므로, 사후확률의 대소 비교는 분자 $f_k(x) \pi_k$ 에 의해서만 결정됩니다.

LDA와 QDA는 클래스별 밀도함수 $f_k(x)$ 가 평균 벡터 $\mu_k$ 와 공분산 행렬 $\Sigma_k$ 를 갖는 **다변량 정규분포(Multivariate Normal Distribution)** $\mathcal{N}(\mu_k, \Sigma_k)$ 를 따른다고 가정합니다.

$$f_k(x) = \frac{1}{(2\pi)^{p/2} \vert{}\Sigma_k\vert{}^{1/2}} \exp\left( -\frac{1}{2} (x - \mu_k)^T \Sigma_k^{-1} (x - \mu_k) \right)$$

---

## 2. QDA (이차 판별분석)의 수식 유도

QDA는 **각 클래스마다 고유한 공분산 행렬을 가진다( $\Sigma_k \neq \Sigma_j$ )** 고 가정합니다.

사후확률 $P(Y=k \mid X=x) \propto f_k(x)\pi_k$ 에 자연로그를 취하면 로그 사후확률이 도출됩니다.

$$\ln \left( f_k(x) \pi_k \right) = -\frac{p}{2} \ln(2\pi) - \frac{1}{2} \ln \vert{}\Sigma_k\vert{} - \frac{1}{2} (x - \mu_k)^T \Sigma_k^{-1} (x - \mu_k) + \ln \pi_k$$

모든 클래스 $k$ 에 공통으로 포함된 상수항 $-\frac{p}{2}\ln(2\pi)$ 를 제외하면, QDA의 **이차 판별함수(Quadratic Discriminant Function)** $\delta_k^{\text{QDA}}(x)$ 가 정의됩니다.

$$\delta_k^{\text{QDA}}(x) = -\frac{1}{2} \ln \vert{}\Sigma_k\vert{} - \frac{1}{2} (x - \mu_k)^T \Sigma_k^{-1} (x - \mu_k) + \ln \pi_k$$

이차형식(Quadratic Form) 항을 $x$ 에 대해 전개하면 다음과 같습니다.

$$(x - \mu_k)^T \Sigma_k^{-1} (x - \mu_k) = x^T \Sigma_k^{-1} x - 2 x^T \Sigma_k^{-1} \mu_k + \mu_k^T \Sigma_k^{-1} \mu_k$$

두 클래스 $k$ 와 $j$ 사이의 결정 경계(Decision Boundary)는 두 판별함수 값이 같아지는 지점( $\delta_k^{\text{QDA}}(x) = \delta_j^{\text{QDA}}(x)$ )입니다.

$$\delta_k^{\text{QDA}}(x) - \delta_j^{\text{QDA}}(x) = -\frac{1}{2} x^T \left( \Sigma_k^{-1} - \Sigma_j^{-1} \right) x + x^T \left( \Sigma_k^{-1} \mu_k - \Sigma_j^{-1} \mu_j \right) + C_{k,j} = 0$$

$\Sigma_k \neq \Sigma_j$ 이므로 최고차항인 $x^T (\Sigma_k^{-1} - \Sigma_j^{-1}) x$ 가 상쇄되지 않고 남아, 결정 경계는 **$x$ 에 대한 2차 곡면(타원, 포물선, 쌍곡선 등)** 을 이룹니다.

---

## 3. LDA (선형 판별분석)의 수식 유도

LDA는 **모든 클래스가 동일한 공분산 행렬을 공유한다( $\Sigma_k = \Sigma, \; \forall k$ )** 는 등분산 가정을 도입합니다.

QDA 판별식에 $\Sigma_k = \Sigma$ 를 대입하고 전개합니다.

$$\ln \left( f_k(x) \pi_k \right) = -\frac{p}{2}\ln(2\pi) - \frac{1}{2}\ln\vert{}\Sigma\vert{} - \frac{1}{2} x^T \Sigma^{-1} x + x^T \Sigma^{-1} \mu_k - \frac{1}{2}\mu_k^T \Sigma^{-1} \mu_k + \ln \pi_k$$

여기서 다음 세 항은 클래스 인덱스 $k$ 와 완전히 독립적입니다.

1. $-\frac{p}{2}\ln(2\pi)$ (상수)
2. $-\frac{1}{2}\ln\vert{}\Sigma\vert{}$ (공통 공분산 행렬식)
3. $-\frac{1}{2} x^T \Sigma^{-1} x$ (클래스 구분에 영향을 주지 않는 $x$ 의 공통 이차형식)

따라서 클래스 간 대소 비교 시 이 세 항을 모두 소거할 수 있으며, 남은 항들로 **선형 판별함수(Linear Discriminant Function)** $\delta_k^{\text{LDA}}(x)$ 가 도출됩니다.

$$\delta_k^{\text{LDA}}(x) = x^T \Sigma^{-1} \mu_k - \frac{1}{2} \mu_k^T \Sigma^{-1} \mu_k + \ln \pi_k$$

두 클래스 $k$ 와 $j$ 의 결정 경계 $\delta_k^{\text{LDA}}(x) = \delta_j^{\text{LDA}}(x)$ 를 정리하면 다음과 같습니다.

$$x^T \Sigma^{-1} (\mu_k - \mu_j) - \frac{1}{2} \mu_k^T \Sigma^{-1} \mu_k + \frac{1}{2} \mu_j^T \Sigma^{-1} \mu_j + \ln\left( \frac{\pi_k}{\pi_j} \right) = 0$$

이 식에서 $x$ 에 대한 최고차항은 1차식이며, 계수 벡터는 $\Sigma^{-1} (\mu_k - \mu_j)$ 입니다. 따라서 결정 경계는 **초평면(Hyperplane, 2차원 공간에서는 직선)** 을 이룹니다.

## 4. 모수(Parameter) 추정 방법

실제 데이터에서는 모집단의 모수를 알 수 없으므로 표본 데이터(총 $N$ 개 관측치, 클래스 $k$ 의 표본 수 $N_k$ )로부터 다음의 비편향 추정량을 사용합니다.

1. **사전확률:** $\hat{\pi}_k = \frac{N_k}{N}$
2. **평균 벡터:** $\hat{\mu}_k = \frac{1}{N_k} \sum_{i \in C_k} x_i$
3. **QDA 표본 공분산 행렬:**

$$\hat{\Sigma}_k = \frac{1}{N_k - 1} \sum_{i \in C_k} (x_i - \hat{\mu}_k)(x_i - \hat{\mu}_k)^T$$


4. **LDA 풀링 공분산 행렬(Pooled Covariance Matrix):**

$$\hat{\Sigma} = \sum_{k=1}^K \frac{N_k - 1}{N - K} \hat{\Sigma}_k$$

## 5. 상세 수치 계산 예제

2차원 특징 공간( $p=2$ )에서 2개 클래스( $K=2$ )를 분류하는 구체적인 수치 예시입니다.

### (1) 데이터 모수 설정

* **사전확률:** $\pi_1 = 0.5$, $\pi_2 = 0.5$ (따라서 $\ln \pi_1 = \ln \pi_2 = \ln 0.5 \approx -0.6931$ )
* **평균 벡터:**

$$\mu_1 = \begin{pmatrix} 0 \\ 0 \end{pmatrix}, \quad \mu_2 = \begin{pmatrix} 2 \\ 2 \end{pmatrix}$$


* **클래스별 공분산 행렬 (QDA용):**

$$\Sigma_1 = \begin{pmatrix} 1 & 0 \\ 0 & 1 \end{pmatrix}, \quad \Sigma_2 = \begin{pmatrix} 2 & 1 \\ 1 & 1 \end{pmatrix}$$


* $\det(\Sigma_1) = 1 \cdot 1 - 0 \cdot 0 = 1 \implies \ln \vert{}\Sigma_1\vert{} = 0$
* $\det(\Sigma_2) = 2 \cdot 1 - 1 \cdot 1 = 1 \implies \ln \vert{}\Sigma_2\vert{} = 0$
* 역행렬:

$$\Sigma_1^{-1} = \begin{pmatrix} 1 & 0 \\ 0 & 1 \end{pmatrix}, \quad \Sigma_2^{-1} = \frac{1}{1} \begin{pmatrix} 1 & -1 \\ -1 & 2 \end{pmatrix} = \begin{pmatrix} 1 & -1 \\ -1 & 2 \end{pmatrix}$$




* **공통 풀링 공분산 행렬 (LDA용, $N_1 = N_2$ 가정):**

$$\Sigma = \frac{1}{2} (\Sigma_1 + \Sigma_2) = \frac{1}{2} \begin{pmatrix} 3 & 1 \\ 1 & 2 \end{pmatrix} = \begin{pmatrix} 1.5 & 0.5 \\ 0.5 & 1.0 \end{pmatrix}$$


* $\det(\Sigma) = 1.5 \times 1.0 - 0.5 \times 0.5 = 1.25 = \frac{5}{4}$
* 역행렬:

$$\Sigma^{-1} = \frac{4}{5} \begin{pmatrix} 1.0 & -0.5 \\ -0.5 & 1.5 \end{pmatrix} = \begin{pmatrix} 0.8 & -0.4 \\ -0.4 & 1.2 \end{pmatrix}$$





---

### (2) 테스트 데이터 포인트 $x = \begin{pmatrix} 1.5 \\ 1.0 \end{pmatrix}$ 분류

#### [LDA 계산 과정]

1. **가중치 벡터 $w_k = \Sigma^{-1} \mu_k$ 계산:**

$$w_1 = \begin{pmatrix} 0.8 & -0.4 \\ -0.4 & 1.2 \end{pmatrix} \begin{pmatrix} 0 \\ 0 \end{pmatrix} = \begin{pmatrix} 0 \\ 0 \end{pmatrix}$$


$$w_2 = \begin{pmatrix} 0.8 & -0.4 \\ -0.4 & 1.2 \end{pmatrix} \begin{pmatrix} 2 \\ 2 \end{pmatrix} = \begin{pmatrix} 0.8(2) - 0.4(2) \\ -0.4(2) + 1.2(2) \end{pmatrix} = \begin{pmatrix} 0.8 \\ 1.6 \end{pmatrix}$$


2. **상수항 $c_k = -\frac{1}{2} \mu_k^T \Sigma^{-1} \mu_k + \ln \pi_k$ 계산:**

$$c_1 = -\frac{1}{2} (0) + \ln 0.5 = -0.6931$$


$$\mu_2^T w_2 = \begin{pmatrix} 2 & 2 \end{pmatrix} \begin{pmatrix} 0.8 \\ 1.6 \end{pmatrix} = 2(0.8) + 2(1.6) = 1.6 + 3.2 = 4.8$$


$$c_2 = -\frac{1}{2} (4.8) + \ln 0.5 = -2.4 - 0.6931 = -3.0931$$


3. **판별값 $\delta_k^{\text{LDA}}(x) = x^T w_k + c_k$ 계산:**

$$\delta_1^{\text{LDA}}(x) = 0 + (-0.6931) = -0.6931$$


$$x^T w_2 = \begin{pmatrix} 1.5 & 1.0 \end{pmatrix} \begin{pmatrix} 0.8 \\ 1.6 \end{pmatrix} = 1.5(0.8) + 1.0(1.6) = 1.2 + 1.6 = 2.8$$


$$\delta_2^{\text{LDA}}(x) = 2.8 - 3.0931 = -0.2931$$


4. **결과 판정:**

$$\delta_2^{\text{LDA}}(x) = -0.2931 > \delta_1^{\text{LDA}}(x) = -0.6931 \implies \mathbf{Class \; 2} \text{ 로 분류}$$



---

#### [QDA 계산 과정]

QDA 판별함수는 $\ln \vert{}\Sigma_k\vert{} = 0$ 이므로 마할라노비스 거리 제곱 $d_{M,k}^2 = (x - \mu_k)^T \Sigma_k^{-1} (x - \mu_k)$ 에 의해 결정됩니다.

$$\delta_k^{\text{QDA}}(x) = -\frac{1}{2} d_{M,k}^2 + \ln 0.5$$

1. **클래스 1과의 거리 계산:**

$$x - \mu_1 = \begin{pmatrix} 1.5 \\ 1.0 \end{pmatrix} - \begin{pmatrix} 0 \\ 0 \end{pmatrix} = \begin{pmatrix} 1.5 \\ 1.0 \end{pmatrix}$$


$$d_{M,1}^2 = \begin{pmatrix} 1.5 & 1.0 \end{pmatrix} \begin{pmatrix} 1 & 0 \\ 0 & 1 \end{pmatrix} \begin{pmatrix} 1.5 \\ 1.0 \end{pmatrix} = 1.5^2 + 1.0^2 = 2.25 + 1.0 = 3.25$$


$$\delta_1^{\text{QDA}}(x) = -\frac{1}{2}(3.25) - 0.6931 = -1.625 - 0.6931 = -2.3181$$


2. **클래스 2와의 거리 계산:**

$$x - \mu_2 = \begin{pmatrix} 1.5 \\ 1.0 \end{pmatrix} - \begin{pmatrix} 2 \\ 2 \end{pmatrix} = \begin{pmatrix} -0.5 \\ -1.0 \end{pmatrix}$$


$$\Sigma_2^{-1} (x - \mu_2) = \begin{pmatrix} 1 & -1 \\ -1 & 2 \end{pmatrix} \begin{pmatrix} -0.5 \\ -1.0 \end{pmatrix} = \begin{pmatrix} -0.5 + 1.0 \\ 0.5 - 2.0 \end{pmatrix} = \begin{pmatrix} 0.5 \\ -1.5 \end{pmatrix}$$


$$d_{M,2}^2 = \begin{pmatrix} -0.5 & -1.0 \end{pmatrix} \begin{pmatrix} 0.5 \\ -1.5 \end{pmatrix} = (-0.5)(0.5) + (-1.0)(-1.5) = -0.25 + 1.50 = 1.25$$


$$\delta_2^{\text{QDA}}(x) = -\frac{1}{2}(1.25) - 0.6931 = -0.625 - 0.6931 = -1.3181$$


3. **결과 판정:**

$$\delta_2^{\text{QDA}}(x) = -1.3181 > \delta_1^{\text{QDA}}(x) = -2.3181 \implies \mathbf{Class \; 2} \text{ 로 분류}$$

### (3) 결정 경계의 기하학적 형태 비교

두 모델의 결정 경계 방정식( $\delta_1(x) = \delta_2(x)$ )을 직접 정리하면 다음과 같습니다.

* **LDA 결정 경계:**

$$\delta_2^{\text{LDA}}(x) - \delta_1^{\text{LDA}}(x) = 0.8 x_1 + 1.6 x_2 - 2.4 = 0 \iff x_1 + 2 x_2 = 3$$



기울기가 $-0.5$ 인 완전한 **직선(선형 경계)** 입니다.
* **QDA 결정 경계:**

$$\delta_1^{\text{QDA}}(x) - \delta_2^{\text{QDA}}(x) = -\frac{1}{2} (d_{M,1}^2 - d_{M,2}^2) = 0 \iff d_{M,1}^2 = d_{M,2}^2$$


$$(x_1^2 + x_2^2) - \left[ (x_1 - 2)^2 - 2(x_1 - 2)(x_2 - 2) + 2(x_2 - 2)^2 \right] = 0$$



이를 전개하여 정리하면:

$$-2 x_1 x_2 + x_2^2 - 4 x_2 + 4 = 0$$



교차항 $x_1 x_2$ 와 2차항 $x_2^2$ 가 남아 있는 **2차 곡선(쌍곡선 형태)** 을 형성합니다.

예를 들어 점 $x = \begin{pmatrix} 1.0 \\ 1.0 \end{pmatrix}$ 의 경우:

* **LDA:** $1.0 + 2(1.0) = 3$ 이므로 정확히 선형 결정 경계선 위에 위치하여 판별값이 동일합니다( $\delta_1 = \delta_2 = -0.6931$ ).
* **QDA:** $d_{M,1}^2 = 2.0$, $d_{M,2}^2 = 1.0$ 이므로 $\delta_2(-1.1931) > \delta_1(-1.6931)$ 이 되어 클래스 2의 영역으로 분류됩니다. 이는 클래스 2의 공분산 구조( $X_1$ 과 $X_2$ 의 양의 상관관계)가 반영되어 거리 척도가 달라졌기 때문입니다.
