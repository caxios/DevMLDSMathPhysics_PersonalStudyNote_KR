# 모멘트(Moment)와 표준화 모멘트

## 1단계: 직관 — "모멘트"라는 이름은 물리학에서 왔어

막대 위에 무게추들이 놓여 있다고 생각해봐.

- **무게중심**(어디서 균형이 잡히나) → 1차 모멘트
- **관성모멘트**(중심에서 얼마나 멀리 퍼져 있나) → 2차 모멘트

통계의 모멘트도 똑같아. 데이터를 막대 위의 점들로 보고 **분포의 모양을 숫자 몇 개로 요약하는 도구**야. 차수(1, 2, 3, 4…)가 올라갈수록 분포의 더 미세한 특징을 잡아내.

---

## 2단계: 원점 모멘트 (raw moment)

확률변수 X의 **k차 원점 모멘트**는 X를 k제곱한 값의 기댓값이야.

$$
\mu'_k = E[X^k]
$$

- 1차: $E[X] = \mu$ → **평균**, 즉 분포의 위치
- 2차: $E[X^2]$ → 원점(0)을 기준으로 얼마나 퍼졌나

문제는 2차 이상부터 **평균이 어디 있느냐에 따라 값이 오염된다**는 거야. 모든 데이터에 100을 더하면 퍼진 정도는 그대로인데 $E[X^2]$는 폭증하지. 그래서 평균을 기준점으로 옮겨야 해.

---

## 3단계: 중심 모멘트 (central moment)

평균을 빼고 k제곱해.

$$
\mu_k = E[(X-\mu)^k]
$$

차수별로 무엇을 측정하는지 보자.

| 차수 | 수식 | 의미 | 제곱의 성질 |
|---|---|---|---|
| 1 | $E[X-\mu]$ | 항상 0 (정의상) | — |
| 2 | $E[(X-\mu)^2] = \sigma^2$ | **분산**: 퍼진 정도 | 짝수 → 부호 사라짐, 크기만 봄 |
| 3 | $E[(X-\mu)^3]$ | **비대칭**: 어느 쪽으로 쏠렸나 | 홀수 → 부호 살아있음 |
| 4 | $E[(X-\mu)^4]$ | **꼬리 두께**: 극단값이 얼마나 자주 나오나 | 짝수 + 4제곱 → 극단값에 극도로 민감 |

- **홀수 차수**는 부호를 보존하니까 **방향**(좌우 비대칭)을 측정해.
- **짝수 차수**는 부호를 지우니까 **크기**(얼마나 멀리 떨어졌나)를 측정해.
- **차수가 높을수록** 평균에서 멀리 떨어진 값의 영향력이 기하급수적으로 커져. 편차 2는 2제곱이면 4, 4제곱이면 16이야. 그래서 고차 모멘트일수록 꼬리(극단값)를 보는 지표가 되고, 동시에 **노이즈에 취약**해져.

---

## 4단계: 중심 모멘트의 치명적 약점 — 단위

수익률을 **소수**(0.01)로 쓰든 **%**(1)로 쓰든 분포 모양은 똑같아. 그런데 X를 c배 하면:

$$
E[(cX - c\mu)^3] = c^3 \, E[(X-\mu)^3]
$$

c = 100이면 3차 중심 모멘트가 **100만 배**가 돼. 모양은 그대로인데 숫자만 바뀌는 거지.

또 삼성전자(변동성 작음)와 코스닥 바이오주(변동성 큼)의 3차 모멘트를 비교하면, 바이오주는 **단지 변동성이 커서** 3차 모멘트의 절댓값도 클 뿐이야. "비대칭이 더 심한지"는 전혀 알 수 없어.

---

## 5단계: 표준화 모멘트 (standardized moment)

해결책은 **k차 중심 모멘트를 σ의 k제곱으로 나누는 것**이야.

$$
\tilde\mu_k = \frac{E[(X-\mu)^k]}{\sigma^k} = E\left[\left(\frac{X-\mu}{\sigma}\right)^k\right]
$$

오른쪽 식을 보면 의미가 명확해져. **데이터를 먼저 z-score로 바꾼 다음 k차 모멘트를 구한 것**과 같아. X를 c배 해도 분자가 $c^k$배, 분모도 $c^k$배라서 **상쇄**돼. 단위도, 변동성 크기도 사라지고 **순수한 모양**만 남는 거야.

| 차수 | 표준화 모멘트 | 이름 | 정규분포일 때 |
|---|---|---|---|
| 1 | 항상 0 | — | 0 |
| 2 | 항상 1 | — | 1 |
| 3 | $E[(X-\mu)^3]/\sigma^3$ | **왜도 (skewness)** | 0 |
| 4 | $E[(X-\mu)^4]/\sigma^4$ | **첨도 (kurtosis)** | 3 |

1차와 2차는 표준화하면 무조건 0과 1이 되어서 정보가 없어. **그래서 표준화 모멘트가 의미를 갖는 건 3차부터야.**

첨도는 정규분포 기준값이 3이라, 3을 뺀 **초과첨도(excess kurtosis)**를 자주 써. 주식 수익률은 초과첨도가 양수, 즉 정규분포보다 극단값이 훨씬 자주 나오는 **fat tail**이 전형적이야.

## 6단계: 숫자로 확인

데이터 X = [1, 2, 3, 4, 10] (모집단 공식, n으로 나눔)

평균: $\mu = 20/5 = 4$
편차: $[-3, -2, -1, 0, 6]$

| | 계산 | 결과 |
|---|---|---|
| 2차 중심 | $(9+4+1+0+36)/5$ | $\sigma^2 = 10$, $\sigma \approx 3.162$ |
| 3차 중심 | $(-27-8-1+0+216)/5$ | $36$ |
| 4차 중심 | $(81+16+1+0+1296)/5$ | $278.8$ |
| **왜도** | $36 / 3.162^3 = 36/31.62$ | **≈ 1.14** |
| **첨도** | $278.8 / 10^2$ | **≈ 2.79** |

**왜도 +1.14:** 10이라는 값 하나가 오른쪽으로 튀어서 양의 비대칭이 생겼어. 3차에서 편차 6의 기여(216)가 나머지 전부(-36)를 압도하는 걸 봐. 이게 3단계에서 말한 "고차일수록 극단값에 지배된다"의 실례야.

이제 모든 값에 100을 곱해 [100, 200, 300, 400, 1000]으로 바꿔도 왜도는 똑같이 1.14야. 직접 해보면 표준화의 효과가 체감될 거야.

---

## 7단계: Realized 지표와 연결

이제 앞에서 본 공식들이 모멘트의 어느 자리에 해당하는지 보여.

| Realized 지표 | 수식 | 대응하는 모멘트 |
|---|---|---|
| Realized Variance | $\sum r_{t,i}^2$ | 2차 (원점) 모멘트의 합 |
| Realized Skewness | $\dfrac{\sqrt{N}\sum r_{t,i}^3}{RV_t^{3/2}}$ | **3차 표준화 모멘트** |
| Realized Kurtosis | $\dfrac{N\sum r_{t,i}^4}{RV_t^{2}}$ | **4차 표준화 모멘트** |

두 가지 의문이 풀릴 거야.

**왜 평균을 안 빼나(중심이 아니라 원점 모멘트를 쓰나)?** 10분봉 수익률의 평균은 사실상 0이라, $E[X^k]$와 $E[(X-\mu)^k]$가 거의 같아. 오히려 39개 표본으로 평균을 추정해서 빼면 추정 오차만 늘어.

**분모의 $RV^{3/2}$는?** 5단계의 $\sigma^3$이야. $RV = \sigma^2$이니 $RV^{3/2} = \sigma^3$이지. 즉 realized skewness는 **5단계 표준화 공식을 그대로 장중 데이터에 적용한 것**이고, $\sqrt{N}$은 합(Σ)을 평균(E)처럼 스케일을 맞춰주는 보정이야.

참고로 4차 모멘트 $\sum r_{t,i}^4$는 나중에 7번 **HARQ**에서 "realized quarticity"라는 이름으로 다시 나와. RV 자체가 얼마나 부정확하게 측정됐는지를 재는 데 쓰여.

---

**한 줄 요약:** 모멘트는 분포 모양을 차수별로 요약한 숫자야(1차 위치, 2차 퍼짐, 3차 비대칭, 4차 꼬리). 표준화 모멘트는 이를 σᵏ로 나눠 **단위와 변동성 크기의 영향을 제거한 순수한 모양 지표**야.

이제 realized skewness가 완전히 연결됐을 거야. 준비되면 2번 realized semivariance로 가자.
