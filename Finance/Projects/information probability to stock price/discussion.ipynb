## 1. Academic Consensus: From Strict EMH to Dynamic Equilibrium

The classical Efficient Market Hypothesis (EMH) formulated by Eugene Fama (1970) posited that asset prices fully reflect all available information within an information filtration set $\mathcal{F}_t$:

$$\mathbb{E}[P_{t+1} \mid \mathcal{F}_t] = (1 + \mu_t) P_t \implies \mathbb{E}[R_{t+1} - \mu_t \mid \mathcal{F}_t] = 0$$

where $\mu_t$ is the expected equilibrium rate of return commensurate with risk. Today, academic discourse has moved past the binary debate of "efficient vs. inefficient" through three formal pillars:

### The Grossman-Stiglitz Information Paradox (1980)

Sanford Grossman and Joseph Stiglitz demonstrated that **costless, complete informational efficiency is mathematically impossible**. If market prices fully and costlessly revealed all information:

* No rational economic agent would expend private resources (research, computing, alternative data) to acquire and process information ($C_{\text{analysis}} > 0$).
* If no agent expends resources to produce information, the market price cannot reflect it.

Therefore, the market settles into an **informationally efficient equilibrium** where mispricings exist in an amount precisely sufficient to compensate informed traders for their costs and risks:

$$\mathbb{E}[\alpha_{\text{informed}}] = C_{\text{analysis}} + \pi_{\text{risk}}$$

where $\pi_{\text{risk}}$ represents the compensation for bearing noise-trader and model risk.

### The Joint Hypothesis Problem

Whenever an empirical model identifies excess returns ($\alpha \neq 0$), academic finance cannot decisively state whether the market is inefficient or whether the underlying asset pricing model (e.g., CAPM, Fama-French 5-Factor) has omitted an unpriced systemic risk factor:

$$R_{i,t} - R_{f,t} = \alpha_i + \sum_{k=1}^K \beta_{i,k} F_{k,t} + \epsilon_{i,t}$$

Consequently, academic consensus accepts that anomalies such as momentum, value, and quality are partly risk premia and partly behavioral under/overreactions maintained by **limits to arbitrage** (capital constraints, short-sale costs, and noise-trader risk as formalized by Shleifer and Vishny, 1997).

### The Adaptive Markets Hypothesis (AMH)

Formalized by Andrew Lo (2004), modern mainstream theory often models efficiency not as a static state, but as an evolutionary ecology. Market efficiency varies across time, asset classes, and market regimes as a function of the number of competing participants, capital availability, and institutional adaptation.

## 2. Industry Consensus: Asymmetric Capacity and Alpha Decay

Quantitative hedge funds and institutional asset managers treat market efficiency through an operational lens:

* **Retail & Unconstrained Long-Only Equities:** Markets are **highly semi-strong efficient**. After accounting for management fees, bid-ask spreads, market impact, and borrowing costs, the net excess return for discretionary public-information trading is negative for over 85–90% of active mutual funds over a 10-year horizon.
* **Systematic & Quantitative Firms (Citadel, Renaissance Technologies, Millennium):** Markets are viewed as **locally inefficient**. Alpha is not an abundant, permanent premium, but a fleeting arbitrage opportunity resulting from structural market microstructure imbalances, cross-asset lead-lag effects, and institutional flow constraints.
* **Alpha Half-Life:** The rate at which an informational edge dissipates ($\tau_{1/2}$) has compressed significantly due to algorithmic competition, shifting alpha extraction from low-frequency fundamental metrics to high-dimensional, non-linear alternative datasets and execution efficiency.

## 3. Mathematical Evaluation of Your Hypotheses

### Point 1: Extracting Proprietary Insights from Public Information

Your intuition directly mirrors the mechanism that sustains the Grossman-Stiglitz equilibrium.

Let public information be a high-dimensional vector $\mathbf{X}_t \in \mathcal{F}_t^{\text{public}}$. The market consensus price reflects an aggregate pricing kernel mapping:

$$P_t = \bar{g}(\mathbf{X}_t)$$

An individual investor does not possess private data $\mathbf{X}_t$, but applies a non-consensus mapping function $g^*(\mathbf{X}_t)$ that models non-linear interactions, higher-order dependencies, or supply-chain propagation:

$$\hat{V}_t = g^*(\mathbf{X}_t)$$

Alpha is generated when the orthogonal component of your valuation estimate relative to the market expectation has positive covariance with the future true liquidation value $V_{t+1}$:

$$\alpha = \mathbb{E}\left[ (g^*(\mathbf{X}_t) - \bar{g}(\mathbf{X}_t)) \cdot (V_{t+1} - P_t) \right] > 0$$

In quantitative finance, this is known as generating **orthogonal alpha from public features** (e.g., natural language processing on corporate filings, supply-chain network centrality, or order-flow imbalances). Industry affirms that this is a valid and primary method for generating excess returns, provided the operational cost to build $g^*(\mathbf{X}_t)$ is lower than the gross alpha captured.

### Point 2: Scenario Probabilities, Momentum, and State-Contingent Pricing

Your framework conceptualizes the asset price through an **Arrow-Debreu state-space model**.

Consider future discrete states of the world $S = \{s_1, s_2, \dots, s_n\}$ occurring at time $t+1$, with objective physical probabilities $p(s_i)$ such that $\sum_{i=1}^n p(s_i) = 1$. The payoff of the asset in state $s_i$ is $X(s_i)$.

In a rational pricing framework, the current price is the risk-adjusted expectation:

$$P_t = \sum_{i=1}^n q(s_i) X(s_i) = \frac{1}{1 + r_f} \sum_{i=1}^n \mathbb{Q}(s_i) X(s_i)$$

where $q(s_i)$ is the state price and $\mathbb{Q}(s_i)$ is the risk-neutral probability assigned by the market.

Your hypothesis states that lower-probability events are "less reflected" in the price. In academic asset pricing, this maps directly to two documented phenomena:

1. **Rational Inattention & Bayesian Underreaction:** Because human and institutional attention is a finite resource, market participants do not continuously update priors on states where $p(s_i)$ is near zero. When an initial catalyst shifts the true probability from $p_t(s_l)$ to $p_{t+1}(s_l) > p_t(s_l)$, the market updates its subjective prior $\hat{p}_t$ with a lag (Bayesian conservatism):
$$\Delta \hat{p}_t < \Delta p_t$$


2. **The Mechanics of Fundamental Momentum:** Because the market updates probabilities incrementally rather than instantaneously, prices exhibit positive serial correlation:
$$\text{Cov}(\Delta P_t, \Delta P_{t+k}) > 0 \quad \text{for intermediate horizons } k \in [3, 12 \text{ months}]$$


As information diffuses and the once-low probability state becomes the consensus baseline, the price moves toward the terminal payoff $X(s_l)$. Momentum is precisely the gradual drift resulting from this delayed probability realization.

However, financial mathematics introduces an important caveat to your second point: **Skewness Preference (Lottery Effect)**. Behavioral finance (Barberis and Huang, 2008) proves that retail investors frequently *overprice* salient, highly visible low-probability tail events (such as out-of-the-money options or penny stock turnarounds) due to cumulative prospect theory probability weighting $w(p) > p$ for small $p$. Therefore, an event is underpriced only when it is low-probability and structurally neglected by institutional attention, rather than merely low-probability.

Modern finance validates your approach: semi-strong efficiency is high enough that raw public data contains no naive arbitrage, but incomplete enough that non-linear synthesis (Point 1) and tracking the dynamic diffusion of probabilities along neglected state paths (Point 2) constitute the primary structural sources of institutional alpha.

## 1. Theoretical Foundation: Probability Weighting & Overpricing

The behavioral mechanism where investors hyper-focus on extreme, low-probability outcomes is mathematically modeled by the probability weighting function $w(p)$ in Cumulative Prospect Theory (Tversky & Kahneman, 1992):

$$w(p) = \frac{p^\gamma}{\left( p^\gamma + (1-p)^\gamma \right)^{1/\gamma}}, \quad 0 < \gamma < 1$$

When an event has an objectively small probability ($p \to 0$), the curvature parameter $\gamma \approx 0.65$ guarantees that:

$$w(p) > p \quad \text{and} \quad \lim_{p \to 0} \frac{w(p)}{p} = \infty$$

### Asset Pricing Implications: Barberis and Huang (2008)

In their seminal paper *"Stocks as Lotteries: The Implications of Probability Weighting for Security Prices"* (American Economic Review), Nicholas Barberis and Ming Huang embedded this probability weighting function into a multi-asset equilibrium model.

Consider a speculative asset whose future terminal payoff distribution exhibits extreme positive idiosyncratic skewness:

* A low probability $p$ of an explosive payoff $X_{\text{high}}$ (e.g., clinical trial success, regulatory approval, sudden short squeeze).
* A high probability $1 - p$ of a baseline or zero payoff $X_{\text{low}}$.

Under rational expected utility theory with objective probabilities $\mathbb{P}$, the fair present value discounted at the required return $k$ is:

$$P_0^* = \frac{\mathbb{E}^\mathbb{P}[X]}{1 + k} = \frac{p X_{\text{high}} + (1 - p) X_{\text{low}}}{1 + k}$$

Under Cumulative Prospect Theory, investors evaluate the asset using decision weights $w(p)$ rather than linear probabilities $p$:

$$V_{\text{CPT}} = w(p) v(X_{\text{high}}) + (1 - w(p)) v(X_{\text{low}})$$

Because $w(p) \gg p$, investors' subjective expected payoff substantially exceeds the objective mathematical expectation:

$$\mathbb{E}^w[X] = w(p) X_{\text{high}} + (1 - w(p)) X_{\text{low}} \gg \mathbb{E}^\mathbb{P}[X]$$

### A Crucial Nuance: Required Return vs. Realized Expected Return

In your description, you noted that the *required return* increases; mathematically, the exact opposite occurs in market equilibrium:

1. Because investors assign disproportionate decision weight $w(p)$ to the explosive payoff, their aggregate demand bids up the equilibrium purchase price:
$$P_0 > P_0^*$$


2. Because the entry price $P_0$ is bid to an inflated level while the objective distribution of future cash flows $\mathbb{E}^\mathbb{P}[X]$ remains unchanged, the **objective expected rate of return** is mathematically compressed:
$$\mathbb{E}^\mathbb{P}[R_{t+1}] = \frac{\mathbb{E}^\mathbb{P}[P_{t+1}]}{P_0} - 1 < R_f$$



Investors in these assets exhibit a **willingness to accept a negative risk premium ($\alpha < 0$)** in exchange for exposure to positive skewness. They pay a "lottery ticket premium," resulting in persistent long-term underperformance.

## 2. Key Empirical Literature and Asset Pricing Factors

The academic literature has empirically documented this dynamic across thousands of securities:

### The MAX Effect: Bali, Cakici, and Whitelaw (2011)

In *"Maxing Out: Stocks as Lotteries and the Cross-Section of Expected Returns"* (Journal of Financial Economics), the authors defined an explicit empirical proxy for lottery-seeking behavior:

$$MAX_{i,t} = \max_{d \in t} \left( R_{i,d} \right)$$

* **Finding:** Stocks with the highest maximum daily returns over the prior month underperform stocks with low maximum daily returns by **1.0% to 1.4% per month in risk-adjusted alpha**.
* **Mechanism:** Investors observe a salient extreme return, overweight the probability that this event will repeat, and bid up the stock price, leading to immediate overvaluation and subsequent long-term price decay.

### Expected Idiosyncratic Skewness: Boyer, Mitton, and Vorkink (2010)

In *"Expected Idiosyncratic Skewness"* (Review of Financial Studies), the authors constructed a predictive model for conditional third moments:

$$\mathbb{E}_t \left[ \epsilon_{i,t+1}^3 \right] = f\left( \text{Skew}(\epsilon_{i,t}), \text{Var}(\epsilon_{i,t}), \text{Volume}_{i,t}, \text{Firm Characteristics} \right)$$

* **Finding:** There is a strong negative cross-sectional relationship between expected idiosyncratic skewness and future stock returns.
* Stocks in the highest expected skewness quintile underperform those in the lowest quintile by substantial margins, confirming that skewness is priced as a negative risk premium.

### Jackpot Stocks: Conrad, Kapadia, and Xing (2014)

In *"Death and Jackpot: Why Do Small Stocks Earn Low Average Returns?"* (Journal of Finance), the authors estimated the conditional objective probability of an extreme positive return ($R > 100\%$ over one year) vs. distress ($R < -75\%$):

* Stocks with high estimated "jackpot probability" earn negative excess returns, proving that the market systematically overpays for the right tail of small-cap equities.

### Investor Demographics: Kumar (2009)

In *"Who Gambles in the Stock Market?"* (Journal of Finance), Alok Kumar established the micro-foundations of this phenomenon:

* Retail investors with socioeconomic profiles matching state lottery ticket buyers disproportionately purchase stocks with low share price, high idiosyncratic volatility, and high idiosyncratic skewness.
* These lottery-like stocks underperform by approximately 2% to 3% annually compared to non-lottery stocks of identical size and book-to-market profiles.

## 3. Why the Mispricing Persists: Limits to Arbitrage

In a friction-free market, rational institutional arbitrageurs would identify this overpricing, short-sell the overvalued asset, and drive the price back to $P_0^*$. Modern finance explains why this mispricing remains uncorrected:

```
Extreme Skewness Preference (Retail) 
       ──> Overweighting w(p) >> p 
       ──> Inflated Demand & Price P₀ 
       ──> High Borrow Fees & Recall Risk (Limits to Arbitrage) 
       ──> Short Leg Blocked 
       ──> Persistent Negative Alpha

```

### Miller's Divergence of Opinion (1977)

Edward Miller demonstrated that when investors hold heterogeneous beliefs ($\sigma_{\text{belief}}^2 > 0$) about an asset with high uncertainty, and the market is subject to short-sale constraints ($c_{\text{borrow}} > 0$ or short sale bans):

$$P_0 = \mathbb{E}_{\text{optimists}}[V] > \mathbb{E}_{\text{all}}[V]$$

The market clearing price reflects only the valuations of the most optimistic participants (those overweighting the tail event), while pessimistic or rational participants are sidelined.

### Arbitrage Asymmetry: Stambaugh, Yu, and Yuan (2012, 2015)

In their studies on financial market anomalies, Stambaugh, Yu, and Yuan demonstrated that:

1. Anomalies related to speculative mispricing exist almost entirely in the **short leg** (the overvalued securities).
2. The borrowing fees ($c_{\text{borrow}}$), low available float, and catastrophic right-tail risk (short squeeze risk where $X_{\text{high}}$ actually occurs) prevent rational quantitative funds from shorting lottery stocks.
3. Therefore, the overpricing remains structurally protected from arbitrage forces.

## 4. Synthesis with Short-Term Momentum vs. Long-Term Reversal

Your framing of this dynamic as "momentum" can be decomposed into two distinct phases:

1. **Short-Term Attention Cascade ($\Delta t < 1 \text{ month}$):** When an initial positive shock occurs, retail investor attention is drawn to the high $MAX$ return. The subjective probability weighting accelerates as more participants enter, causing a temporary continuation of returns ($\text{Cov}(\Delta P_t, \Delta P_{t+1}) > 0$).
2. **Fundamental Reversion ($\Delta t \in [1, 12 \text{ months}]$):** Because the objective probability remains low ($p \ll 1$), the high payoff fails to materialize in the vast majority of cases ($X = X_{\text{low}}$). The inflated expectation collapses, resulting in severe negative abnormal returns ($\alpha < 0$).

Mainstream asset pricing confirms that retail participants do not miss low-probability events; they systematically overpay for them, creating an asset class characterized by inflated prices, low required risk premia by speculative buyers, and consistently negative long-term realized returns.
