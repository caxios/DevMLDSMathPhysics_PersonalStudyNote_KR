Qustion1 : 주가 종목의 수익률의 레짐을 파악하고, 각각의 레짐 속에서 10분봉의 변동성이 일일 주가 수익률의 상승과 하락에, 그리고 또 일일 주가 수익률의 변동성이 향후의 상승, 하락에 대한 어떤 시그널을 주는지 연구하는 프로젝트를 진행하고 싶은데.. 어떻게 발전시켜나갈 수 있지? 애초에 내가 하고자 하는 연구가 실제로 많이 연구되고 있는 거야? 그렇다면, 최신 연구 동향에 대해서 알려주고, 그렇지 않다면 주가 수익률의 변동성에 대해 최근에 어떤 연구가 이뤄지고 있는지 알려줘봐.

---

짧게 답하면, 네가 하려는 연구의 **구성요소는 매우 많이 연구된 분야**야. 다만 "레짐별로, 장중 변동성이 다음 날 수익률의 **방향**을 예측하는가"라는 조합은 표준적인 주제가 아니야. 새로워서라기보다는 **변동성은 변동성을 잘 예측하지만 방향은 거의 예측하지 못하기 때문**이야. 지금 구상 그대로 진행하면 "유의미한 결과 없음"으로 끝날 가능성이 높아. 그래서 문제 설정부터 손봐야 해.

## 1. 지금 구상의 문제점

**동시성 문제.** "10분봉 변동성이 일일 수익률의 상승·하락에 주는 시그널"이 같은 날(t)의 관계라면 예측이 아니야. 그냥 레버리지 효과(하락일에 변동성이 커짐)를 다시 확인하는 셈이고, 이건 1976년 Black 이후 누구나 아는 사실이야. 반드시 t일의 장중 측정치로 t+1일(또는 t+1~t+5일) 수익률을 예측하는 구조여야 해.

**레짐 순환성.** 레짐을 수익률로 정의한 뒤 그 레짐 안에서 수익률 방향을 예측하면 순환 논리가 돼. 레짐을 정의하는 변수와 예측 변수는 분리해야 해.

**Look-ahead bias.** HMM의 smoothed probability(전체 표본을 보고 추정한 레짐)를 쓰면 백테스트 결과가 크게 부풀려져. 이 분야에서 가장 흔한 실수야. 반드시 filtered 또는 online 추정만 써야 해.

**표본 부족.** 단일 종목 시계열을 레짐 수 × 상승/하락으로 쪼개면 관측치가 금방 바닥나. 종목 패널이나 횡단면으로 가야 검정력이 생겨.

**측정 노이즈.** 한국장 정규장을 10분봉으로 나누면 하루 약 39개 구간이야. 이 정도로 계산한 실현변동성은 노이즈가 커. 동시호가와 오버나이트 수익률은 따로 분리해야 해.

## 2. 실제 연구 계보: 네 질문과 가장 가까운 것부터

**(1) 장중 수익률의 "모양"이 미래 수익률을 예측하는가.** 네 아이디어에 가장 가까운 주제야. 핵심은 변동성의 **크기**가 아니라 **부호 분해**라는 점이야.
- Amaya, Christoffersen, Jacobs, Vasquez (2015, JFE): 장중 수익률로 계산한 realized skewness가 다음 주 횡단면 수익률을 음(−)으로 예측한다.
- Barndorff-Nielsen, Kinnebrock, Shephard (2010)의 realized semivariance(상방/하방 변동성 분해)
- Patton & Sheppard (2015) "Good volatility, bad volatility"
- Bollerslev, Li, Zhao (2020, JFQA): signed jump variation으로 횡단면 수익률을 예측한다.
- Bollerslev, Patton, Quaedvlieg (2022, JFE): realized semibeta

**(2) 변동성이 어떻게 방향 예측력을 만드는가.** Christoffersen & Diebold (2006)는 평균 수익률을 예측할 수 없어도, 기대수익률이 0이 아니고 변동성이 예측 가능하면 **부호(방향)가 예측 가능해진다**는 걸 보였어. 네 연구의 이론적 근거로 반드시 읽어야 할 논문이야.

**(3) 변동성과 미래 수익률 (일별·월별).** 레버리지/변동성 피드백 가설(Campbell & Hentschel 1992), 고유변동성 퍼즐(Ang, Hodrick, Xing, Zhang 2006), volatility-managed portfolio(Moreira & Muir 2017)와 이에 대한 반박(Cederburg et al. 2020)이 여기에 속해. "일일 수익률의 변동성이 향후 등락에 주는 시그널"은 사실상 이 영역이야.

**(4) 실현변동성 예측 자체.** 측정의 기초는 Andersen, Bollerslev, Diebold, Labys(ABDL)와 Barndorff-Nielsen & Shephard의 실현변동성 이론이고, 여기서 HAR(Corsi 2009)과 HARQ(측정오차 보정)가 표준 벤치마크로 자리 잡았어. 최근에는 ML이 많이 들어오고 있어. 예를 들어 Betz et al.(2023)은 과거 일·주·월 단위로 집계한 장중 수익률을 이미지처럼 구성해 CNN을 학습시켰고, 다우존스 28개 종목의 분 단위 데이터에서 HAR 벤치마크보다 표본 내외 RMSE가 모두 낮았다.

**(5) 레짐 식별의 최신 흐름.** HMM과 Markov-switching에서 **statistical jump model(JM)** 쪽으로 무게가 옮겨가고 있어. HMM은 레짐이 불균형하고 지속성이 강하며 신호 대 잡음비가 낮고 데이터가 제한적인 환경에서 지속성과 안정성이 부족한 상태열을 만들어 잦은 오경보를 낸다는 한계가 지적돼 왔어. JM은 각 상태 전환에 jump penalty를 부과해 레짐 지속성을 높이고, 수익률 시계열에서만 뽑은 위험·수익 피처를 쓰며, 전략 성과를 직접 최적화하는 시계열 교차검증으로 penalty를 선택해(Shu, Yu, Mulvey 2024). 변동성 예측에 레짐을 결합한 최근 연구로는 2014~2025년 SPX의 5분 수익률로 계산한 실현변동성에 soft Markov switching, XGBoost 기반 분포적 스펙트럴 클러스터링 등 여러 레짐 전환 방법을 적용한 연구가 있어.

**정신이 번쩍 들 만한 반례.** 2026년의 한 논문은 장중 변동성·거래량·갭으로 레짐 분류기를 만들었어. 분류기는 기술적(descriptive)으로는 유효했지만, 방향성 매매 신호로 바꾸는 데는 실패했어. 가장 좋았던 설정도 연도별 안정성 검정에서 무너져 2024년에는 순손실을 냈고, 저자는 결국 분류기를 보조적 맥락 필터나 변동성 레짐 지표로만 보존했어. 네 프로젝트가 가장 빠지기 쉬운 결말이 바로 이거야.

## 3. 발전시키는 방법

**가설을 검증 가능한 형태로 다시 써.**
- H1: t일의 realized semivariance와 realized skewness가 t+1~t+5일 수익률의 횡단면 순위를 예측한다.
- H2: 그 예측력은 시장 레짐(고변동/저변동, 강세/약세)에 따라 부호나 크기가 달라진다.
- H3: 일별 변동성이 다음 날 방향 예측에 기여하는 부분은 Christoffersen-Diebold 채널, 즉 변동성 대비 기대수익 비율로 설명된다.

**레짐은 종목이 아니라 시장 수준에서 정의해.** KOSPI 지수에 JM이나 HMM을 online으로 적용하면 순환성 문제와 표본 부족 문제가 함께 해결돼. 종목별 레짐은 나중에 확장으로 두면 돼.

**분석 순서는 계량이 먼저, ML은 나중이야.**
1. KOSPI200 또는 유동성 상위 종목의 10분봉 패널을 구축한다.
2. RV, RS⁺/RS⁻, RSkew, 점프 성분, 오버나이트 수익률을 분리해 계산한다.
3. Fama-MacBeth 회귀와 decile long-short 포트폴리오를 만든다.
4. 레짐 더미와의 상호작용항으로 H2를 검정한다.
5. 마지막으로 XGBoost로 비선형성을 확인한다.

처음부터 ML로 가면 "왜 되는지"를 설명할 수 없어서 연구로서의 가치가 떨어져.

**평가 지표를 바꿔.** 방향 적중률 52% 같은 숫자는 의미가 없어. Rank IC, 거래비용을 차감한 long-short 샤프, 연도별 안정성, 레짐 조건부 모형과 무조건부 모형 간 Diebold-Mariano 검정으로 평가해야 해.

**차별화 포인트는 한국 시장에 있어.** 미국 데이터로는 이미 다 해본 주제라서 한국만의 구조가 네 무기야. 가격제한폭(상하한가 근처에서 실현변동성이 절단되는 문제), 높은 개인투자자 비중, 그리고 특히 **2023년 11월~2025년 3월 공매도 전면 금지 기간**을 외생적 레짐으로 쓰면 좋아. 공매도가 막혔을 때 하방 semivariance의 예측력이 달라지는지 보는 건 통계적 레짐보다 훨씬 설득력 있는 자연실험이야.

솔직히 말하면, 원래 구상("레짐별 변동성 → 방향")은 포트폴리오 프로젝트로는 흔하고 결과가 약하게 나올 공산이 커. 반면 "realized semivariance/skewness의 횡단면 예측력이 한국 시장의 제도적 레짐(공매도 금지)에서 어떻게 달라지는가"로 좁히면, 방법론은 검증된 것을 쓰면서도 기여점이 분명한 연구가 돼.
