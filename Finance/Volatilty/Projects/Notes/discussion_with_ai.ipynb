(26.9.18) Qustion : 주가 종목의 수익률의 레짐을 파악하고, 각각의 레짐 속에서 10분봉의 변동성이 일일 주가 수익률의 상승과 하락에, 그리고 또 일일 주가 수익률의 변동성이 향후의 상승, 하락에 대한 어떤 시그널을 주는지 연구하는 프로젝트를 진행하고 싶은데.. 어떻게 발전시켜나갈 수 있지? 애초에 내가 하고자 하는 연구가 실제로 많이 연구되고 있는 거야? 그렇다면, 최신 연구 동향에 대해서 알려주고, 그렇지 않다면 주가 수익률의 변동성에 대해 최근에 어떤 연구가 이뤄지고 있는지 알려줘봐.

짧게 답하면, 네가 하려는 연구의 **구성요소는 매우 많이 연구된 분야**야. 다만 "레짐별로, 장중 변동성이 다음 날 수익률의 **방향**을 예측하는가"라는 조합은 표준적인 주제가 아니야. 새로워서라기보다는 **변동성은 변동성을 잘 예측하지만 방향은 거의 예측하지 못하기 때문**이야. 지금 구상 그대로 진행하면 "유의미한 결과 없음"으로 끝날 가능성이 높아. 그래서 문제 설정부터 손봐야 해.

## 1. 지금 구상의 문제점

**동시성 문제.** "10분봉 변동성이 일일 수익률의 상승·하락에 주는 시그널"이 같은 날(t)의 관계라면 예측이 아니야. 그냥 레버리지 효과(하락일에 변동성이 커짐)를 다시 확인하는 셈이고, 이건 1976년 Black 이후 누구나 아는 사실이야. 반드시 t일의 장중 측정치로 t+1일(또는 t+1~t+5일) 수익률을 예측하는 구조여야 해.

**레짐 순환성.** 레짐을 수익률로 정의한 뒤 그 레짐 안에서 수익률 방향을 예측하면 순환 논리가 돼. 레짐을 정의하는 변수와 예측 변수는 분리해야 해.

**Look-ahead bias.** HMM의 smoothed probability(전체 표본을 보고 추정한 레짐)를 쓰면 백테스트 결과가 크게 부풀려져. 이 분야에서 가장 흔한 실수야. 반드시 filtered 또는 online 추정만 써야 해.

**표본 부족.** 단일 종목 시계열을 레짐 수 × 상승/하락으로 쪼개면 관측치가 금방 바닥나. 종목 패널이나 횡단면으로 가야 검정력이 생겨.

**측정 노이즈.** 한국장 정규장을 10분봉으로 나누면 하루 약 39개 구간이야. 이 정도로 계산한 실현변동성은 노이즈가 커. 동시호가와 오버나이트 수익률은 따로 분리해야 해.

## 2. 실제 연구 계보: 네 질문과 가장 가까운 것부터

**(1) 장중 수익률의 "모양"이 미래 수익률을 예측하는가.** 네 아이디어에 가장 가까운 주제야. 핵심은 변동성의 **크기**가 아니라 **부호 분해**라는 점이야.
- Amaya, Christoffersen, Jacobs, Vasquez (2015, JFE): 장중 수익률로 계산한 realized skewness가 다음 주 횡단면 수익률을 음(−)으로 예측한다.
- Barndorff-Nielsen, Kinnebrock, Shephard (2010)의 realized semivariance(상방/하방 변동성 분해)
- Patton & Sheppard (2015) "Good volatility, bad volatility"
- Bollerslev, Li, Zhao (2020, JFQA): signed jump variation으로 횡단면 수익률을 예측한다.
- Bollerslev, Patton, Quaedvlieg (2022, JFE): realized semibeta

**(2) 변동성이 어떻게 방향 예측력을 만드는가.** Christoffersen & Diebold (2006)는 평균 수익률을 예측할 수 없어도, 기대수익률이 0이 아니고 변동성이 예측 가능하면 **부호(방향)가 예측 가능해진다**는 걸 보였어. 네 연구의 이론적 근거로 반드시 읽어야 할 논문이야.

**(3) 변동성과 미래 수익률 (일별·월별).** 레버리지/변동성 피드백 가설(Campbell & Hentschel 1992), 고유변동성 퍼즐(Ang, Hodrick, Xing, Zhang 2006), volatility-managed portfolio(Moreira & Muir 2017)와 이에 대한 반박(Cederburg et al. 2020)이 여기에 속해. "일일 수익률의 변동성이 향후 등락에 주는 시그널"은 사실상 이 영역이야.

**(4) 실현변동성 예측 자체.** 측정의 기초는 Andersen, Bollerslev, Diebold, Labys(ABDL)와 Barndorff-Nielsen & Shephard의 실현변동성 이론이고, 여기서 HAR(Corsi 2009)과 HARQ(측정오차 보정)가 표준 벤치마크로 자리 잡았어. 최근에는 ML이 많이 들어오고 있어. 예를 들어 Betz et al.(2023)은 과거 일·주·월 단위로 집계한 장중 수익률을 이미지처럼 구성해 CNN을 학습시켰고, 다우존스 28개 종목의 분 단위 데이터에서 HAR 벤치마크보다 표본 내외 RMSE가 모두 낮았다.

**(5) 레짐 식별의 최신 흐름.** HMM과 Markov-switching에서 **statistical jump model(JM)** 쪽으로 무게가 옮겨가고 있어. HMM은 레짐이 불균형하고 지속성이 강하며 신호 대 잡음비가 낮고 데이터가 제한적인 환경에서 지속성과 안정성이 부족한 상태열을 만들어 잦은 오경보를 낸다는 한계가 지적돼 왔어. JM은 각 상태 전환에 jump penalty를 부과해 레짐 지속성을 높이고, 수익률 시계열에서만 뽑은 위험·수익 피처를 쓰며, 전략 성과를 직접 최적화하는 시계열 교차검증으로 penalty를 선택해(Shu, Yu, Mulvey 2024). 변동성 예측에 레짐을 결합한 최근 연구로는 2014~2025년 SPX의 5분 수익률로 계산한 실현변동성에 soft Markov switching, XGBoost 기반 분포적 스펙트럴 클러스터링 등 여러 레짐 전환 방법을 적용한 연구가 있어.

**정신이 번쩍 들 만한 반례.** 2026년의 한 논문은 장중 변동성·거래량·갭으로 레짐 분류기를 만들었어. 분류기는 기술적(descriptive)으로는 유효했지만, 방향성 매매 신호로 바꾸는 데는 실패했어. 가장 좋았던 설정도 연도별 안정성 검정에서 무너져 2024년에는 순손실을 냈고, 저자는 결국 분류기를 보조적 맥락 필터나 변동성 레짐 지표로만 보존했어. 네 프로젝트가 가장 빠지기 쉬운 결말이 바로 이거야.

## 3. 발전시키는 방법

**가설을 검증 가능한 형태로 다시 써.**
- H1: t일의 realized semivariance와 realized skewness가 t+1~t+5일 수익률의 횡단면 순위를 예측한다.
- H2: 그 예측력은 시장 레짐(고변동/저변동, 강세/약세)에 따라 부호나 크기가 달라진다.
- H3: 일별 변동성이 다음 날 방향 예측에 기여하는 부분은 Christoffersen-Diebold 채널, 즉 변동성 대비 기대수익 비율로 설명된다.

**레짐은 종목이 아니라 시장 수준에서 정의해.** KOSPI 지수에 JM이나 HMM을 online으로 적용하면 순환성 문제와 표본 부족 문제가 함께 해결돼. 종목별 레짐은 나중에 확장으로 두면 돼.

**분석 순서는 계량이 먼저, ML은 나중이야.**
1. KOSPI200 또는 유동성 상위 종목의 10분봉 패널을 구축한다.
2. RV, RS⁺/RS⁻, RSkew, 점프 성분, 오버나이트 수익률을 분리해 계산한다.
3. Fama-MacBeth 회귀와 decile long-short 포트폴리오를 만든다.
4. 레짐 더미와의 상호작용항으로 H2를 검정한다.
5. 마지막으로 XGBoost로 비선형성을 확인한다.

처음부터 ML로 가면 "왜 되는지"를 설명할 수 없어서 연구로서의 가치가 떨어져.

**평가 지표를 바꿔.** 방향 적중률 52% 같은 숫자는 의미가 없어. Rank IC, 거래비용을 차감한 long-short 샤프, 연도별 안정성, 레짐 조건부 모형과 무조건부 모형 간 Diebold-Mariano 검정으로 평가해야 해.

**차별화 포인트는 한국 시장에 있어.** 미국 데이터로는 이미 다 해본 주제라서 한국만의 구조가 네 무기야. 가격제한폭(상하한가 근처에서 실현변동성이 절단되는 문제), 높은 개인투자자 비중, 그리고 특히 **2023년 11월~2025년 3월 공매도 전면 금지 기간**을 외생적 레짐으로 쓰면 좋아. 공매도가 막혔을 때 하방 semivariance의 예측력이 달라지는지 보는 건 통계적 레짐보다 훨씬 설득력 있는 자연실험이야.

솔직히 말하면, 원래 구상("레짐별 변동성 → 방향")은 포트폴리오 프로젝트로는 흔하고 결과가 약하게 나올 공산이 커. 반면 "realized semivariance/skewness의 횡단면 예측력이 한국 시장의 제도적 레짐(공매도 금지)에서 어떻게 달라지는가"로 좁히면, 방법론은 검증된 것을 쓰면서도 기여점이 분명한 연구가 돼.

## 0. 바뀐 조건이 설계를 어떻게 바꾸나

### yfinance의 현실

yfinance는 비공식 라이브러리라 정책이 바뀔 수 있지만, 일반적으로 알려진 조회 한도는 이래.

| 봉 간격 | 조회 가능한 과거 기간 |
|---|---|
| 1분 | 최근 약 30일 (한 번에 7일씩) |
| 2·5·15·30분 | **최근 약 60일** |
| 60분 | **최근 약 730일 (2년)** |
| 일봉 | 전체 역사 |

그리고 **10분봉은 기본 제공 간격이 아니야.** 5분봉을 받아 합쳐야 하는데, 5분봉은 60일치밖에 없어.

**직설적으로 말하면, 이전 계획처럼 "10분봉 10년치"로 하는 연구는 yfinance로는 불가능해.** 그래서 설계를 바꿔야 해.

### 해결책: 세 가지 해상도로 나눠서 설계

핵심 아이디어는 이거야. **realized 지표는 "기간을 잘게 쪼갠 수익률"로 만드는 거지, 반드시 10분봉일 필요는 없어.** 월간 RSJ는 그달의 일별 수익률 21개로, 주간 RSJ는 그 주의 60분봉 35개로도 만들 수 있어.

| 층 | 데이터 | 기간 | 만들 수 있는 지표 | 봉 개수 N | 측정 오차 $\sqrt{2/N}$ |
|---|---|---|---|---|---|
| **A층** | 일봉 | 15년 이상 | **월간** RS±, RSJ, RSkew / 일별 레인지 변동성 | 약 21 | 약 31% |
| **B층** | 60분봉 | 2년 | **주간** RS±, RSJ, RSkew | 약 35 (하루 7개 × 5일) | 약 24% |
| **C층** | 5분봉 **직접 누적 수집** | 지금부터 쌓임 | **일간** RV, RS±, RSJ | 약 78 | 약 16% |

(측정 오차는 7번 HARQ에서 유도한 공식이야. 참고로 원래 계획의 10분봉 일간 지표는 N = 39로 약 23%였어. B층 주간 지표가 그것과 비슷한 정밀도야.)

**각 층의 역할:**
- **A층 (주력):** 긴 역사가 있어서 레짐 에피소드가 많고(2008, 2011, 2015~16, 2018년 4분기, 2020, 2022, 2025년 관세 충격), 공매도 금지 전후 비교가 가능해. **본 분석과 백테스트의 중심**이야.
- **B층 (검증):** 더 정밀한 주간 지표로 A층 결과가 해상도를 높여도 유지되는지 확인해.
- **C층 (미래 자산):** 오늘부터 매주 5분봉을 받아 저장하면, 프로젝트가 끝날 때쯤 몇 달치, 1년 뒤면 1년치 자체 데이터베이스가 생겨. **지금 당장 시작해야 하는 유일한 작업**이야.

이 구조는 약점이면서 동시에 기여점이 될 수 있어. **"측정 해상도(월간·주간·일간)에 따라 시그널이 얼마나 살아남는가"**는 그 자체로 흥미로운 질문이고, 무료 데이터로 전략을 만들려는 사람에게 실용적인 답이야.

### yfinance의 또 다른 함정: 생존편향

yfinance는 대부분 **현재 상장된 종목**만 조회돼. 상장폐지된 종목은 빠져 있어. 그래서 "현재 S&P 500 종목"으로 과거를 백테스트하면, **살아남은 승자들만으로** 과거를 평가하는 셈이라 성과가 부풀려져.

대응 방법은 이래.
- **시점별 구성종목 목록**을 무료로 구해서 쓴다 (미국: 공개된 S&P 500 과거 구성종목 데이터셋, 한국: pykrx의 지수 구성종목 조회).
- 그 목록 중 yfinance에 데이터가 없는 종목이 몇 개인지 **세어서 보고**한다.
- 전략 성과는 **같은 universe로 만든 동일가중 벤치마크 대비**로 평가한다. 편향이 양쪽에 비슷하게 들어가서 상대 성과는 덜 왜곡돼 (완전히 사라지진 않아).
- 대형주 universe로 한정한다. 대형주는 상장폐지가 드물어서 편향이 작아.

### 무료 보조 데이터

가격은 yfinance로 하되, 가격이 아닌 데이터는 다른 무료 소스가 필요해.

| 데이터 | 무료 소스 | 용도 |
|---|---|---|
| 한국 시가총액, 지수 구성종목, 공매도 잔고 | **pykrx** (KRX 정보데이터시스템 기반) | 규모 통제, 생존편향 완화, H4 |
| 미국 팩터 (시장, 규모, 가치, 반전, 모멘텀) | Kenneth French 데이터 라이브러리 | 팩터 알파 |
| 미국 공매도 잔고 | FINRA 공개 데이터 | H4 보조 |
| 무위험 금리 | FRED (미국), 한국은행 ECOS (한국) | 샤프 비율 |
| VIX | yfinance (^VIX) | 미국 강건성 레짐 |

pykrx는 KRX 사이트 구조에 의존해서 작동이 안 될 때가 있으니, **가장 먼저 작동 여부를 테스트**해봐.

## 1. 수정된 연구 질문과 전략 목표

**연구 질문:**

> 기간 내 수익률의 방향 비대칭(realized semivariance, RSJ)은 미래 변동성과 횡단면 수익률을 예측하는가? 그 예측력은 시장 변동성 레짐과 공매도 제약에 따라 달라지며, 한국과 미국에서 어떻게 다른가?

> 위 예측력을 거래비용, 세금, 공매도 제약 등 **개인투자자가 실제로 마주하는 조건**에서 실행 가능한 전략으로 바꿨을 때, 단순 벤치마크 대비 위험조정 성과를 개선하는가?

## 3. 측정 설계: 층별 변수

| 변수 | A층 (일봉, 월간) | B층 (60분봉, 주간) | C층 (5분봉, 일간) |
|---|---|---|---|
| RV | 월 내 일별 $r^2$ 합 | 주 내 시간별 $r^2$ 합 | 일 내 5분 $r^2$ 합 |
| RS±, RSJ | 월 내 일별 수익률로 | 주 내 시간별 수익률로 | 일 내 5분 수익률로 |
| RSkew | 월 내 일별 수익률로 | 주 내 시간별 수익률로 | 일 내 5분 수익률로 |
| RQ | 월 내 일별 $r^4$로 | 주 내 시간별 $r^4$로 | 일 내 5분 $r^4$로 |
| IVOL, MAX | 월 내 일별 CAPM 잔차 / 최대 일수익률 | — | — |

**A층을 위한 추가 도구: 레인지 기반 일별 변동성**

일봉에는 시가, 고가, 저가, 종가가 있어서, 종가만 쓰는 것보다 훨씬 정밀하게 **그날의** 변동성을 추정할 수 있어. 대표적인 게 이 두 가지야.

$$
\sigma^2_{\text{Parkinson}} = \frac{1}{4\ln 2}\left(\ln\frac{H}{L}\right)^2
$$

$$
\sigma^2_{\text{Garman-Klass}} = \frac{1}{2}\left(\ln\frac{H}{L}\right)^2 - (2\ln 2 - 1)\left(\ln\frac{C}{O}\right)^2
$$

고가와 저가의 차이는 하루 동안 가격이 얼마나 넓게 움직였는지를 담고 있어서, 종가 하나보다 훨씬 많은 정보가 있어. 이걸로 **15년치 일별 변동성 시계열**을 만들 수 있고, H1의 HAR 분석에 쓸 수 있어.

**반분산의 무료 대체물: LHAR**

일봉에는 장중 부호 정보가 없으니 일별 RS±를 직접 만들 수는 없어. 대신 Corsi & Renò(2012)의 **레버리지 HAR(LHAR)**을 써. 과거 **음의 수익률** 크기를 별도 변수로 넣는 방식이야.

$$
\sigma^2_{t+1} = \beta_0 + \beta_d\sigma^2_t + \beta_w\sigma^{2(w)}_t + \beta_m\sigma^{2(m)}_t + \gamma_d\,r^-_t + \gamma_w\,r^{-(w)}_t + \gamma_m\,r^{-(m)}_t + \varepsilon_{t+1}
$$

$r^-_t = \min(r_t, 0)$이야. "하락이 미래 변동성을 더 키우는가"라는 semivariance와 같은 질문을 일봉만으로 검정하는 방법이야. 2년치 B층과 C층으로는 진짜 RS±를 쓴 SHAR로 같은 결론이 나오는지 확인하면 돼.

## 4. 레짐 정의 (사전 확정)

한국과 미국에 **같은 규칙**을 적용해야 비교가 공정해.

| 역할 | 레짐 | 정의 |
|---|---|---|
| **주 1** | 시장 변동성 레짐 | 각국 지수(^GSPC, ^KS11) 일별 수익률로 만든 피처(여러 반감기의 EWMA 수익률, 하방편차, 소르티노 비율)로 **JM, K = 2**, 확장 윈도우, 월별 재추정, 전진 계산(online). λ는 중첩 워크포워드로 선택 |
| **주 2** | 제도적 레짐 | 한국 공매도 전면 금지 기간 더미. **미국은 같은 기간에 금지가 없었으므로 대조군** |
| 강건성 1 | HMM 레짐 | 지수 수익률 + 로그 레인지 변동성, K = 2, online 필터링 |
| 강건성 2 | 관측 레짐 | 미국 VIX, 한국은 지수 레인지 변동성의 확장 윈도우 상위 20% |

A층 데이터가 15년 이상이라 **이제 H3(레짐 조건부)를 제대로 검정할 수 있어.** 이전 계획에서는 10분봉 기간이 짧을까 봐 걱정했던 부분이야.

## 5. 가설

### H1: 변동성 예측 (연구의 기반)

- **A층:** LHAR의 $\gamma$(음의 수익률 계수)가 유의한가, 레짐 상호작용이 있는가 (한국·미국, 15년 이상)
- **B·C층:** SHAR에서 $\beta^- > \beta^+$가 재현되는가 (2년 / 누적분)
- **표본 외 비교:** HAR vs LHAR vs HARQ, QLIKE 기준

**이게 전략의 위험관리 엔진이 돼** (6단계 S1). 변동성을 잘 예측하면 포지션 크기를 조절할 수 있으니까.

### H2: 횡단면 수익률 예측

- **A층:** 월간 RSJ, RSkew가 다음 달 수익률을 음(−)으로 예측하는가 (Fama-MacBeth, Rank IC, 분위 포트폴리오)
- **B층:** 주간 RSJ가 다음 주 수익률을 예측하는가
- 통제: 단기 반전, IVOL, MAX, 유동성(거래대금), 한국은 pykrx 시가총액

### H3: 시장 레짐 조건부

$$
IC_t = \theta_0 + \theta_1 D_t + u_t
$$

A층의 긴 역사로 검정해. $D_t$는 online JM 레짐이야.

### H4: 공매도 제약 — 국가 간 이중차분

미국을 대조군으로 쓸 수 있게 되면서 설계가 훨씬 강해졌어.

$$
b_{c,t} = \theta_0 + \theta_1\,Korea_c + \theta_2\,BanPeriod_t + \theta_3\,(Korea_c \times BanPeriod_t) + \theta_4\,D_{c,t} + u_{c,t}
$$

- $b_{c,t}$: 국가 c, t월의 RSJ Fama-MacBeth 기울기 (또는 IC)
- **$\theta_3$**: "공매도 금지 기간에 **한국에서만** 시그널이 달라졌는가"

같은 기간 두 시장에 공통으로 영향을 준 사건(글로벌 금리, 2024년 8월 급락, 2025년 관세 충격)은 미국 대조군이 상쇄해줘. 한국 고유 사건(밸류업 프로그램, 2024년 12월 계엄 사태 급락)은 여전히 교란 요인이니 $D_{c,t}$로 통제하고, **플라시보 검정**(가짜 금지 기간)도 해.

한국 내에서는 pykrx 공매도 잔고로 **금지 직전 공매도가 많았던 종목 vs 적었던 종목**의 이중차분도 추가해.

### H5: 한국과 미국의 차이

- 한국은 개인투자자 비중이 높아서, 복권 선호 계열 시그널(RSJ, RSkew, MAX, IVOL)의 효과가 **더 강할 거라는 가설**이 가능해.
- 반대로 가격제한폭과 VI 때문에 극단적 움직임이 잘려서, 측정 자체가 약해질 수도 있어.

어느 쪽이 맞는지는 데이터가 말해줄 거야. 국가 더미와 상호작용항으로 검정해.

## 6. 전략 설계 — 실제 투자로 연결

학술 롱숏은 개인에게 실현이 어려워(18번 6단계). 그래서 **개인이 실제로 할 수 있는 형태**로 세 가지 전략을 설계해. 각 전략은 앞의 가설 하나씩과 연결돼.

### S1: 레짐 기반 위험 조절 (H1, H3 기반)

**가장 현실적이고 결과가 나올 가능성이 높은 전략이야.** Shu-Yu-Mulvey(2024)의 0/1 전략을 기반으로 해.

- **자산:** 지수 ETF (미국 SPY, 한국 KODEX 200 같은 KOSPI200 추종 ETF)
- **규칙 A (레짐 스위치):** online JM이 평온 레짐이면 100% 보유, 혼란 레짐이면 현금(또는 단기채 ETF)
- **규칙 B (변동성 타기팅):** 6번에서 건너뛴 volatility-managed portfolio의 핵심 아이디어야. 한 줄로 말하면 **예측 변동성이 높을 때 비중을 줄이고 낮을 때 늘리는 것**이야.

$$
w_{t} = \min\left(\frac{\sigma_{\text{목표}}}{\hat\sigma_{t+1}},\ w_{\max}\right)
$$

$\hat\sigma_{t+1}$은 H1의 LHAR이나 HARQ 예측값이야. 레버리지를 쓰지 않도록 $w_{\max} = 1$로 제한해.

- **규칙 C:** A와 B의 결합
- **벤치마크:** 같은 ETF 단순 보유

### S2: 종목 선택 (H2 기반)

- **Universe:** S&P 500, KOSPI200 (대형주, 유동성 확보) 또는 섹터 ETF
- **시그널:** 월간 RSJ + RSkew + MAX + IVOL을 결합한 점수. 각 시그널을 매월 횡단면 순위로 바꿔서 평균 내는 게 가장 단순하고 견고해.
- **구성:** **롱온리**. 점수 상위 20% 종목을 동일가중으로 보유. 롱숏 수익은 연구 결과로만 보고하고, 실전은 롱온리로 해.
- **리밸런싱:** 월 1회. 회전율을 줄이기 위해 **버퍼 규칙**을 둬. 예를 들어 보유 종목이 상위 30% 안에 있으면 계속 보유하고, 벗어날 때만 교체해.
- **벤치마크:** 같은 universe 동일가중

### S3: 결합 전략 (S1 + S2)

- 종목 선택은 S2로, 전체 주식 비중은 S1으로 결정해.
- 레짐에 따라 **시그널 가중치를 바꾸는** 버전도 시험해봐. H3에서 "혼란 레짐에서 RSJ가 더 강하다"가 나오면, 혼란 레짐에서는 RSJ 비중을 높이는 식이야. 단, 이건 H3 결과를 보고 설계하는 거라 12번의 선택 편향이 생겨. **마지막 테스트 구간에서만** 평가해야 해.

### 실행 규칙 (모든 전략 공통)

**신호와 체결 시점:** t일 종가까지의 데이터로 신호를 계산하고, **t+1일 시가**(yfinance에 있음)나 **t+1일 종가**로 체결한다고 가정해. 두 가정을 모두 돌려서 결과 차이를 봐. 차이가 크면 시그널이 체결 타이밍에 예민하다는 경고야.

**거래비용:**

| 항목 | 한국 | 미국 |
|---|---|---|
| 매도 세금 | 증권거래세 (연도별로 세율이 바뀌어 왔으니 기간별 적용) | 없음 (소액의 SEC 수수료 등) |
| 수수료 | 증권사별 (비대면 계좌는 매우 낮음) | 대부분 0에 가까움 |
| 스프레드 + 시장충격 | 대형주 기준 편도 약 0.05~0.1%로 가정하고 민감도 분석 | 대형주 기준 편도 약 0.02~0.05% |
| 환전 | — | **한국에서 투자 시 환전 수수료** |

**세금:** 한국 거주자가 미국 주식으로 번 매매차익은 연간 기본공제(250만 원)를 넘는 부분에 22%(지방세 포함)의 양도소득세가 붙어. 국내 상장주식은 대주주가 아니면 현행 기준 매매차익이 비과세야. 이 차이 때문에 **같은 전략이라도 세후 성과는 한국과 미국에서 크게 달라**. 세전과 세후를 둘 다 보고해. (세법은 바뀔 수 있으니 실제 적용 전에 확인해.)

**공매도:** 개인이 개별 종목을 공매도하는 건 현실적으로 어려워. 하락 방어가 필요하면 **인버스 ETF**나 현금 비중 조절로 해.

### 전략 평가 지표

- 연율 수익률, 변동성, **샤프 비율과 그 표준오차** (18번)
- 최대 낙폭, 소르티노, 왜도
- 벤치마크 대비 초과수익, 정보비율
- **회전율과 손익분기 거래비용**
- 세전 / 세후
- **레짐별 성과**: 평온기와 혼란기에 각각 어떻게 했나
- 연도별 성과: 특정 몇 년에만 의존하는지

## 7. 검증 파이프라인: 백테스트에서 실전까지

실전 적용이 목표라면 백테스트 결과만으로는 부족해. **네 단계를 순서대로** 통과시켜.

### 1단계: 개발 구간 백테스트

2010년 ~ 2020년 정도를 개발 구간으로 써서 λ, 시그널 조합, 버퍼 폭을 결정해. 모든 결정은 **중첩 워크포워드**(12번)로 해.

### 2단계: 최종 테스트 구간 (단 한 번)

2021년 ~ 현재를 **한 번도 보지 않은** 구간으로 남겨두고, 1단계에서 확정한 규칙을 딱 한 번 적용해. 이 결과가 네가 보고할 "진짜 성과"야. 결과가 나빠도 규칙을 바꾸지 마. 바꾸는 순간 이 구간도 오염돼.

### 3단계: 페이퍼 트레이딩 (최소 3~6개월)

규칙을 코드로 고정하고, 매주(또는 매월) 실제로 신호를 계산해서 **"이렇게 샀을 것이다"를 날짜와 함께 기록**해. 실제 돈은 쓰지 않아. 이 단계에서 확인할 것은 이거야.

- 백테스트에서는 안 보였던 **실행상의 문제** (데이터가 늦게 올라옴, yfinance 오류, 휴장일 처리)

# 1. JM (Statistical Jump Model)이 뭐였나

## 1-1. 이 모형으로 하려는 일

매일 장이 끝나면, 오늘 시장(KOSPI 또는 S&P 500 지수)이 **평온한 상태**인지 **혼란한 상태**인지 판정하고 싶어. 결과는 날짜마다 0(평온) 또는 1(혼란)이라는 라벨 하나야. 이 라벨이 나중에 레짐 HAR의 $D_t$가 되고, 전략 S1에서 "주식을 들고 있을지 현금으로 갈지"를 정하는 근거가 돼.

## 1-2. 판정의 재료: 피처

하루의 수익률 하나만 보고 판정하면 너무 불안정해. 그래서 **최근 며칠~몇 주의 수익률을 요약한 숫자 여러 개**를 매일 계산해. 이걸 피처(feature)라고 하고, 날짜 t의 피처들을 묶은 걸 $x_t$라고 써.

우리가 쓰기로 한 피처는 세 종류야.

- **최근 평균 수익률:** 최근 수익률을 지수가중평균한 값. 요즘 시장이 오르는 중인지 내리는 중인지를 나타내.
- **하방편차:** 최근 **음의 수익률만** 제곱해서 평균 낸 뒤 제곱근. 요즘 하락이 얼마나 크고 잦은지를 나타내.
- **소르티노 비율:** 최근 평균 수익률 ÷ 하방편차. 하락 위험 대비 수익이 얼마나 되는지.

이 세 가지를 반감기 5일, 10일, 21일로 각각 계산하면 하루에 숫자 9개가 생겨. 그래서 $x_t$는 9개 숫자로 된 벡터야. 단위가 서로 다르니까 각 피처를 표준화(평균 0, 표준편차 1)해서 써.

## 1-3. 레짐마다 "전형적인 피처 값"이 있다

평온 레짐에서는 하방편차가 작고 소르티노 비율이 높아. 혼란 레짐에서는 하방편차가 크고 소르티노 비율이 낮거나 음수야. 각 레짐의 **전형적인 피처 값**을 그 레짐의 **중심**이라고 부르고 $\theta_0$(평온), $\theta_1$(혼란)로 써. 중심도 피처와 같은 9개 숫자짜리 벡터야.

어떤 날의 피처 $x_t$가 $\theta_1$에 가까우면 그날은 혼란 레짐처럼 보이는 거고, $\theta_0$에 가까우면 평온 레짐처럼 보이는 거야. 가까운 정도는 거리의 제곱으로 재.

$$
\text{거리 비용}_t(k) = \frac{1}{2}\left\| x_t - \theta_k \right\|^2
$$

## 1-4. 거리만 보면 생기는 문제

매일 그냥 더 가까운 중심을 고르면(이게 k-means 클러스터링이야), 하루짜리 급락이 있을 때 그날만 혼란으로 판정됐다가 다음 날 다시 평온으로 돌아와. 레짐이 너무 자주 바뀌어서 전략으로 쓰면 매매가 폭증하고, 대부분이 가짜 경보야.

## 1-5. JM의 해결책: 레짐을 바꿀 때마다 벌점

JM은 레짐 라벨 전체 $s_1, s_2, \dots, s_T$와 중심 $\theta_0, \theta_1$을 다음 식이 최소가 되도록 정해.

$$
\sum_{t=1}^{T} \frac{1}{2}\left\| x_t - \theta_{s_t} \right\|^2 \;+\; \lambda \times (\text{레짐이 바뀐 횟수})
$$

첫째 항은 "각 날을 그날 피처와 가까운 레짐에 배정하라"는 요구고, 둘째 항은 "레짐을 자주 바꾸지 마라"는 요구야. λ는 레짐 전환 한 번의 벌점이야.

10번에서 계산했던 숫자 예시를 다시 보면, 1차원 피처가 `[0.1, 0.2, 2.9, 0.3, 0.1, 2.8, 3.1, 3.0, 2.9]`이고 중심이 0(평온)과 3(혼란)일 때,

- 3일째 하루짜리 튐(2.9)을 혼란으로 바꾸면 거리 비용이 4.2만큼 줄지만, 들어갔다 나오느라 전환이 2번 생겨서 벌점이 2λ야. 그래서 λ가 2.1보다 크면 **무시**돼.
- 6~9일째처럼 4일 연속 높은 값은, 평온으로 두면 거리 비용이 17.4인데 혼란으로 바꾸면 전환 1번(λ)만 내면 돼. 그래서 λ가 17.4보다 작으면 **감지**돼.

즉 λ를 2.1과 17.4 사이로 정하면, 하루짜리 튐은 무시하고 여러 날 지속되는 변화만 레짐 전환으로 잡아. **튀는 크기뿐 아니라 지속 기간까지 보고 판정**하는 게 JM의 핵심이야.

## 1-6. 어떻게 계산하나

두 단계를 번갈아 반복해.

1. **라벨을 고정하고 중심을 갱신:** 레짐 0으로 배정된 날들의 피처 평균이 새 $\theta_0$, 레짐 1로 배정된 날들의 평균이 새 $\theta_1$.
2. **중심을 고정하고 라벨을 갱신:** 벌점 때문에 날짜마다 따로 고를 수 없어서, 동적계획법으로 전체 라벨 경로를 구해. 1일부터 차례로 "오늘이 레짐 k로 끝나는 가장 싼 경로의 비용" $V_t(k)$를 계산하는 방식이야.

$$
V_t(k) = \frac{1}{2}\|x_t - \theta_k\|^2 + \min_{j}\big[V_{t-1}(j) + \lambda \cdot \mathbf{1}\{j \neq k\}\big]
$$

"오늘 레짐 k의 비용 = 오늘의 거리 비용 + (어제 레짐 j로 끝나는 비용 + j에서 k로 바뀌면 λ) 중 가장 작은 것"이야.

## 1-7. 실전에서 쓰는 방식 (online)

- **매일의 판정:** $V_t(0)$과 $V_t(1)$ 중 작은 쪽이 오늘의 레짐이야. $V_t$는 오늘까지의 데이터만으로 계산되니까 미래 정보가 없어. (동적계획법의 역추적은 과거 라벨을 다시 고치는 단계라 쓰지 않아.)
- **중심의 재추정:** 매달 말, 그때까지의 데이터로만 중심을 다시 계산해.
- **λ 선택:** 5개 정도의 후보를 정하고, 개발 구간(2005~2020) 안에서 워크포워드로 성과가 가장 좋은 걸 골라서 고정해.

## 1-8. HMM과의 차이

HMM도 평온/혼란 레짐을 찾는 모형이고, 레짐 유지 확률(전이확률)을 데이터에서 추정해. 그런데 금융 데이터는 잡음이 커서 이 추정이 불안정하고, 결과적으로 레짐이 너무 자주 바뀌는 경향이 있어. JM은 "얼마나 쉽게 레짐을 바꿀지"를 추정하지 않고 λ로 **직접 정해.** 그래서 가짜 경보를 통제하기 쉬워. 우리 계획은 JM을 주 모형, HMM을 결과 확인용(강건성)으로 쓰는 거였어.

# 2. HAR, LHAR, SHAR 다시, 그리고 프로젝트에서의 사용법

## 2-1. 출발점: 하루에 변동성 숫자 하나

세 모형 모두 **"날마다 측정한 변동성 숫자"의 시계열**을 가지고 작동해. 이 숫자를 $RV_t$라고 쓸게. 이걸 어떻게 측정하느냐는 데이터에 따라 달라.

**일봉만 있을 때 (A층, 15년 이상):** 그날의 고가(H)와 저가(L)로 계산하는 Parkinson 추정치를 써.

$$
RV_t = \frac{1}{4\ln 2}\left(\ln\frac{H_t}{L_t}\right)^2
$$

예를 들어 고가 102, 저가 99면 $\ln(102/99) = 0.0299$이고, 제곱해서 $4\ln 2 = 2.77$로 나누면 약 0.00032야. 퍼센트 제곱 단위로 바꾸면 3.2이고, 제곱근을 취하면 그날 변동성이 약 1.8%라는 뜻이야.

**장중 봉이 있을 때 (C층, 5분봉 수집분):** 하루 동안의 5분 수익률을 제곱해서 모두 더해.

$$
RV_t = \sum_{i=1}^{78} r_{t,i}^2
$$

**여기서 짚어둘 것:** 장중 봉(5분이든 10분이든)은 **그날의 $RV_t$ 하나를 만드는 재료**야. 지난번에 "10분 변동성이 1일, 1주일 변동성에 영향을 준다"고 이해했던 부분이 틀린 이유가 이거야. 10분 변동성이 무언가를 예측하는 게 아니라, 10분 수익률들이 모여서 하루 변동성 숫자가 되는 거야.

## 2-2. HAR: 내일 변동성을 과거 변동성으로 예측

$$
RV_{t+1} = \beta_0 + \beta_d\,RV_t + \beta_w\,RV_t^{(w)} + \beta_m\,RV_t^{(m)} + \varepsilon_{t+1}
$$

- $RV_{t+1}$: **내일**의 변동성. 예측하려는 대상(종속변수)이야.
- $RV_t$: 오늘의 변동성
- $RV_t^{(w)}$: 최근 5 거래일 변동성의 평균
- $RV_t^{(m)}$: 최근 22 거래일 변동성의 평균

오늘 하나만 쓰면 하루의 우연한 튐에 휘둘리고, 한 달 평균만 쓰면 최근 변화에 둔해. 그래서 세 가지를 섞는데, **얼마씩 섞을지(β)는 과거 데이터로 회귀해서 정해.** 계수 추정은 네가 해본 OLS 회귀와 똑같아. 날짜마다 한 행을 만들고, 왼쪽에 오늘·주평균·월평균, 오른쪽에 내일 값을 두고 회귀하는 거야.

**숫자 예시:** 추정된 계수가 $\beta_0 = 0.1$, $\beta_d = 0.4$, $\beta_w = 0.35$, $\beta_m = 0.2$이고, 오늘 4.0, 최근 5일 평균 2.5, 최근 22일 평균 1.5라면

$$
\widehat{RV}_{t+1} = 0.1 + 0.4 \times 4.0 + 0.35 \times 2.5 + 0.2 \times 1.5 = 2.875
$$

오늘의 급등(4.0)이 일부는 내일까지 이어지지만 전부는 아니라고 예측하는 거야.

**HAR이 알려주지 않는 것:** 내일 **얼마나 크게** 움직일지는 예측하지만, **오를지 내릴지**는 전혀 알려주지 않아. $RV$는 수익률을 제곱해서 만든 거라 부호가 없으니까.

## 2-3. LHAR: 오늘 떨어졌는지를 추가 정보로 쓴다

HAR의 한계가 하나 있어. 오늘 변동성이 4.0인 날이 두 개 있다고 하자. 하나는 3% 폭락한 날이고 하나는 3% 급등한 날이야. HAR은 두 날의 내일 예측이 똑같아. 그런데 실제 시장에서는 **하락한 날 다음에 변동성이 더 커지는** 경향이 강해. 이걸 레버리지 효과라고 해.

LHAR은 이걸 반영하려고 **하락의 크기**를 변수로 추가해.

$$
L_t = \max(-r_t,\ 0)
$$

$r_t$는 오늘의 일별 수익률(%)이야. 오늘 −3%면 $L_t = 3$이고, +3%면 $L_t = 0$이야.

$$
RV_{t+1} = \beta_0 + \beta_d RV_t + \beta_w RV_t^{(w)} + \beta_m RV_t^{(m)} + \gamma_d L_t + \gamma_w L_t^{(w)} + \gamma_m L_t^{(m)} + \varepsilon_{t+1}
$$

$L_t^{(w)}$와 $L_t^{(m)}$는 최근 5일, 22일 동안의 하락 크기 평균이야.

**중요한 점 세 가지를 다시 확인할게.**

- LHAR이 예측하는 것도 **내일의 변동성**이야. HAR과 종속변수가 같아.
- LHAR도 **모든 날**을 써. 오른 날은 $L_t = 0$이라서 추가 항이 0이 될 뿐이고, 표본에서 빠지는 게 아니야.
- 기대하는 결과는 $\gamma > 0$, 즉 하락이 클수록 내일 변동성이 커진다는 거야.

**숫자 예시:** $\gamma_d = 0.3$이라면 (주·월 하락 항은 생략)

| | 오늘 수익률 | $L_t$ | HAR 예측 | 추가분 | LHAR 예측 |
|---|---|---|---|---|---|
| 폭락한 날 | −3% | 3 | 2.875 | 0.9 | 3.775 |
| 급등한 날 | +3% | 0 | 2.875 | 0 | 2.875 |

**LHAR은 일봉만 있으면 돼.** 수익률의 부호는 종가로 바로 알 수 있으니까. 그래서 15년치 일봉으로 하는 우리 주 분석에서 쓸 수 있어.

## 2-4. SHAR: 장중의 하락 봉과 상승 봉을 나눈다

LHAR은 **하루 종가 기준 부호 하나**만 봐. 그런데 종가는 +0.5%로 끝났어도 장중에 −3%까지 빠졌다가 회복한 날이 있을 수 있어. LHAR은 이날을 하락한 날로 보지 않아.

SHAR은 장중 봉 **하나하나의 부호**를 봐. 오늘의 $RV_t$를 상승 봉 부분과 하락 봉 부분으로 나눠.

$$
RS_t^+ = \sum_{\text{상승 봉}} r_{t,i}^2, \qquad RS_t^- = \sum_{\text{하락 봉}} r_{t,i}^2, \qquad RS_t^+ + RS_t^- = RV_t
$$

그리고 HAR의 $\beta_d RV_t$ 자리를 이 두 개로 바꿔.

$$
RV_{t+1} = \beta_0 + \beta^+ RS_t^+ + \beta^- RS_t^- + \beta_w RV_t^{(w)} + \beta_m RV_t^{(m)} + \varepsilon_{t+1}
$$

기대하는 결과는 $\beta^- > \beta^+$, 즉 하락 봉에서 나온 변동성이 내일 변동성에 더 크게 이어진다는 거야. 만약 $\beta^+ = \beta^-$라면 그냥 HAR과 같아지니까, SHAR의 핵심 검정은 "두 계수가 다른가"야.

**숫자 예시:** $\beta^+ = 0.1$, $\beta^- = 0.7$, 주·월 부분 합이 1.275라면

| | $RS^+$ | $RS^-$ | 오늘 부분 | SHAR 예측 |
|---|---|---|---|---|
| 장중 하락 위주 | 0.4 | 3.6 | 0.04 + 2.52 | 3.835 |
| 장중 상승 위주 | 3.6 | 0.4 | 0.36 + 0.28 | 1.915 |

**SHAR은 장중 봉이 있어야 해.** 그래서 yfinance 조건에서는 5분봉을 몇 달 모은 뒤에야 제대로 쓸 수 있어.

## 2-5. 세 모형 비교

| | HAR | LHAR | SHAR |
|---|---|---|---|
| 예측 대상 | 내일 변동성 | 내일 변동성 | 내일 변동성 |
| 추가 정보 | 없음 | 종가 기준 하락 크기 | 장중 하락 봉의 변동성 |
| 필요 데이터 | 일별 변동성 | 일봉 | 장중 봉 |
| 우리 프로젝트에서 쓸 수 있는 기간 | 15년 이상 | 15년 이상 | 5분봉 수집 이후 |

## 2-6. 우리 프로젝트에서 쓰는 법

**용도 ① 연구 (H1): "하락이 미래 변동성을 더 키우는가"를 검정**

1. KOSPI와 S&P 500 지수 일봉으로 날마다 Parkinson $RV_t$와 하락 크기 $L_t$를 계산한다.
2. 개발 구간(2005~2020) 안에서 워크포워드로 HAR과 LHAR의 내일 변동성 예측값을 만든다. 워크포워드란, 매달 그때까지의 데이터로만 계수를 추정하고 다음 한 달을 예측하는 걸 반복하는 거야.
3. 예측 오차(QLIKE)를 비교해서 LHAR이 HAR보다 나은지 확인한다.
4. JM 레짐 $D_t$를 곱한 항을 추가해서, **혼란 레짐에서 하락의 영향($\gamma$)이 더 커지는지** 본다. 이게 레짐 LHAR이야.
5. 5분봉이 쌓이면 같은 질문을 SHAR로 반복해서, 장중 정보로도 같은 결론이 나오는지 확인한다.

**용도 ② 전략 (S1): 투자 비중을 정하는 입력값**

LHAR로 예측한 내일 변동성을 투자 비중 계산에 써. 이게 변동성 타기팅이야.

$$
w_t = \min\left(\frac{\sigma_{\text{목표}}}{\hat\sigma_{t+1}},\ 1\right), \qquad \hat\sigma_{t+1} = \sqrt{\widehat{RV}_{t+1}}
$$

목표 일변동성이 1.0%일 때,

- 예측 변동성이 1.5%면 $w = 1.0/1.5 = 0.67$ → 주식 67%, 현금 33%
- 예측 변동성이 0.8%면 $1.0/0.8 = 1.25$인데 상한이 1이라 → 주식 100%

**용도에 없는 것:** HAR 계열로 **내일 오를지 내릴지를 예측하지 않아.** 방향 예측은 세 모형 중 어디에도 없어. 이게 바로 3번 질문(6번 항목)과 연결돼.

# 3. 6번(시그널화)과 7번(백테스트) 상세 설명

## 3-1. 6번 항목: 무엇이 문제였나

네가 쓴 목록의 1~5번은 전부 **변동성**에 관한 거였어. 그리고 6번에서 "이걸 투자 시그널로 만든다"고 했지. 그런데 변동성 예측은 방향을 알려주지 않아. 그러면 자연스럽게 이런 질문이 생겨. **방향을 모르는데 어떻게 돈을 벌지?**

이 질문에 대한 답이 목록에 없었다는 게 지적의 핵심이야. 답은 두 가지가 있고, 둘은 전혀 다른 메커니즘이야.

## 3-2. 먼저, "시그널"을 정확히 정의하자

시그널은 **t일(또는 t월) 장 마감 시점에 이미 알 수 있는 숫자**이고, 이 숫자로 **다음 기간의 포지션**을 정해.

$$
\text{시그널}_t \;\longrightarrow\; \text{비중}_t \;\longrightarrow\; \text{수익}_{t+1} = \text{비중}_t \times \text{수익률}_{t+1}
$$

시그널이 무엇을 정하느냐에 따라 두 종류로 나뉘어.

| 종류 | 시그널이 정하는 것 | 우리 프로젝트의 전략 |
|---|---|---|
| 위험 조절 시그널 | 주식에 **얼마나** 투자할지 (0~100%) | S1 |
| 종목 선택 시그널 | **어떤 종목**을 살지 | S2 |

## 3-3. 메커니즘 ①: 방향을 몰라도 위험 조절로 성과를 개선할 수 있다 (S1)

직관에 반하는 것 같지만, 방향을 전혀 예측하지 못해도 샤프 비율을 높일 수 있어. 수치로 보여줄게.

시장이 두 레짐을 오간다고 가정하자.

| 레짐 | 시간 비율 | 일평균 수익률 | 일변동성 |
|---|---|---|---|
| 평온 | 70% | +0.05% | 0.8% |
| 혼란 | 30% | 0.00% | 2.5% |

**바이앤홀드 (항상 100% 보유):**

- 평균 수익률 $= 0.7 \times 0.05 + 0.3 \times 0 = 0.035\%$
- 분산은 레짐별 (분산 + 평균²)을 비율대로 더한 뒤 전체 평균²을 빼면 돼: $0.7 \times (0.64 + 0.0025) + 0.3 \times (6.25 + 0) - 0.035^2 \approx 2.32$
- 변동성 $\approx 1.52\%$
- 일 샤프 $= 0.035 / 1.52 \approx 0.023$, 연율화($\times\sqrt{252}$)하면 **약 0.36**

**레짐 스위치 (평온일 때만 보유, 혼란일 때 현금, 현금 수익률 0으로 가정):**

- 평균 수익률 $= 0.7 \times 0.05 = 0.035\%$ (혼란 레짐의 평균이 원래 0이었으니 잃은 게 없어)
- 분산 $= 0.7 \times (0.64 + 0.0025) - 0.035^2 \approx 0.449$
- 변동성 $\approx 0.67\%$
- 연 샤프 **약 0.83**

**평균 수익률은 똑같은데 변동성이 절반 이하로 줄어서 샤프가 두 배 넘게 올랐어.** 방향 예측은 전혀 하지 않았어. "혼란 레짐은 수익에 기여하는 건 별로 없으면서 위험만 크다"는 사실만 이용한 거야.

**그런데 이 계산에는 두 가지 비현실적인 가정이 있어.** 그리고 이게 바로 백테스트로 확인해야 하는 부분이야.

- **레짐을 완벽하게, 즉시 안다고 가정했어.** 실제 online 레짐은 혼란 진입을 며칠~몇 주 늦게 감지해. 그러면 폭락 초반의 손실을 피하지 못해.
- **혼란 레짐의 평균 수익률이 0이라고 가정했어.** 실제로는 폭락 직후의 큰 반등이 혼란 레짐 안에서 일어나는 경우가 많아. 현금으로 빠져 있으면 이 반등을 놓쳐. 한국 시장은 특히 급락 후 반등이 빠른 경우가 있어서 이 비용이 클 수 있어.

그래서 이론상 0.36 → 0.83이었던 개선이 실제로는 얼마나 남는지, 혹은 사라지는지는 **백테스트를 해봐야만** 알 수 있어.

**변동성 타기팅도 같은 원리야.** 레짐처럼 0/1로 나누는 대신, LHAR 예측 변동성에 반비례해서 비중을 연속적으로 조절하는 거야. 변동성이 높을 것 같은 날 비중을 줄이면, 위험이 큰 구간의 노출이 줄어서 전체 변동성이 떨어져.

**정리하면 S1은 H1(변동성 예측)과 레짐을 투자에 연결하는 방법이야.** 변동성을 예측하는 연구가 "무엇을 사야 하나"가 아니라 "얼마나 투자해야 하나"라는 결정으로 이어지는 거지.

## 3-4. 메커니즘 ②: 종목 간 순위를 예측해서 종목을 고른다 (S2)

S1은 지수 하나에 얼마나 투자할지만 정해. 그러면 "어떤 종목을 살까"는 어디서 나올까? 이게 네 목록에서 빠진 **H2(횡단면 수익률 예측)**야.

**H2의 질문:** 이번 달 RSJ, MAX, IVOL 같은 값을 보고, **다음 달 어떤 종목이 다른 종목보다 더 오를지**를 예측할 수 있나?

여기서 예측하는 건 "삼성전자가 다음 달 오를까 내릴까"가 아니야. "KOSPI200 200개 종목 중에서 **상대적으로** 어느 종목이 더 나을까"야. 시장 전체가 떨어지는 달에도, 덜 떨어지는 종목을 고르는 건 가능해.

**실제 운용 절차 (매월 반복):**

1. 매월 마지막 거래일 장 마감 후, KOSPI200 종목 200개 각각에 대해 이번 달 일별 수익률로 RSJ, MAX, IVOL을 계산한다.
2. 4단계에서 효과가 확인된 시그널들을 각각 순위로 바꾸고 평균 내서 종목별 점수를 만든다.
3. 점수 상위 20%(40종목)를 동일 비중으로 산다.
4. 한 달 보유한 뒤 다시 1번부터 반복한다. 기존 보유 종목이 상위 30% 안에 남아 있으면 팔지 않는다(버퍼 규칙, 거래비용 절감용).

**왜 롱숏이 아니라 롱온리인가:** 학술 연구에서는 상위 20%를 사고 하위 20%를 공매도하는 롱숏 수익을 보고해. 그런데 개인이 한국 개별 종목을 공매도하는 건 현실적으로 어렵고, 2023년 11월~2025년 3월에는 아예 금지됐었어. 그래서 실전 전략은 상위 종목만 사는 롱온리로 하고, 롱숏 결과는 연구 결과로만 보고해.

**S2의 성패는 4단계 결과에 달려 있어.** 시그널이 단기 반전 효과를 통제한 뒤에도 살아남지 못하면, S2는 축소하거나 생략해야 해.

## 3-5. S3: 두 메커니즘의 결합

- **무엇을 살지:** S2의 종목 선택
- **얼마나 살지:** S1의 비중 조절

예를 들어 S1이 오늘 비중을 60%로 정했다면, S2가 고른 40종목에 총자산의 60%를 나눠 넣고 40%는 현금으로 둬.

## 3-6. 가설에서 전략까지의 연결 전체

| 연구 결과 | 무엇을 예측하나 | 어떤 결정에 쓰이나 | 전략 |
|---|---|---|---|
| H1 + 레짐 | 내일의 변동성 크기, 시장 상태 | 주식 비중 (얼마나) | S1 |
| H2 (H3 포함) | 다음 달 종목 간 상대 순위 | 종목 구성 (무엇을) | S2 |
| 둘 다 | | 둘 다 | S3 |

네 목록의 6번이 "시그널화한다"로만 되어 있었던 건, 이 표의 가운데 두 열이 비어 있었던 거야. 특히 H2가 없어서 **"무엇을" 결정할 근거가 없었어.**

## 3-7. 7번 항목: "백테스트한다" 한 줄에 빠진 것들

백테스트는 **과거 데이터로 전략을 모의 운용해보는 것**이야. 문제는 과거 데이터를 쓰기 때문에, 조금만 잘못해도 **당시에는 알 수 없었던 정보**를 쓰게 되고, 성과가 실제보다 훨씬 좋게 나온다는 거야. 학생 프로젝트에서 "샤프 3.0"이 나오는 경우는 거의 다 여기서 실수한 거야. 빠진 요소를 하나씩 설명할게.

### (1) 개발 구간과 최종 테스트 구간의 분리

전략을 만들 때는 여러 선택을 해. λ를 얼마로 할지, 상위 20%로 할지 30%로 할지, 목표 변동성을 얼마로 할지 같은 것들이야. 이 선택을 할 때마다 과거 성과를 보고 좋은 쪽을 고르게 돼.

문제는, **여러 개를 시도해보고 가장 좋은 걸 고르면 그 성과는 실력보다 운이 섞여서 부풀려진다**는 거야. 12번에서 계산했던 예시를 다시 보면, 실력이 전혀 없는(진짜 샤프 0인) 전략 변형 20개를 5년 데이터로 평가하고 그중 최고를 고르면, 그 최고 전략의 샤프가 평균적으로 약 0.84로 나와. 아무 실력이 없는데도.

그래서 데이터를 둘로 나눠.

| 구간 | 기간 | 용도 |
|---|---|---|
| 개발 구간 | 2005 ~ 2020 | 모든 선택과 시행착오 |
| 최종 테스트 구간 | 2021 ~ 현재 | 선택이 끝난 전략을 **한 번만** 평가 |

테스트 구간의 성과가 **보고할 진짜 성과**야. 개발 구간 성과는 선택 편향이 섞여 있어서 참고용이야.

### (2) 규칙 동결

"테스트를 한 번만 한다"는 게 실제로 지켜지려면, 테스트 전에 모든 규칙을 **문서로 적고 바꾸지 않아야** 해. 테스트 결과가 나쁘다고 λ를 바꿔서 다시 돌리면, 그 순간 테스트 구간도 개발 구간이 돼버려. 그래서 규칙을 설정 파일에 적고 git에 커밋한 뒤 태그를 붙여서 날짜를 남기기로 했지. 나중에 누가 물으면 "이 규칙은 이 날짜에 정했고, 테스트 데이터는 그 뒤에 처음 열었다"고 증명할 수 있어.

### (3) 미래 정보 차단 (look-ahead 방지)

가장 흔하고 가장 치명적인 오류야. 세 군데서 생겨.

**레짐:** 레짐 라벨이나 확률을 계산할 때, 그날까지의 데이터만 써야 해. HMM의 스무딩 확률이나 JM의 역추적 경로는 나중 데이터까지 보고 과거를 판정한 거라 쓰면 안 돼.

**파라미터:** LHAR 계수, JM 중심, 표준화에 쓰는 평균과 표준편차 모두 그 시점까지의 데이터로만 계산해야 해. 전체 기간으로 한 번 추정한 값을 과거 모든 날에 쓰면 미래 정보가 섞여.

**매매 시점:** t일 종가까지의 데이터로 계산한 시그널은 t일 종가에 체결할 수 없어. 종가가 나와야 계산이 되니까. 그래서 그 비중은 **t+1일 수익률**에 적용해야 해.

$$
\text{틀린 계산: } \text{비중}_t \times r_t \qquad \text{맞는 계산: } \text{비중}_t \times r_{t+1}
$$

틀린 계산을 하면, −3% 폭락한 날 "혼란 레짐 → 현금"이라는 판정을 **그 폭락이 일어나기 전에** 한 것처럼 계산돼. 성과가 극적으로 좋아지지만 완전히 허구야. 코드에서 비중을 하루 밀지 않고 돌려보면 성과가 얼마나 부풀려지는지 직접 확인할 수 있고, 그 차이가 이 오류가 얼마나 위험한지 보여줘.

### (4) 거래비용과 세금

전략이 매매를 할 때마다 비용이 들어.

$$
\text{순수익}_t = \text{총수익}_t - \text{편도 비용} \times \text{회전율}_t
$$

회전율은 그날(또는 그달) 비중이 얼마나 바뀌었는지야. 100% 주식에서 100% 현금으로 가면 회전율이 1이야.

**S1의 경우 (가정한 숫자로):** 레짐 스위치가 1년에 6번 바뀌고 편도 비용이 0.1%라면 연 0.6% 정도야. 변동성 타기팅은 매일 비중이 조금씩 바뀌어서, 하루 평균 3%p씩 비중이 움직이면 1년에 약 7.5번 전체를 사고판 셈이라 연 0.75% 정도가 돼. 그래서 비중 변화가 10%p 이상일 때만 조정하는 버퍼를 넣기로 했어.

**S2의 경우:** 매월 40종목 중 절반이 바뀌면 회전율이 높아서 비용이 훨씬 커. 한국은 매도 시 증권거래세도 붙어.

**세금:** 한국 거주자 기준으로 국내 상장주식 매매차익은 일반 투자자라면 현행 비과세이고, 미국 주식은 연간 공제액을 넘는 매매차익에 22%가 붙어. 국내 주식형 ETF는 증권거래세가 면제야. 같은 전략이라도 세후 성과가 나라마다 달라지니까 세전과 세후를 둘 다 계산해야 해. 세법은 바뀔 수 있으니 실제 적용할 때 다시 확인해.

### (5) 벤치마크

"바이앤홀드보다 샤프가 높다"만으로는 부족해. 더 단순한 방법으로도 같은 효과를 얻을 수 있다면, 복잡한 모형을 쓸 이유가 없으니까. 그래서 세 가지와 비교하기로 했어.

| 벤치마크 | 이 비교가 알려주는 것 |
|---|---|
| 바이앤홀드 | 전략이 아무것도 안 하는 것보다 나은가 |
| 200일 이동평균 규칙 (지수가 200일선 위면 보유, 아래면 현금) | JM 레짐이 가장 흔한 단순 타이밍 규칙보다 나은가 |
| 단순 변동성 타기팅 (최근 22일 실제 변동성으로 비중 조절) | LHAR 예측이 단순히 과거 변동성을 쓰는 것보다 나은가 |

S1이 바이앤홀드는 이기지만 200일선 규칙에 진다면, "레짐 모형은 단순 규칙 이상의 가치가 없다"가 결론이야. 그것도 중요한 결과야.

### (6) 성과 지표

수익률 하나로 평가하지 않아.

- 연 수익률, 연 변동성
- **샤프 비율과 그 표준오차.** 샤프 비율도 추정치라 오차가 있어서, 10년 데이터로도 샤프 0.5는 0과 통계적으로 구별이 안 될 수 있어.
- 최대 낙폭: 고점 대비 가장 크게 떨어진 폭. S1의 핵심 목적이 이걸 줄이는 거야.
- 회전율과 손익분기 비용: 비용이 얼마 이하여야 수익이 나는지
- **레짐별 성과:** 평온기와 혼란기에 각각 어떻게 했나. 전략의 성과가 어디서 나왔는지 보여줘.

### (7) 백테스트 이후: 페이퍼 트레이딩과 소액 실전

실제 투자에 쓰겠다는 게 목표니까, 백테스트는 끝이 아니야. 백테스트에서는 절대 드러나지 않는 문제들이 있어. yfinance가 어느 날 데이터를 안 주거나, 휴장일 처리가 틀리거나, 데이터가 늦게 갱신되거나, 실제로 체결할 수 있는 가격이 가정과 다르거나 하는 것들이야.

그래서 순서가 이래.

1. **페이퍼 트레이딩 (3~6개월):** 규칙을 고정한 코드로 매일 또는 매월 실제로 시그널을 계산하고, "이 날 이렇게 샀을 것이다"를 기록해. 돈은 쓰지 않아.
2. **성과 점검:** 페이퍼 트레이딩 성과가 백테스트에서 예상한 범위 안에 있는지 확인해.
3. **소액 실전:** 잃어도 괜찮은 금액으로 시작해.
4. **중단 규칙을 미리 정하기:** 예를 들어 실전 최대 낙폭이 백테스트 최대 낙폭의 1.5배를 넘으면 멈추고 원인을 분석해.

백테스트나 페이퍼 트레이딩을 통과해도 미래 성과가 보장되지는 않아. 실전 투입 규모는 네가 감당할 수 있는 범위에서 스스로 판단해야 해.

## 3-8. 7번 항목을 체크리스트로 정리하면

- [ ] 개발 구간(2005~2020)과 최종 테스트 구간(2021~현재)을 나눴다
- [ ] 모든 선택은 개발 구간 안에서만 했다
- [ ] 규칙을 문서화하고 git 태그로 동결했다
- [ ] 레짐, 파라미터, 표준화 모두 그 시점까지의 데이터로만 계산했다
- [ ] t일 시그널을 t+1일 수익률에 적용했다
- [ ] 거래비용과 세금을 반영했다 (세전·세후 모두)
- [ ] 바이앤홀드, 200일선, 단순 변동성 타기팅과 비교했다
- [ ] 샤프 비율의 표준오차와 최대 낙폭을 함께 보고했다
- [ ] 테스트 구간은 한 번만 돌렸다
- [ ] 통과한 전략만 페이퍼 트레이딩으로 넘겼다

# 전체 요약

**JM**은 매일 지수의 최근 수익률·하방편차·소르티노 비율을 보고 평온/혼란 레짐을 판정하는데, 레짐을 바꿀 때마다 벌점 λ를 매겨서 하루짜리 튐은 무시하고 지속되는 변화만 잡는 모형이야.

**HAR, LHAR, SHAR**은 모두 **내일의 변동성**을 예측하는 회귀야. HAR은 오늘·지난주·지난달 변동성을 쓰고, LHAR은 여기에 종가 기준 하락 크기를 추가하고, SHAR은 장중 하락 봉의 변동성을 따로 쓴다. 우리 프로젝트에서는 일봉 15년치로 **LHAR과 레짐 LHAR이 주 분석**이고, SHAR은 5분봉이 쌓이면 확인용이야. 그리고 LHAR 예측값은 **전략 S1의 투자 비중**을 정하는 데 쓰여.

**6번 지적**의 핵심은, 변동성 예측은 방향을 알려주지 않으니 시그널이 두 종류로 나뉜다는 거야. **S1은 변동성과 레짐으로 "얼마나" 투자할지**를 정해서 위험을 줄이는 방식이고, **S2는 H2의 횡단면 시그널로 "무엇을" 살지** 정하는 방식이야. 네 목록에는 H2가 빠져 있어서 S2의 근거가 없었어.

**7번 지적**의 핵심은, 백테스트는 과거 데이터를 쓰기 때문에 미래 정보가 섞이기 쉽고, 여러 번 시도하면 운이 성과로 둔갑한다는 거야. 그래서 구간 분리, 규칙 동결, 미래 정보 차단, 비용·세금, 단순 벤치마크 비교, 그리고 페이퍼 트레이딩까지가 전부 "백테스트한다" 안에 들어가야 해.

1. 이해한 내용 : 레짐을 파악한다. HMM으로. 레짐은 변동성의 크기로 정의한다.
2. 질문 : JM이 그래서 뭐고 어떤 역할을 한다는 건지 이해가 안 돼. 갑자기 lambda는 왜 나오는지도 모르겠고. 벌점은 왜 나오는 거고, 라벨 경로를 구한다는 건 또 뭔 말인지도 전혀 모르겠어.
3. 질문 : 그래서 장중 봉을 가져다가 왜 계산을 하느냐고. 장중봉이 그날의 RV를 만드는 재료겠지. 그런데 뭐 어쩌라고, 그건 그냥 일봉만으로도 구할 수 있잖아.

# 1. 이해한 내용에 대한 피드백

> 레짐을 파악한다. HMM으로. 레짐은 변동성의 크기로 정의한다.

**틀린 문장은 없지만, 프로젝트를 설명하는 문장으로는 부족해.** 네 가지를 짚을게.

**① 레짐 파악은 목적이 아니라 도구야.** 이 문장만 보면 프로젝트의 목표가 "레짐을 찾는 것"처럼 읽혀. 그런데 레짐은 두 곳에 쓰이는 **조건 변수**야. 하나는 "하락 변동성이 미래 변동성을 키우는 효과가 레짐에 따라 달라지는가", "종목 선택 시그널의 예측력이 레짐에 따라 달라지는가" 같은 연구 질문의 조건이고, 다른 하나는 전략에서 주식 비중을 줄일지 말지 정하는 근거야. 레짐을 찾았다는 것만으로는 아무 결론도 나오지 않아.

**② 누구의 레짐인지가 빠졌어.** 종목마다 레짐을 찾는 게 아니라 **시장 지수(KOSPI, S&P 500)의 레짐**이야. 모든 종목이 같은 날 같은 레짐 라벨을 공유해.

**③ HMM이라고 쓰는 건 괜찮지만, 조건이 붙어야 해.** HMM을 주 도구로 써도 돼. 다만 "그날까지의 데이터만으로 계산한 확률(필터링 확률)을 쓰고, 파라미터도 그 시점까지의 데이터로 주기적으로 다시 추정한다"는 조건 없이 HMM을 돌리면, 라이브러리 기본값이 미래 데이터를 섞어서 결과가 전부 오염돼. 그리고 HMM에는 레짐이 너무 자주 바뀌는 문제가 있어서, 그걸 어떻게 처리할지도 답할 수 있어야 해. 그게 2번 질문의 JM이 나온 이유야.

**④ 레짐이 하나 더 있어.** 공매도 금지 기간이라는 **제도적 레짐**이 빠졌어. 통계 모형으로 찾는 레짐과 달리, 정책으로 정해진 기간이라 데이터에서 추정할 필요가 없고, 이 프로젝트의 가장 차별화된 부분이야.

**고쳐 쓰면:** "시장 지수의 일별 데이터로, 그날까지의 정보만 써서 저변동성/고변동성 레짐을 매일 판정한다(HMM 또는 JM). 이 레짐과 공매도 금지 기간을 조건으로 두고, 변동성 예측과 종목 선택 시그널의 효과가 달라지는지 검정하며, 전략의 주식 비중을 정하는 데도 쓴다."

# 2. JM이 뭐고, λ·벌점·라벨 경로는 왜 나오나

처음부터 다시, 문제 상황부터 시작할게.

## 2-1. 풀려는 문제

매일 장이 끝나면 그날 시장이 평온(0)인지 혼란(1)인지 판정하고 싶어. 판정 재료로, 날마다 시장 상태를 나타내는 숫자 하나가 있다고 하자. 예를 들어 "최근 며칠의 변동성" 같은 값이야. 이 숫자를 $x_t$라고 할게.

8일치 값이 이렇다고 하자.

| 날 | 1 | 2 | 3 | 4 | 5 | 6 | 7 | 8 |
|---|---|---|---|---|---|---|---|---|
| $x_t$ | 1 | 1 | **3** | 1 | 1 | **3** | **3** | **3** |

평온한 날은 대체로 1, 혼란한 날은 대체로 3이라고 하자. 눈으로 보면, 3일째는 하루만 튄 것이고 6일째부터는 진짜로 상태가 바뀐 것 같아.

## 2-2. 가장 단순한 방법과 그 문제

"2보다 크면 혼란"처럼 날마다 따로 판정하면 이렇게 돼.

| 날 | 1 | 2 | 3 | 4 | 5 | 6 | 7 | 8 |
|---|---|---|---|---|---|---|---|---|
| 판정 | 0 | 0 | **1** | 0 | 0 | 1 | 1 | 1 |

3일째에 혼란으로 갔다가 4일째에 평온으로 돌아왔어. 레짐이 8일 동안 세 번 바뀐 거야. 이 판정으로 매매하면 3일째에 주식을 다 팔고 4일째에 다시 사게 되는데, 그 하루의 튐이 일회성이었다면 거래비용만 날린 거야. 실제 일별 데이터는 이런 튐이 훨씬 많아서, 날마다 따로 판정하면 레짐이 쉴 새 없이 바뀌어.

**문제의 원인:** 날마다 **따로** 판정하니까 "어제까지 평온이 이어졌다"는 정보가 전혀 반영되지 않아.

## 2-3. 해결 아이디어: 8일치 판정을 한꺼번에 정한다

날마다 따로 정하지 말고, **8일 전체의 판정 순서를 한 묶음으로** 정하자는 거야. 이 판정 순서 한 묶음을 **라벨 경로**라고 불러. 예를 들어 이런 것들이 라벨 경로야.

- 경로 A: 0, 0, 1, 0, 0, 1, 1, 1
- 경로 B: 0, 0, 0, 0, 0, 1, 1, 1
- 경로 C: 0, 0, 0, 0, 0, 0, 0, 0

날이 8일이고 날마다 0 또는 1이니까, 가능한 라벨 경로는 $2^8 = 256$개야. 이 중에서 **가장 좋은 경로 하나**를 고르는 게 목표야.

## 2-4. "좋은 경로"의 기준 두 가지

**기준 ① 데이터와 맞아야 한다.** 값이 3인 날을 평온(0)으로 판정하거나, 값이 1인 날을 혼란(1)으로 판정하면 데이터와 안 맞는 거야. 안 맞는 정도를 숫자로 재야 비교할 수 있으니까, "그 레짐의 전형적인 값에서 얼마나 떨어져 있나"로 재.

평온의 전형적인 값은 1, 혼란의 전형적인 값은 3이야. 이 전형적인 값을 **중심**이라고 불러. 각 날의 **불일치 비용**은 거리 제곱의 절반으로 정의할게.

$$
\text{불일치 비용} = \frac{1}{2}(x_t - \text{중심})^2
$$

- 값 1인 날을 평온(중심 1)으로 판정: $\frac{1}{2}(1-1)^2 = 0$
- 값 1인 날을 혼란(중심 3)으로 판정: $\frac{1}{2}(1-3)^2 = 2$
- 값 3인 날을 평온으로 판정: $2$
- 값 3인 날을 혼란으로 판정: $0$

**기준 ② 레짐이 자주 바뀌면 안 된다.** 경로 안에서 판정이 0→1 또는 1→0으로 바뀐 횟수를 세.

## 2-5. 두 기준을 하나의 점수로 합치기: 여기서 λ와 벌점이 나온다

두 기준은 서로 충돌해. 기준 ①만 보면 경로 A(날마다 가까운 쪽)가 불일치 비용 0으로 최고야. 기준 ②만 보면 한 번도 안 바뀌는 경로 C가 최고야. 둘 사이의 균형을 잡으려면 두 기준을 **같은 단위로 바꿔서 더해야** 해.

그래서 "레짐이 한 번 바뀔 때마다 불일치 비용 몇 점에 해당하는 벌점을 매긴다"고 정하는 거야. 그 **한 번 바뀔 때의 벌점이 λ**야.

$$
\text{경로의 총비용} = (\text{모든 날의 불일치 비용 합}) + \lambda \times (\text{레짐이 바뀐 횟수})
$$

**총비용이 가장 작은 경로를 고른다.** 이게 JM(Statistical Jump Model)의 전부야. "Jump"는 가격 점프가 아니라 **레짐이 바뀌는 것**을 가리켜.

## 2-6. 세 경로의 총비용을 직접 계산해보자

| 경로 | 불일치 비용 합 | 바뀐 횟수 | 총비용 |
|---|---|---|---|
| A: 0,0,1,0,0,1,1,1 | 0 | 3 | $3\lambda$ |
| B: 0,0,0,0,0,1,1,1 | 2 (3일째 값 3을 평온으로) | 1 | $2 + \lambda$ |
| C: 0,0,0,0,0,0,0,0 | 8 (3·6·7·8일째 값 3을 평온으로) | 0 | $8$ |

λ 값에 따라 어느 경로가 이기는지 보자.

| λ | A | B | C | 선택 | 결과 |
|---|---|---|---|---|---|
| 0.5 | **1.5** | 2.5 | 8 | A | 하루짜리 튐까지 레짐 전환으로 잡음 |
| 3 | 9 | **5** | 8 | B | 튐은 무시, 6일째부터의 진짜 변화는 잡음 |
| 10 | 30 | 12 | **8** | C | 진짜 변화까지 놓침 |

A와 B를 비교하면 $3\lambda < 2 + \lambda$, 즉 $\lambda < 1$이면 A가 이겨. B와 C를 비교하면 $2 + \lambda < 8$, 즉 $\lambda < 6$이면 B가 이겨. 그래서 **λ가 1과 6 사이면 우리가 원하는 경로 B**가 선택돼.

**λ가 하는 일을 정리하면:** "얼마나 확실한 증거가 있어야 레짐을 바꿀 것인가"를 정하는 값이야. 3일째 하루의 튐은 불일치 비용을 2만큼 줄여주지만, 그러려면 들어갔다 나오느라 두 번 바꿔야 해서 벌점이 $2\lambda$야. λ가 1보다 크면 벌점이 더 커서 무시돼. 반면 6~8일째는 3일 연속이라, 평온으로 두면 불일치 비용이 6이나 쌓이는데 혼란으로 바꾸면 벌점 λ 한 번만 내면 돼. 그래서 λ가 6보다 작으면 전환이 선택돼. **튄 크기뿐 아니라 며칠 지속됐는지까지** 판정에 들어가는 거야.

## 2-7. 256개 경로를 다 계산하지 않고 최선의 경로를 찾는 법

8일이면 256개라 다 계산할 수도 있지만, 실제 데이터는 수천 일이라 $2^{5000}$개를 계산할 수는 없어. 그래서 이렇게 해.

**하루씩 앞으로 가면서, 각 레짐별로 "오늘 이 레짐으로 끝나는 경로 중 가장 싼 것의 총비용"만 기억한다.** 이걸 $V_t(0)$, $V_t(1)$이라고 부를게.

오늘 레짐 0으로 끝나는 경로는 어제 0에서 왔거나(벌점 없음), 어제 1에서 왔거나(벌점 λ) 둘 중 하나야. 그러니까

$$
V_t(0) = (\text{오늘을 0으로 판정한 불일치 비용}) + \min\big(V_{t-1}(0),\ V_{t-1}(1) + \lambda\big)
$$

$$
V_t(1) = (\text{오늘을 1로 판정한 불일치 비용}) + \min\big(V_{t-1}(1),\ V_{t-1}(0) + \lambda\big)
$$

어제까지의 가장 싼 경로만 알면 오늘의 가장 싼 경로를 알 수 있으니까, 나머지 수많은 경로는 계산할 필요가 없어. λ = 3으로 직접 계산해볼게.

| 날 | $x_t$ | 0으로 판정 비용 | 1로 판정 비용 | $V_t(0)$ | $V_t(1)$ | 오늘까지 기준 판정 |
|---|---|---|---|---|---|---|
| 1 | 1 | 0 | 2 | 0 | 2 | 0 |
| 2 | 1 | 0 | 2 | 0 + min(0, 2+3) = **0** | 2 + min(2, 0+3) = **4** | 0 |
| 3 | 3 | 2 | 0 | 2 + min(0, 4+3) = **2** | 0 + min(4, 0+3) = **3** | 0 |
| 4 | 1 | 0 | 2 | 0 + min(2, 3+3) = **2** | 2 + min(3, 2+3) = **5** | 0 |
| 5 | 1 | 0 | 2 | 0 + min(2, 5+3) = **2** | 2 + min(5, 2+3) = **7** | 0 |
| 6 | 3 | 2 | 0 | 2 + min(2, 7+3) = **4** | 0 + min(7, 2+3) = **5** | 0 |
| 7 | 3 | 2 | 0 | 2 + min(4, 5+3) = **6** | 0 + min(5, 4+3) = **5** | **1** |
| 8 | 3 | 2 | 0 | 2 + min(6, 5+3) = **8** | 0 + min(5, 6+3) = **5** | 1 |

마지막 날 $V_8(1) = 5$가 가장 작고, 이건 경로 B의 총비용 $2 + 3 = 5$와 정확히 같아. 256개를 다 계산하지 않고도 최선의 경로를 찾은 거야.

**맨 오른쪽 열이 실전에서 쓰는 판정이야.** 매일 장이 끝났을 때, 그날까지의 데이터로 $V_t(0)$과 $V_t(1)$ 중 작은 쪽을 오늘의 레짐으로 판정해. 미래 데이터는 전혀 쓰지 않아.

이 열에서 두 가지를 볼 수 있어.

- **3일째:** 하루짜리 튐이 왔지만 $V_3(0) = 2 < V_3(1) = 3$이라 평온을 유지했어. 가짜 경보를 피했어.
- **6일째:** 진짜 변화가 시작됐는데 아직 평온으로 판정했어. **7일째에야** 혼란으로 바뀌었어. 하루 늦게 감지한 거야. 가짜 경보를 줄이는 대가로 진짜 변화도 조금 늦게 잡는다는 뜻이고, λ가 클수록 이 지연이 길어져.

## 2-8. 중심은 어디서 오나

위 예시에서는 중심을 1과 3으로 미리 알고 있었지만, 실제로는 몰라. 그래서 두 단계를 번갈아 반복해.

1. 중심을 임시로 정하고, 위 방법으로 최선의 라벨 경로를 구한다.
2. 그 경로에서 평온으로 판정된 날들의 $x_t$ 평균을 새 평온 중심으로, 혼란으로 판정된 날들의 평균을 새 혼란 중심으로 바꾼다.
3. 경로가 더 이상 안 바뀔 때까지 1~2를 반복한다.

실제로는 $x_t$가 숫자 하나가 아니라 여러 개(최근 평균 수익률, 최근 하락폭의 크기, 둘의 비율 등을 기간별로)야. 그래도 원리는 같아. 불일치 비용이 여러 숫자에 대한 거리 제곱의 합이 될 뿐이야.

## 2-9. λ는 어떻게 정하나

이론적으로 정해지는 값이 아니야. 후보 몇 개(예: 5개)를 정하고, 개발 구간 데이터에서 각 λ로 레짐을 판정해 전략을 돌려본 뒤, 성과가 가장 좋은 λ를 골라서 고정해. 테스트 구간에서는 바꾸지 않아.

## 2-10. 우리 프로젝트에서의 역할, 그리고 HMM과의 관계

**역할:** 매일 0/1 레짐 라벨을 만드는 것. 그 라벨은 연구 질문의 조건 변수와 전략의 주식 비중 판단에 쓰여.

**HMM과의 관계:** HMM도 같은 일을 해. 다만 HMM은 "레짐이 바뀔 확률"을 데이터에서 추정하고, 판정을 확률로 내놔. 그 전환 확률이 사실상 JM의 λ와 같은 역할을 하는데, 데이터로 추정하다 보니 잡음 많은 금융 데이터에서는 레짐이 너무 자주 바뀌게 추정되는 경향이 있어. JM은 그 값을 추정하지 않고 λ로 **직접 정해서** 전환 빈도를 통제해.

그래서 선택지는 둘이야. HMM을 주 도구로 쓰고 전환이 잦은 문제를 따로 다루거나, JM을 주 도구로 쓰고 HMM은 결과가 같은지 확인하는 용도로 쓰거나. 어느 쪽이든 결과가 두 방법에서 비슷하게 나와야 신뢰할 수 있어.

# 3. 일봉으로도 되는데 왜 장중 봉으로 계산하나

**네 반박은 반은 맞아.** 일봉만으로도 하루 변동성을 구할 수 있고, 실제로 yfinance 제약 때문에 이 프로젝트의 주 분석은 일봉으로 해. 그런데 장중 봉에는 일봉이 **줄 수 없는 것이 두 가지** 있어. 정밀도와 하루 안에서의 방향 분해야.

## 3-1. 정밀도: 같은 "하루 변동성"이라도 오차 크기가 다르다

하루 변동성의 참값을 $\sigma^2$라고 하고, 각 방법으로 추정했을 때 추정값이 참값 주변에서 얼마나 흔들리는지 보자.

**일봉 종가 수익률 하나로 추정:** 그날 수익률 $r_t$ 하나를 제곱해서 $r_t^2$를 하루 분산으로 쓰는 거야. 수익률이 평균 0, 분산 $\sigma^2$인 정규분포라면

- $E[r_t^2] = \sigma^2$ → 평균적으로는 맞아.
- $\text{Var}(r_t^2) = E[r_t^4] - (E[r_t^2])^2 = 3\sigma^4 - \sigma^4 = 2\sigma^4$ (정규분포의 4차 모멘트가 $3\sigma^4$이므로)
- 표준편차는 $\sqrt{2}\,\sigma^2$, 즉 **참값 대비 약 141%**

추정값이 참값의 0배에서 3배 사이를 오가는 수준이야. 하루치 추정으로는 거의 쓸모가 없어.

**장중 봉 N개로 추정:** 하루를 N개 구간으로 쪼개면 각 구간 수익률의 분산은 $\sigma^2/N$이고, 제곱해서 더하면 위와 같은 계산으로 상대오차가 $\sqrt{2/N}$이 돼. 서로 독립인 오차 N개를 더하니까 상대오차가 $\sqrt{N}$배로 줄어드는 거야.

**일봉 고가·저가로 추정 (Parkinson):** 종가 하나보다는 훨씬 나아. 고가와 저가는 하루 동안 가격이 어디까지 움직였는지 담고 있으니까. 대략 종가 수익률 5개 정도의 정보량이라, 상대오차가 약 60% 수준이야.

| 방법 | 상대오차 (대략) |
|---|---|
| 일봉 종가 수익률 제곱 | 약 141% |
| 일봉 고가·저가 (Parkinson) | 약 60% |
| 60분봉 (하루 7개) | 약 53% |
| 10분봉 (하루 39개) | 약 23% |
| 5분봉 (하루 78개) | 약 16% |

(정규분포를 가정한 이론값이라, 실제 데이터에서는 이보다 조금 나빠.)

**이 오차가 왜 중요한가:** HAR 같은 회귀에서 설명변수(오늘 변동성)에 오차가 크면, 추정된 계수가 0 쪽으로 쪼그라들어. 오늘의 변동성이 내일에 주는 진짜 영향보다 약하게 추정된다는 뜻이야. 그리고 예측을 평가할 때 정답으로 쓰는 "내일 변동성"도 오차투성이면, 어느 모형이 더 나은지 구별하기가 어려워져.

## 3-2. 하루 안에서의 방향 분해: 일봉으로는 불가능하다

이게 더 결정적인 이유야. 이런 하루를 생각해봐.

- 시가 100에서 시작
- 오전에 97까지 3% 급락
- 오후에 천천히 회복해서 100.2로 마감

| 데이터 | 이날에 대해 알 수 있는 것 |
|---|---|
| 일봉 종가 | 전일 대비 +0.2%. 조용하게 조금 오른 날로 보임 |
| 일봉 고가·저가 | 저가 97, 고가 100.2. 크게 움직인 날이라는 건 알지만, **먼저 떨어졌는지 먼저 올랐는지는 모름** |
| 장중 봉 | 오전에 하락 봉들이 몰렸고 오후에 상승 봉들이 이어졌음. **하락 변동성과 상승 변동성을 따로 계산 가능** |

하락 변동성($RS^-$)과 상승 변동성($RS^+$)을 **날마다** 나누려면, 하루 안의 움직임을 봉 단위로 봐야 해. 일봉에는 그 정보가 아예 없어.

그럼 일봉으로는 방향 분해를 전혀 못 하느냐? 두 가지 대안이 있고, 우리 계획에서 이걸 쓰기로 했던 거야.

- **하루 단위 부호만 쓰기 (LHAR):** "오늘 종가 기준으로 떨어졌나, 얼마나 떨어졌나"를 변수로 쓰는 것. 위 예시 같은 날은 하락으로 안 잡히는 한계가 있어.
- **기간을 늘려서 쓰기:** 한 달 동안의 일별 수익률 21개로 그달의 $RS^+$, $RS^-$를 계산하는 것. 방향 분해는 되지만, 이건 "하루 안의 급락"이 아니라 "한 달 동안 크게 떨어진 날이 많았나"를 재는 거라 다른 양이야. 그리고 그달 수익률과 기계적으로 얽혀 있어서, 단순히 "지난달 많이 떨어진 종목"과 구별하기 어려워져.

## 3-3. 정리

일봉으로 할 수 있는 것과 없는 것이 분명히 갈려.

| | 일봉 | 장중 봉 |
|---|---|---|
| 하루 변동성 | 가능하지만 오차 큼 (약 60%) | 정밀함 (약 16%, 5분봉) |
| 하루 단위 하락/상승 변동성 분리 | **불가능** | 가능 |
| 장기 이력 (yfinance) | 15년 이상 | 60분봉 2년, 5분봉 60일 |

그래서 이 프로젝트는 **긴 기간이 필요한 분석(레짐 에피소드가 여러 번 들어가야 하는 분석)은 일봉으로, 하루 안의 방향 분해가 필요한 분석은 장중 봉으로** 나눈 거야. 네 말대로 일봉으로 할 수 있는 건 일봉으로 해. 다만 "일봉으로도 똑같이 구할 수 있다"는 건 틀렸어. **정밀도가 떨어지고, 하루 단위 방향 분해는 원천적으로 불가능해.**

# 4. 종목을 많이 쓰면 장중 봉 표본 부족이 해결되나

**부분적으로만 해결돼. 무엇이 해결되고 무엇이 안 되는지 나눠서 봐야 해.**

먼저 "표본 부족"이 무엇을 말하는지 정확히 하자. 여기서 문제는 **한 종목의 하루 변동성을 계산할 때 쓰는 봉 개수(N)가 적어서, 그 종목의 그날 값에 측정오차가 크다**는 거야. 종목 수가 많아지면 이 오차가 어떻게 되는지 보자.

## 4-1. 해결되는 것: 여러 종목에 걸친 "평균적인 효과"의 추정 정밀도

각 종목의 측정오차가 서로 독립이라고 하자. 예를 들어 "하락 변동성이 다음 날 변동성을 키우는 정도"를 500개 종목을 묶어서 추정하면, 종목별 오차가 서로 상쇄되면서 **추정치의 표준오차가 줄어들어.** 종목이 4배 많으면 표준오차가 대략 절반이 돼. 이 부분은 네 말이 맞아.

## 4-2. 해결되지 않는 것 ①: 측정오차로 인한 편향

측정오차가 있는 변수를 설명변수로 쓰면, 계수가 **0 쪽으로 쪼그라드는 편향**이 생겨. 진짜 값을 $x^*$, 측정오차를 $u$, 관측값을 $x = x^* + u$라고 하면, 회귀계수는 이 값으로 수렴해.

$$
\hat{b} \;\to\; b \times \frac{\text{Var}(x^*)}{\text{Var}(x^*) + \text{Var}(u)}
$$

예를 들어 진짜 값의 분산과 오차의 분산이 같다면 이 비율이 0.5라서, 추정된 효과는 진짜 효과의 절반이야. **이 비율은 종목 수와 무관해.** 종목 수를 10배로 늘리면 표준오차는 줄지만, "진짜의 절반"이라는 틀린 값을 **더 정밀하게** 추정하게 될 뿐이야. 편향을 줄이려면 각 종목의 측정 자체를 정밀하게 해야 하고(봉을 더 잘게), 종목 수로는 안 돼.

## 4-3. 해결되지 않는 것 ②: 개별 종목 단위의 의사결정

전략 S2는 매달 **종목별 시그널 값으로 순위를 매겨서** 상위 종목을 사. 이때 쓰이는 건 여러 종목의 평균이 아니라 **각 종목의 개별 값**이야. 각 종목의 값에 오차가 크면, 진짜로는 상위권인 종목이 오차 때문에 중위권으로 밀리고 그 반대도 생겨. 순위가 뒤섞이는 거지.

순위와 미래 수익률의 상관(IC)으로 보면, 진짜 IC에 대략 $\sqrt{\text{Var}(x^*)/(\text{Var}(x^*) + \text{Var}(u))}$가 곱해져서 약해져. 위 예시(비율 0.5)라면 IC가 약 0.71배가 돼. 종목이 200개든 2,000개든 **매달 각 종목의 순위가 흔들리는 정도는 똑같아.**

## 4-4. 해결되지 않는 것 ③: 오차가 종목끼리 독립이 아닐 때

4-1은 오차가 종목끼리 독립이라는 전제였어. 그런데 같은 날에는 모든 종목이 같은 시장 충격을 받고, 같은 시간대에 같은 방식으로 거래가 몰려. 그래서 오차가 종목끼리 상관돼 있어. 상관된 오차는 평균을 내도 상쇄가 덜 돼서, 4-1의 이득도 기대보다 작아.

## 4-5. 해결되지 않는 것 ④: 시간 방향의 표본 수

종목을 늘리면 **같은 날의 관측**이 늘어날 뿐, **날짜나 사건의 수**는 늘지 않아. 두 가지가 여기에 걸려.

- **레짐:** 레짐은 시장 지수 하나로 정의돼. 종목을 늘려도 위기 에피소드가 2008년, 2020년 같은 몇 번뿐이라는 사실은 그대로야. 60분봉 2년치라면 그 2년 안에 들어 있는 레짐 전환만 볼 수 있어.
- **공매도 금지 기간:** 약 17개월이라는 기간은 종목을 몇 개 쓰든 17개월이야.

## 4-6. 정리

| 문제 | 종목 수를 늘리면 |
|---|---|
| 평균 효과 추정의 표준오차 | 줄어듦 (단, 오차가 종목 간 상관되면 덜 줄어듦) |
| 측정오차로 인한 계수 편향 | **그대로** |
| 종목별 순위 판단의 정확도 (전략) | **그대로** |
| 레짐 에피소드 수, 공매도 금지 기간 길이 | **그대로** |

**결론:** 종목을 많이 쓰는 건 통계적 정밀도를 높이는 데는 도움이 되지만, 장중 봉 개수가 적어서 생기는 문제의 핵심인 **편향과 개별 종목 판단의 부정확성**은 해결하지 못해. 그래서 측정 해상도(봉을 얼마나 잘게 쓰느냐)와 종목 수는 서로 대체재가 아니라 **다른 문제를 푸는 도구**야.

# 5. good vol / bad vol을 구분하는데 왜 "방향을 모른다"고 하나

**좋은 지적이고, 내 표현이 부정확했어.** "방향"이라는 단어가 두 가지 다른 것을 가리키는데 구분하지 않고 썼어.

## 5-1. "방향"의 두 가지 의미

| | 의미 | semivariance로 알 수 있나 |
|---|---|---|
| **과거 움직임의 방향** | 오늘의 변동성이 하락 움직임에서 왔나, 상승 움직임에서 왔나 | **알 수 있어.** 이게 good vol / bad vol이야 |
| **미래 수익률의 방향** | 내일(다음 달) 주가가 오를까, 내릴까 | **변동성 예측으로는 알 수 없어** |

내가 "변동성 예측은 방향을 모른다"고 한 건 **두 번째 의미**였어. 정확하게는 이렇게 말했어야 해. **"변동성 모형에 과거의 방향 정보를 넣어도, 그 모형이 예측하는 건 미래 변동성의 크기이지 미래 수익률의 방향이 아니다."**

## 5-2. 입력과 출력을 나눠서 보면 명확해진다

변동성 예측 모형(SHAR)을 다시 보자.

$$
RV_{t+1} = \beta_0 + \beta^+ RS_t^+ + \beta^- RS_t^- + \beta_w RV_t^{(w)} + \beta_m RV_t^{(m)} + \varepsilon_{t+1}
$$

- **입력(오른쪽):** 오늘의 good vol($RS_t^+$)과 bad vol($RS_t^-$). 과거 움직임의 방향 정보가 **들어 있어.**
- **출력(왼쪽):** 내일의 $RV_{t+1}$. 수익률을 제곱해서 더한 값이라 **부호가 없어.** 내일 −3% 움직여도, +3% 움직여도 같은 값이야.

그래서 이 모형이 알려주는 건 **"오늘 bad vol이 컸으니 내일은 크게 흔들릴 것이다"**야. "내일 떨어질 것이다"가 아니야. 입력에 방향 정보를 넣었다고 출력에 방향 정보가 생기지는 않아.

## 5-3. 그럼 bad vol이 크면 내일 떨어지는 경향은 없나?

이건 **별개의 질문**이고, 별개의 회귀로 확인해야 해. 예측 대상이 부호 있는 수익률로 바뀌어.

$$
r_{t+1} = a + b \cdot (\text{bad vol 비중})_t + \dots
$$

이게 우리 프로젝트의 H2(수익률 예측)야. 그리고 여기서 주의할 점이 있어. 기존 연구에서 나온 결과는 **직관과 반대 방향**이야. 상승 변동성 비중이 높았던 종목(급등 위주)이 이후 수익률이 **낮고**, 하락 변동성 비중이 높았던 종목(급락 위주)이 이후 수익률이 **높은** 경향이 보고됐어. 급등에 과잉반응했다가 되돌아오는 것으로 해석해. 그리고 이 효과도 크지 않고, 단순히 "최근 많이 오른 종목이 되돌아오는" 단기 반전 효과와 상당 부분 겹쳐.

즉 **bad vol이 크다고 해서 이후에도 계속 떨어진다는 보장은 없고, 오히려 반대일 수 있어.** 이게 과거 방향 정보와 미래 방향이 별개라는 걸 보여주는 예야.

## 5-4. 프로젝트에서의 두 가지 용도

같은 good vol / bad vol이 두 군데에서 다른 목적으로 쓰여.

| 용도 | 예측 대상 | 알려주는 것 | 전략에서의 쓰임 |
|---|---|---|---|
| 변동성 예측 (H1) | 미래 변동성 크기 | "bad vol이 크면 앞으로 더 크게 흔들린다" | 주식 비중을 얼마나 줄일지 (S1) |
| 수익률 예측 (H2) | 미래 수익률 (종목 간 상대 순위) | "bad vol 비중이 높은 종목이 다음 달 상대적으로 나을 수 있다" (검정 대상) | 어떤 종목을 살지 (S2) |

**결론:** 네 말대로 이 프로젝트는 good vol과 bad vol을 구분하고, 과거 움직임의 방향은 알 수 있어. 내가 부정확하게 말한 거야. 정확한 요점은, **변동성 예측의 결과물은 미래 수익률의 방향을 알려주지 않으며, 미래 수익률의 방향은 H2라는 별도의 질문으로 따로 검정해야 한다**는 거야.

# JM에서 말하는 "비용", "가장 싼 경로", "원하는 경로"

세 표현이 모두 헷갈리는 이유는, 내가 **"비용"이라는 단어를 설명 없이 먼저 쓰고** 그 정의를 나중에 붙였기 때문이야. 순서를 바로잡아서 처음부터 다시 설명할게.

---

## 1. 먼저, JM이 푸는 문제를 "표 고르기"로 다시 정의하자

8일치 데이터가 있어. 날마다 "최근 시장 변동성" 같은 숫자 하나야.

| 날 | 1 | 2 | 3 | 4 | 5 | 6 | 7 | 8 |
|---|---|---|---|---|---|---|---|---|
| 값 | 1 | 1 | 3 | 1 | 1 | 3 | 3 | 3 |

우리가 최종적으로 만들고 싶은 건 **날마다 평온(0) 또는 혼란(1)을 적은 판정표 하나**야. 예를 들어 이런 표들이 후보가 될 수 있어.

| 후보 | 1 | 2 | 3 | 4 | 5 | 6 | 7 | 8 |
|---|---|---|---|---|---|---|---|---|
| 판정표 A | 0 | 0 | 1 | 0 | 0 | 1 | 1 | 1 |
| 판정표 B | 0 | 0 | 0 | 0 | 0 | 1 | 1 | 1 |
| 판정표 C | 0 | 0 | 0 | 0 | 0 | 0 | 0 | 0 |

날마다 0 또는 1이니까 가능한 판정표는 모두 $2^8 = 256$개야. 앞에서 "라벨 경로"라고 부른 게 바로 이 **판정표 하나**야. 날짜 순서대로 늘어선 0과 1의 줄이라서 "경로"라고 부르는 거고, 다른 뜻은 없어.

**JM이 하는 일은 256개의 판정표 중에서 하나를 고르는 것**이야.

---

## 2. 표를 고르려면 채점 기준이 필요하다: 이게 "비용"이다

256개 중에서 하나를 고르려면, 각 표가 **얼마나 나쁜지**를 숫자로 매길 기준이 있어야 해. 그래야 컴퓨터가 비교할 수 있으니까.

여기서 **"비용"은 돈이 아니라, 판정표가 얼마나 나쁜지를 나타내는 점수**야. 우리가 직접 정의한 채점 규칙이고, **점수가 낮을수록 좋은 판정표**야. 골프 점수처럼 낮을수록 좋다고 생각하면 돼.

그리고 "레짐의 비용"이라는 표현은 부정확했어. **레짐 자체에 비용이 있는 게 아니야.** 판정표에 점수를 매기는 항목이 두 가지 있고, 그걸 각각 설명할게.

### 채점 항목 ①: 판정이 데이터와 맞지 않는 정도

평온한 날의 전형적인 값은 1, 혼란한 날의 전형적인 값은 3이라고 하자. 그러면 어떤 날의 판정이 그날 값과 얼마나 안 맞는지를 이렇게 잴 수 있어.

$$
\text{불일치 점수} = \frac{1}{2}(\text{그날 값} - \text{판정한 레짐의 전형적인 값})^2
$$

| 그날 값 | 평온(0)으로 판정 | 혼란(1)으로 판정 |
|---|---|---|
| 1 | $\frac{1}{2}(1-1)^2 = 0$ (잘 맞음) | $\frac{1}{2}(1-3)^2 = 2$ (안 맞음) |
| 3 | $\frac{1}{2}(3-1)^2 = 2$ (안 맞음) | $\frac{1}{2}(3-3)^2 = 0$ (잘 맞음) |

값이 1인 날을 혼란으로 판정하거나, 값이 3인 날을 평온으로 판정하면 2점씩 쌓여.

### 채점 항목 ②: 판정이 바뀐 횟수

판정표 안에서 0→1이나 1→0으로 바뀐 횟수를 세고, **한 번 바뀔 때마다 λ점**을 더해. 이게 "벌점"이야. λ는 우리가 정하는 숫자고, 여기서는 3으로 두자.

### 왜 두 번째 항목이 필요한가

항목 ①만 쓰면, 날마다 값에 가장 가까운 레짐을 고르는 판정표 A가 0점으로 최고점이야. 그런데 A는 3일째 하루의 튐 때문에 혼란으로 갔다가 다음 날 돌아와. 이 판정표로 매매하면 하루 만에 팔았다가 다시 사게 되고, 그 튐이 일회성이었다면 거래비용만 날린 거야.

그래서 "판정을 바꾸려면 그만한 이유가 있어야 한다"는 요구를 점수에 넣은 거야. 판정을 바꾸면 λ점을 잃으니까, **불일치 점수를 λ점보다 많이 줄여줄 때만** 바꾸는 게 이득이 돼.

### 판정표의 총점

$$
\text{총점} = (\text{모든 날의 불일치 점수 합}) + \lambda \times (\text{판정이 바뀐 횟수})
$$

---

## 3. 세 후보를 채점해보자 (λ = 3)

| 판정표 | 불일치 점수 합 | 바뀐 횟수 | 벌점 | **총점** |
|---|---|---|---|---|
| A: 0,0,1,0,0,1,1,1 | 0 (모든 날이 가까운 쪽) | 3 | 9 | **9** |
| B: 0,0,0,0,0,1,1,1 | 2 (3일째 값 3을 평온으로) | 1 | 3 | **5** |
| C: 0,0,0,0,0,0,0,0 | 8 (3·6·7·8일째 값 3을 평온으로) | 0 | 0 | **8** |

나머지 253개 표도 같은 방식으로 점수를 매길 수 있고, 그중 **총점이 가장 낮은 표가 B**야(총점 5).

**"가장 싼 경로"는 이 총점이 가장 낮은 판정표를 말하는 거야.** 그리고 JM은 **그 판정표를 답으로 채택**해. 중요한 건 총점의 숫자 5가 아니라, **그 점수를 받은 판정표 B 자체**야. 총점은 표들을 비교하기 위한 수단일 뿐이야.

---

## 4. "우리가 원하는 경로"라는 말의 뜻

이 8일치 데이터는 내가 **설명하려고 일부러 만든** 거야. 3일째는 하루만 튄 거고, 6일째부터는 진짜로 상태가 바뀐 거라고 **답을 알고** 만들었어. 그래서 사람이 보기에 올바른 판정표는 B야. 하루짜리 튐은 무시하고, 지속되는 변화는 잡은 표니까. "우리가 원하는 경로"는 이 **정답 판정표 B**를 가리킨 거야.

그리고 λ를 바꿔가며 채점해본 이유는, **λ에 따라 JM이 고르는 표가 달라진다**는 걸 보여주려는 거였어.

| λ | A 총점 | B 총점 | C 총점 | JM이 고르는 표 |
|---|---|---|---|---|
| 0.5 | **1.5** | 2.5 | 8 | A: 튐까지 레짐 전환으로 잡음 |
| 3 | 9 | **5** | 8 | B: 정답 |
| 10 | 30 | 12 | **8** | C: 진짜 변화까지 놓침 |

λ가 너무 작으면 벌점이 약해서 사소한 튐에도 판정을 바꾸고, 너무 크면 벌점이 강해서 진짜 변화에도 판정을 안 바꿔.

**여기서 짚어야 할 점:** 실제 시장 데이터에서는 **정답 판정표를 미리 알 수 없어.** 3일째 같은 날이 일회성 튐인지 진짜 위기의 시작인지는 사후에나 알 수 있지. 그래서 실제로는 "정답 표를 맞히는 λ"를 고를 수 없고, 대신 **여러 λ로 판정표를 만들어 전략을 돌려보고, 개발 구간에서 성과가 가장 좋은 λ**를 골라. "원하는 경로"는 원리를 설명하기 위한 장치였고, 실전의 λ 선택 기준은 아니야.

---

## 5. 실전에서는 매일 어떻게 판정하나

지금까지는 8일치를 **다 본 뒤에** 판정표를 고르는 방식이었어. 그런데 실전에서는 매일 장이 끝날 때마다, **그날까지의 데이터만 보고** 오늘의 레짐을 정해야 해. 7일째 저녁에는 8일째 값을 모르니까.

그래서 매일 이렇게 물어.

> **"오늘까지의 데이터에 대해, 오늘을 평온으로 끝내는 판정표 중 최고 점수와, 오늘을 혼란으로 끝내는 판정표 중 최고 점수를 비교하면 어느 쪽이 더 낮은가?"**

더 낮은 쪽이 오늘까지의 데이터를 더 잘 설명하는 판정이니까, 그걸 오늘의 레짐으로 정해. 앞에서 $V_t(0)$, $V_t(1)$이라고 쓴 게 바로 이 두 점수야.

- $V_t(0)$: 오늘을 **평온으로 끝내는** 모든 판정표 중 가장 낮은 총점
- $V_t(1)$: 오늘을 **혼란으로 끝내는** 모든 판정표 중 가장 낮은 총점

직접 해볼게 (λ = 3).

**3일째 저녁 (지금까지 값: 1, 1, 3)**

| 오늘을 어떻게 끝내나 | 그중 가장 좋은 판정표 | 불일치 | 벌점 | 총점 |
|---|---|---|---|---|
| 평온(0) | 0, 0, 0 | 2 (3일째 값 3을 평온으로) | 0 | **2** |
| 혼란(1) | 0, 0, 1 | 0 | 3 (한 번 바뀜) | **3** |

평온 쪽 점수가 낮아. **오늘은 평온으로 판정.** 튐 하루로는 판정을 바꿀 만한 근거가 부족하다는 뜻이야.

**6일째 저녁 (지금까지 값: 1, 1, 3, 1, 1, 3)**

| 오늘을 어떻게 끝내나 | 그중 가장 좋은 판정표 | 불일치 | 벌점 | 총점 |
|---|---|---|---|---|
| 평온(0) | 0, 0, 0, 0, 0, 0 | 4 (3일째, 6일째) | 0 | **4** |
| 혼란(1) | 0, 0, 0, 0, 0, 1 | 2 (3일째) | 3 | **5** |

아직 평온 쪽이 낮아. **오늘도 평온으로 판정.** 진짜 변화가 시작됐지만 아직 하루치 증거뿐이라서.

**7일째 저녁 (지금까지 값: 1, 1, 3, 1, 1, 3, 3)**

| 오늘을 어떻게 끝내나 | 그중 가장 좋은 판정표 | 불일치 | 벌점 | 총점 |
|---|---|---|---|---|
| 평온(0) | 0, 0, 0, 0, 0, 0, 0 | 6 (3·6·7일째) | 0 | **6** |
| 혼란(1) | 0, 0, 0, 0, 0, 1, 1 | 2 (3일째) | 3 | **5** |

이제 혼란 쪽이 낮아. **오늘은 혼란으로 판정.** 높은 값이 이틀 연속 나오니까, 평온으로 버티는 불일치 점수가 벌점보다 커졌어.

**정리하면:** "가장 싼 경로의 총비용"이 중요한 이유는, **매일 저녁 "평온으로 끝나는 최선의 설명"과 "혼란으로 끝나는 최선의 설명"을 비교해서 오늘의 레짐을 정하는 기준**이기 때문이야. 그리고 이 예시에서 JM은 3일째의 튐은 무시했고, 6일째 시작된 진짜 변화는 **하루 늦은 7일째**에 잡았어. 가짜 경보를 줄이는 대신 감지가 약간 늦어지는 거지.

(날이 수천 일이면 매일 모든 판정표를 다 채점할 수 없어서, 어제의 두 점수만으로 오늘의 두 점수를 계산하는 방법을 써. 그 계산 방법이 앞에서 본 $V_t$ 점화식인데, 하는 일은 위 표와 똑같아.)

---

## 한 줄 요약

**비용**은 판정표가 얼마나 나쁜지를 매기는 점수이고, "데이터와 안 맞는 정도"와 "판정을 바꾼 횟수 × λ"를 더한 거야. **가장 싼 경로**는 이 점수가 가장 낮은 판정표이고, JM은 그 표를 레짐 판정 결과로 채택해. 실전에서는 매일 "오늘을 평온으로 끝내는 최선의 판정표 점수"와 "혼란으로 끝내는 최선의 판정표 점수"를 비교해서 낮은 쪽으로 오늘을 판정해. **원하는 경로**는 설명용 데이터에서 내가 미리 정답으로 정해둔 판정표를 말한 것이고, 실제 데이터에서는 정답을 모르니 λ를 전략 성과로 골라.

| **H3** | H2의 예측력이 평온/혼란 레짐에 따라 달라지나? | 레짐 조건부 |
| **H4** | 한국 공매도 금지 기간(2023.11~2025.3)에 시그널 효과가 달라졌나? (미국을 대조군으로) | 제도 효과 |
| **H5** | 한국과 미국에서 효과의 크기가 다른가? | 국가 비교 |

**우선순위:** H1과 H2가 기본, H4가 차별화 포인트.

## 3. 어떤 방법론으로 하나

| 대상 | 방법 | 핵심 |
|---|---|---|
| **변동성 측정** | 일봉: 고가·저가 기반 변동성 / 장중 봉: 수익률 제곱합 | 장중 봉이 있으면 하락·상승 변동성을 날마다 분리 가능 |
| **레짐 판정** | 지수 일별 데이터로 **JM**(주) 또는 **HMM**(확인용) | 시장 수준, 저/고변동성 2개, **그날까지의 데이터만 사용** |
| **H1** | **HAR → LHAR → 레짐 LHAR** (장중 데이터가 쌓이면 SHAR) | 내일 변동성 예측, 표본 외 예측오차(QLIKE)로 비교 |
| **H2, H3** | **Rank IC**, 분위 포트폴리오, **Fama-MacBeth 회귀**, 이중정렬 | 반전·MAX·IVOL 통제 후에도 살아남나 |
| **H4** | **이중차분(DiD)**: 한국 vs 미국, 공매도 많던 종목 vs 적던 종목 + 가짜 기간 플라시보 | 17개월이라 결론은 조심스럽게 |
| **H5** | 같은 달 두 나라 IC 차이의 평균 검정 | |

## 4. 전략 (연구 결과 → 투자)

| 전략 | 무엇을 정하나 | 근거 |
|---|---|---|
| **S1** | 주식을 **얼마나** 들고 있을지 (레짐 스위치 + 변동성 타기팅) | H1, 레짐 |
| **S2** | **어떤 종목**을 살지 (시그널 상위 20% 롱온리, 월 1회) | H2 |
| **S3** | S2 종목 구성 + S1 비중 조절 | 둘 다 |

**검증 원칙:** 2005~2020년(개발 구간)에서만 규칙을 정하고, 규칙을 동결한 뒤 2021년~현재(테스트 구간)에서 **한 번만** 평가. 거래비용·세금 반영, 바이앤홀드·200일선 규칙·단순 변동성 타기팅과 비교. 통과하면 페이퍼 트레이딩 → 소액 실전.

## 6. 진행 순서

1. **5분봉 수집 시작** + 지수로 HAR vs LHAR
2. 지수 레짐(JM) + 레짐 LHAR + **S1**
3. 종목 데이터 구축
4. 횡단면 분석 (H2, H3)
5. 공매도·한미 비교 (H4, H5) ┐ 병행 (단, S2·S3 규칙 동결 후 H4 본 분석)
6. 종목 선택 전략 (S2, S3)    ┘
7. 장중 데이터로 결과 확인
8. 페이퍼 트레이딩 → 소액 실전
9. 보고서, 코드, 면접용 요약

**인턴 지원용으로는 2단계까지만 제대로 끝내도 완결된 이야기가 돼.**
