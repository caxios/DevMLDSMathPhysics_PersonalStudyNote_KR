3. Signed Jump Variation (부호 있는 점프 변동)

앞 개념에서 $RS^+ - RS^-$가 "점프의 방향"을 드러낸다고 했지. 이번엔 이걸 제대로 파고들 거야. 순서는 점프가 뭔가 → 부호 없는 점프 측정 → 부호 있는 점프 측정 → 상대 지표(RSJ) → 실증 → 함정 → 코드야.

---

## 1단계: "점프"가 정확히 뭔가

주가 움직임을 수식으로 쓰면 보통 이렇게 모형화해.

$$
dp_t = \mu_t\,dt + \sigma_t\,dW_t + \kappa_t\,dN_t
$$

- $p_t$: 로그주가
- $\mu_t\,dt$: 기대수익률(드리프트). 짧은 시간 단위에서는 무시해도 될 만큼 작아.
- $\sigma_t\,dW_t$: **연속 부분(diffusion)**. $W_t$는 브라운 운동이고, 매 순간 작고 무작위로 오르내리는 평상시 움직임이야.
- $\kappa_t\,dN_t$: **점프 부분**. $N_t$는 "이벤트가 몇 번 발생했나"를 세는 카운터(포아송 과정)이고, $\kappa_t$는 발생했을 때의 점프 크기야. 양수면 급등, 음수면 급락이지.

직관적으로 보면:

- 연속 부분은 **호가창이 조금씩 밀리는 평소 움직임**이야.
- 점프는 **실적 서프라이즈, 공시, 속보처럼 가격이 한 번에 불연속적으로 뛰는 움직임**이야.

---

## 2단계: RV는 두 성분의 합을 측정한다

N → ∞일 때 RV가 수렴하는 값을 **이차변동(quadratic variation, QV)**이라고 해.

$$
RV_t \;\to\; QV_t = \underbrace{\int_{t-1}^{t} \sigma_s^2\,ds}_{IV_t:\ \text{연속 부분}} + \underbrace{\sum_{j=1}^{N_t} \kappa_j^2}_{\text{점프 부분}}
$$

RV만 보면 "오늘 변동성이 컸다"는 건 알지만, **평소 움직임이 컸던 건지 점프 한 번이 컸던 건지** 구분이 안 돼. 둘은 경제적 의미가 전혀 달라. 연속 변동성은 꾸준히 이어지는 성질(지속성)이 강하고, 점프는 일회성 이벤트에 가까워. 그래서 이 둘을 분리하려는 연구가 나왔어.

---

## 3단계: 부호 없는 점프 측정 — Bipower Variation

Barndorff-Nielsen & Shephard(2004)는 **점프에 둔감한** 변동성 추정량을 만들었어.

$$
BV_t = \frac{\pi}{2} \sum_{i=2}^{N} |r_{t,i}|\,|r_{t,i-1}|
$$

**핵심 아이디어:** 제곱 $r_i^2$ 대신 **인접한 두 봉의 절댓값 곱**을 써.

- 점프는 드물어서, 점프가 난 봉 옆의 봉은 대개 평범한 크기야.
- 그래서 점프 봉($|r_i|$ 큼)에 옆의 평범한 봉($|r_{i-1}|$ 작음)이 곱해지면, 점프의 영향이 크게 희석돼.
- $\pi/2$는 정규분포에서 $E|Z| = \sqrt{2/\pi}$라서 붙는 보정상수야. $(\sqrt{2/\pi})^2 = 2/\pi$를 상쇄해서 $BV \to IV$가 되게 해줘.

결과적으로 $BV_t \to IV_t$, 즉 **연속 부분만** 측정해. 그러면 점프 부분은 차이로 얻을 수 있어.

$$
JV_t = \max(RV_t - BV_t,\ 0)
$$

**문제:** 이것도 제곱 기반이라 **부호가 없어.** 급등 점프인지 급락 점프인지 모르는 거야. 네 연구 질문("방향")에는 치명적이지.

---

## 4단계: Signed Jump Variation의 정의

여기서 semivariance가 해결책이 돼. 앞 개념의 이론적 결과를 다시 쓰면:

$$
RS_t^{+} \to \frac{1}{2}IV_t + \sum_{\kappa_j > 0} \kappa_j^2, \qquad RS_t^{-} \to \frac{1}{2}IV_t + \sum_{\kappa_j < 0} \kappa_j^2
$$

연속 부분은 양쪽에 **정확히 반반**씩 들어가니까, 빼면 사라져.

$$
\boxed{\Delta J_t = RS_t^{+} - RS_t^{-} \;\to\; \sum_{\kappa_j > 0} \kappa_j^2 - \sum_{\kappa_j < 0} \kappa_j^2}
$$

이게 Patton & Sheppard(2015)가 정의한 **signed jump variation**이야.

- $\Delta J_t > 0$: 오늘은 **급등 점프**가 지배적이었다.
- $\Delta J_t < 0$: 오늘은 **급락 점프**가 지배적이었다.
- $\Delta J_t \approx 0$: 점프가 없었거나, 양방향 점프가 상쇄됐다.

3단계의 BV 방식과 비교하면 장점이 분명해. **BV를 추정할 필요 없이, semivariance 두 개의 차이만으로** 점프의 크기와 방향을 동시에 얻어.

Patton & Sheppard는 이걸 양수 부분과 음수 부분으로 다시 쪼개서도 썼어.

$$
\Delta J_t^{+} = \Delta J_t \cdot \mathbf{1}\{\Delta J_t > 0\}, \qquad \Delta J_t^{-} = \Delta J_t \cdot \mathbf{1}\{\Delta J_t < 0\}
$$

좋은 점프와 나쁜 점프가 미래에 **비대칭적으로** 영향을 주는지 보려는 거야.

---

## 5단계: 상대 지표 — Relative Signed Jump Variation (RSJ)

$\Delta J_t$ 원값은 변동성이 큰 종목일수록 절댓값도 커. 모멘트 표준화 때와 똑같은 문제지. 그래서 Bollerslev, Li, Zhao(2020)는 RV로 나눈 상대 지표를 썼어.

$$
RSJ_t = \frac{RS_t^{+} - RS_t^{-}}{RV_t} = \frac{RS_t^{+} - RS_t^{-}}{RS_t^{+} + RS_t^{-}} \in [-1,\ 1]
$$

- **+1**: 모든 변동이 상승 봉에서 나옴
- **−1**: 모든 변동이 하락 봉에서 나옴
- **0**: 상승과 하락 변동이 대칭

앞에서 본 $RS^-/RV$와는 1:1 관계야.

$$
RSJ_t = 1 - 2\cdot\frac{RS_t^-}{RV_t}
$$

같은 정보를 −1~+1 척도로 바꾼 것일 뿐이야.

---

## 6단계: 숫자 예시

앞에서 쓴 세 종목을 그대로 쓸게.

| 종목 | RS⁺ | RS⁻ | RV | ΔJ | RSJ | RSkew |
|---|---|---|---|---|---|---|
| A (막판 급락) | 0.10 | 1.00 | 1.10 | **−0.90** | **−0.82** | −1.90 |
| B (막판 급등) | 1.00 | 0.10 | 1.10 | **+0.90** | **+0.82** | +1.90 |
| C (고른 등락) | 0.86 | 0.50 | 1.36 | +0.36 | +0.26 | 작음 |

**RSJ와 RSkew의 비교:** 둘 다 비대칭을 잡고 부호도 같이 움직여. 차이는 이거야.

- RSkew는 **3제곱** → 극단 봉 하나에 극도로 민감하고 노이즈가 커.
- RSJ는 **2제곱** + 범위가 [−1, 1]로 제한 → 훨씬 안정적이야.

그래서 같은 "장중 비대칭" 정보를 담으면서 실증에서는 RSJ가 더 선호되는 경향이 있어.

---

## 7단계: 실증 결과

**(1) 미래 변동성 예측 (Patton & Sheppard 2015)**

HAR 모형에 $\Delta J^+$와 $\Delta J^-$를 넣으면 이런 결과가 나와.

- **음의 점프**($\Delta J^-$, 급락)는 미래 변동성을 **크게 높여.**
- **양의 점프**($\Delta J^+$, 급등)는 미래 변동성을 오히려 **낮추는** 방향이었어.

같은 크기의 점프라도 방향에 따라 이후의 불안이 반대로 움직인다는 거야. 앞에서 본 "나쁜 변동성이 더 중요하다"를 점프 수준에서 확인한 결과지.

**(2) 미래 수익률 예측 (Bollerslev, Li, Zhao 2020)**

매주 종목들을 RSJ로 정렬했더니 이런 결과가 나왔어.

- **RSJ가 높은 종목**(급등 점프 위주) → 다음 주 수익률이 **낮았어.**
- **RSJ가 낮은 종목**(급락 점프 위주) → 다음 주 수익률이 **높았어.**
- RSJ 하위를 사고 상위를 파는 롱숏 포트폴리오가 유의한 초과수익을 냈어.

부호가 realized skewness 결과와 같은 방향이라는 걸 봐둬. 해석도 비슷해. 투자자들이 급등 점프에 **과잉반응**하거나 복권형 종목을 **과대평가**하고, 이후 되돌림이 온다는 거지.

**비판적으로 봐야 할 점:** RSJ가 높은 종목은 대체로 **그 주에 이미 많이 오른 종목**이야. 그러니 이 효과의 상당 부분이 그냥 잘 알려진 **단기 반전(short-term reversal)** 효과일 수 있어. 논문은 이를 통제해도 효과가 남는다고 보고하지만, 네가 한국 데이터로 재현할 때는 **전주 수익률을 반드시 통제변수로 넣어야 해.** 안 그러면 "새로운 발견"이 아니라 반전 효과를 다른 이름으로 다시 측정한 것일 수 있어.

---

## 8단계: 네 프로젝트에 쓸 때의 함정

**"점프"라는 이름을 과신하지 마.** 4단계의 이론은 N → ∞에서 성립해. N = 39면 연속 부분이 양쪽에 반반으로 딱 나뉘지 않아서, **점프가 없어도 ΔJ가 꽤 큰 값**을 가질 수 있어. 실무적으로는 "점프의 측정치"라기보다 **"장중 변동의 방향 비대칭 지표"**로 해석하는 게 정직해. 주 단위로 합산하면 이 노이즈가 꽤 줄어.

**한국 시장 특유의 문제: 변동성완화장치(VI).** KRX에서는 가격이 급변하면 VI가 발동돼 약 2분간 단일가 매매로 전환돼. 그러면 10분봉 안에서 인위적인 불연속이 생기고, 그게 "점프"로 잡혀. VI 발동 여부를 표시해두거나, 최소한 이 문제를 연구 한계로 명시해야 해. 오히려 **VI 발동 종목의 이후 수익률**은 그 자체로 흥미로운 하위 연구가 될 수도 있어.

**점프의 상당수는 오버나이트에 일어나.** 실적 발표나 공시는 장 마감 후나 장 시작 전에 몰리는 경우가 많아. 장중 봉만 쓰면 **가장 큰 점프를 통째로 놓쳐.** 오버나이트 수익률을 별도 변수로 두고, 장중 RSJ와 오버나이트 수익률이 각각 어떤 정보를 주는지 분리하는 게 좋아.

**가격제한폭.** 상한가에 닿으면 이후 봉이 0으로 잘려서 $RS^+$가 과소측정돼. 상·하한가 도달일은 따로 처리해야 해.

---

## 계산 코드

```python
import numpy as np
import pandas as pd

def jump_measures(r: pd.Series) -> pd.Series:
    """r: 하루치 10분봉 로그수익률 (첫 봉/오버나이트 제외)"""
    r = r.dropna().values
    rs_pos = np.sum(r[r > 0] ** 2)
    rs_neg = np.sum(r[r < 0] ** 2)
    rv = rs_pos + rs_neg
    # Bipower variation (부호 없는 점프 분리용)
    bv = (np.pi / 2) * np.sum(np.abs(r[1:]) * np.abs(r[:-1]))
    dJ = rs_pos - rs_neg
    return pd.Series({
        'RV': rv,
        'BV': bv,
        'JV': max(rv - bv, 0),              # 부호 없는 점프
        'dJ': dJ,                           # signed jump variation
        'dJ_pos': max(dJ, 0),
        'dJ_neg': min(dJ, 0),
        'RSJ': dJ / rv if rv > 0 else np.nan,
    })

daily = (df.groupby(['ticker', 'date'])['ret10m']
           .apply(jump_measures)
           .unstack())

# 주 단위 RSJ: 합을 먼저 구하고 나서 비율 계산 (일별 비율의 평균이 아님)
weekly = daily.groupby(['ticker', 'week'])[['dJ', 'RV']].sum()
weekly['RSJ_w'] = weekly['dJ'] / weekly['RV']
```

주 단위 RSJ는 **분자와 분모를 각각 합산한 뒤 나누는 게** 맞아. 일별 RSJ를 평균하면 변동성이 거의 없던 날의 극단적인 비율(예: RV가 아주 작은데 RSJ = −1)이 같은 가중치로 들어가서 왜곡돼.

## 네 프로젝트와의 연결

지금까지 배운 세 개념은 사실 **같은 질문에 대한 세 가지 답**이야. 질문은 "장중 변동이 어느 방향으로 비대칭이었나?"이고:

| 지표 | 방식 | 장점 | 약점 |
|---|---|---|---|
| RSkew | 3차 표준화 모멘트 | 이론적으로 직관적 | 노이즈 큼 |
| RS⁺, RS⁻ | 분산을 부호별로 분해 | 안정적, 변동성 예측에 강함 | 원값은 크기 효과 섞임 |
| ΔJ, RSJ | 두 반분산의 차이 | 점프 해석 + 방향 + 범위 제한 | N 작으면 "점프" 해석 약함 |

네 연구 설계에 넣는다면 이렇게 쓰면 돼.

- **변동성 예측 파트**(신호가 강함): HAR에 $RS^+$, $RS^-$, $\Delta J^+$, $\Delta J^-$를 넣고, **계수가 레짐별로 달라지는지** 검정해.
- **수익률 예측 파트**(신호가 약함): 주 단위 RSJ로 종목을 정렬해 다음 주 수익률을 보고, 전주 수익률을 통제한 뒤에도 효과가 남는지, 그게 **레짐에 따라 달라지는지** 봐.

---

**한 줄 요약:** Signed jump variation $\Delta J = RS^+ - RS^-$는 평상시 연속 움직임을 상쇄시키고 **점프의 방향과 크기**만 남긴 지표야. RV로 나눈 RSJ는 [−1, 1] 범위의 안정적인 비대칭 지표가 되고, 급등 점프 위주 종목이 이후 부진하다는 횡단면 결과가 있어. 다만 단기 반전 효과와 겹치는지 반드시 확인해야 해.

다음 4번 **realized semibeta**는 지금까지의 "부호 분해" 아이디어를 **개별 종목과 시장의 공분산(베타)**으로 확장한 거야. 준비되면 말해줘.

---

Qustion : Bipower Variation이 이해가 안 되는데, 수식과 함께 유도과정을 다시 단계적으로 설명해줘, 쉽게.

Bipower Variation, 처음부터 다시 유도하기

3단계가 어려웠던 건 "왜 이웃한 두 봉의 절댓값을 곱하면 점프가 사라지나"와 "π/2는 어디서 나왔나"를 건너뛰었기 때문이야. 이 두 가지를 수식으로 하나씩 유도해볼게.

---

## 0단계: 목표 정리

풀려는 문제는 이거야.

$$
RV_t = \underbrace{IV_t}_{\text{평상시 변동}} + \underbrace{\kappa^2}_{\text{점프}}
$$

RV는 둘을 합쳐서 보여주니까, **점프에는 반응하지 않고 평상시 변동($IV$)만 측정하는 다른 지표**를 만들 수 있다면 빼서 점프를 분리할 수 있어. 그 지표가 BV(bipower variation)야.

---

## 1단계: 평상시(점프 없음) 10분봉 수익률을 수식으로 쓰기

하루 분산을 $\sigma^2$라고 하자(하루 변동성이 σ). 이걸 N개 봉으로 쪼개면 봉 하나의 분산은 $\sigma^2/N$, 표준편차는 $\sigma/\sqrt{N}$이야. 그러면 봉 하나의 수익률은 이렇게 쓸 수 있어.

$$
r_i = \frac{\sigma}{\sqrt{N}}\, z_i, \qquad z_i \sim N(0,1)\ \text{독립}
$$

$z_i$는 표준정규분포를 따르는 "주사위"고, $\sigma/\sqrt{N}$은 그 주사위 눈금의 크기라고 생각하면 돼.

---

## 2단계: 점프가 없을 때 RV가 σ²가 되는 걸 확인

$$
RV = \sum_{i=1}^{N} r_i^2 = \sum_{i=1}^{N} \frac{\sigma^2}{N} z_i^2 = \sigma^2 \cdot \frac{1}{N}\sum_{i=1}^{N} z_i^2
$$

표준정규분포에서 $E[z^2] = 1$(분산이 1이니까)이므로, N이 크면 $\frac{1}{N}\sum z_i^2 \approx 1$이야.

$$
RV \approx \sigma^2 \quad \checkmark
$$

---

## 3단계: 점프가 하나 끼면 RV가 어떻게 되나

k번째 봉에서 크기 κ의 점프가 났다고 하자.

$$
r_k = \frac{\sigma}{\sqrt{N}} z_k + \kappa
$$

이 봉을 제곱하면:

$$
r_k^2 = \frac{\sigma^2}{N}z_k^2 + 2\cdot\frac{\sigma}{\sqrt{N}}z_k\,\kappa + \kappa^2
$$

- 첫째 항은 평상시 성분이라 원래 RV에 들어갈 몫이야.
- 둘째 항은 $\sigma/\sqrt{N}$이 붙어 있어서 N이 커지면 0으로 가.
- **셋째 항 $\kappa^2$는 N과 무관하게 그대로 남아.**

따라서:

$$
RV \approx \sigma^2 + \kappa^2
$$

**RV는 점프를 통째로 흡수해.** 여기까지가 문제 제기야.

---

## 4단계: 아이디어 — "이웃 봉과 곱해서 증거를 요구하자"

제곱 $r_i^2 = r_i \times r_i$는 **자기 자신과 곱하는** 거야. 그래서 한 봉이 크게 튀면 그 크기가 그대로 제곱돼서 들어가.

대신 **바로 옆 봉과 곱하면** 어떨까?

$$
|r_i| \times |r_{i-1}|
$$

비유하자면 이래. 제곱은 "혼자서 주장하면 그대로 인정"이고, 이웃 곱은 "**옆 사람도 같이 커야** 크게 인정"이야. 점프는 한 봉에서만 갑자기 일어나니까 옆 봉이 받쳐주지 않아. 그래서 점프의 영향이 희석돼.

**왜 절댓값을 붙이나?** 그냥 $r_i \times r_{i-1}$을 쓰면 부호가 랜덤이라 +와 −가 섞여 평균이 0이 돼버려. 우리는 "크기"를 재고 싶으니 절댓값을 붙여.

---

## 5단계: 점프가 없을 때 이웃 곱의 기댓값 계산

1단계의 식을 넣으면:

$$
|r_i|\,|r_{i-1}| = \frac{\sigma}{\sqrt{N}}|z_i| \cdot \frac{\sigma}{\sqrt{N}}|z_{i-1}| = \frac{\sigma^2}{N}\,|z_i|\,|z_{i-1}|
$$

$z_i$와 $z_{i-1}$은 독립이니 기댓값은 각각의 곱이야.

$$
E\big[|r_i|\,|r_{i-1}|\big] = \frac{\sigma^2}{N}\,E|z_i|\cdot E|z_{i-1}| = \frac{\sigma^2}{N}\,\big(E|z|\big)^2
$$

이제 $E|z|$ 값만 알면 돼.

---

## 6단계: $E|z|$ 유도 — π/2가 여기서 나와

표준정규분포의 밀도함수는 $\phi(z) = \frac{1}{\sqrt{2\pi}}e^{-z^2/2}$야.

$$
E|z| = \int_{-\infty}^{\infty} |z|\,\phi(z)\,dz
$$

좌우 대칭이니까 양수 쪽만 계산해서 2배 하면 돼.

$$
E|z| = 2\int_{0}^{\infty} z\,\frac{1}{\sqrt{2\pi}}e^{-z^2/2}\,dz
$$

적분 부분만 따로 보자. $u = z^2/2$로 치환하면 $du = z\,dz$니까:

$$
\int_{0}^{\infty} z\,e^{-z^2/2}\,dz = \int_{0}^{\infty} e^{-u}\,du = 1
$$

따라서:

$$
E|z| = \frac{2}{\sqrt{2\pi}} = \sqrt{\frac{2}{\pi}} \approx 0.798
$$

5단계 식에 넣으면:

$$
E\big[|r_i|\,|r_{i-1}|\big] = \frac{\sigma^2}{N}\cdot\frac{2}{\pi}
$$

---

## 7단계: 합하고 보정하면 σ²

이웃 쌍은 $N-1$개인데, N이 크면 N개로 봐도 돼.

$$
\sum_{i=2}^{N}|r_i|\,|r_{i-1}| \approx N\cdot\frac{\sigma^2}{N}\cdot\frac{2}{\pi} = \frac{2}{\pi}\,\sigma^2
$$

$\sigma^2$보다 조금 작게(약 0.64배) 나오지. 그래서 $\pi/2$를 곱해서 되돌려줘.

$$
BV = \frac{\pi}{2}\sum_{i=2}^{N}|r_i|\,|r_{i-1}| \approx \sigma^2 \quad \checkmark
$$

**π/2의 정체:** 절댓값의 평균($\sqrt{2/\pi}$)이 제곱의 평균(1)보다 작아서 생기는 차이를 메우는 **보정상수**일 뿐이야. 신비로운 게 아니야.

---

## 8단계: 점프가 끼면 BV는 어떻게 되나 (핵심)

k번째 봉에 점프 κ가 있으면, κ가 들어가는 이웃 곱은 **딱 두 개**야.

$$
|r_k|\,|r_{k-1}| \quad \text{와} \quad |r_{k+1}|\,|r_k|
$$

각각은 대략 이렇게 돼.

$$
|r_k|\,|r_{k-1}| \approx |\kappa| \times \underbrace{\frac{\sigma}{\sqrt{N}}|z_{k-1}|}_{\text{평범한 옆 봉}}
$$

RV와 나란히 비교해보자.

| | 점프가 더하는 양 | N이 커지면? |
|---|---|---|
| RV | $\kappa^2$ | **그대로** |
| BV | $\dfrac{\pi}{2}\times 2 \times |\kappa| \times \dfrac{\sigma}{\sqrt{N}}|z|$ | **0으로 사라짐** ($\sqrt{N}$이 분모) |

RV에서는 점프가 **점프 × 점프**로 들어가지만, BV에서는 **점프 × 작은 옆 봉**으로 들어가. 옆 봉은 N이 커질수록 작아지니까 결국 점프의 흔적이 사라져.

$$
BV \to \sigma^2 = IV, \qquad RV \to \sigma^2 + \kappa^2
$$

$$
\Rightarrow\quad RV - BV \to \kappa^2
$$

이게 점프의 크기야.

---

## 9단계: 숫자로 체감하기

하루 분산 $\sigma^2 = 1$(하루 변동성 1%), 점프 $\kappa = 3\%$라고 하자. 옆 봉의 크기는 평균적으로 $\frac{\sigma}{\sqrt{N}}\times 0.8$ 정도야.

| N (하루 봉 개수) | 옆 봉 크기 | RV에 섞이는 점프 | BV에 섞이는 점프 ($\pi \times 3 \times$ 옆 봉) |
|---|---|---|---|
| 10 | 0.25 | 9 | 2.38 |
| **39 (10분봉)** | **0.13** | **9** | **1.21** |
| 390 (1분봉) | 0.04 | 9 | 0.38 |

- RV에는 N과 상관없이 점프가 **9만큼** 고스란히 들어가. 평상시 분산 1의 9배야.
- BV에도 점프가 조금 새어 들어가지만, N이 커질수록 줄어들어.
- **10분봉(N = 39)에서는 1.21 정도가 새어 들어가.** 평상시 분산 1보다 커. 즉 10분봉에서는 BV가 점프를 **완전히 걸러내지 못해.** 앞에서 "N = 39면 점프 해석을 과신하지 마라"고 한 이유가 이 숫자야.

---

## 10단계: 점프 변동과 max(·, 0)

$$
JV = \max(RV - BV,\ 0)
$$

`max(·, 0)`을 붙이는 이유는, 유한한 N에서는 우연히 BV가 RV보다 커지는 날이 있기 때문이야. 점프 크기가 음수일 수는 없으니 0으로 잘라.

그리고 이 JV에 **부호가 없다**는 게 3번 개념으로 이어지는 연결고리야. RV도 BV도 제곱이나 절댓값으로 만들어졌으니 급등이든 급락이든 똑같이 양수로 나와. 그래서 방향까지 보려고 semivariance의 차이 $\Delta J = RS^+ - RS^-$를 쓰는 거야.

---

**한 줄 요약:** BV는 봉을 **자기 자신과 곱하는 대신 옆 봉과 곱해서**, 한 봉에만 갑자기 나타나는 점프가 "작은 옆 봉"에 곱해져 희석되게 만든 지표야. $\pi/2$는 절댓값 평균($\sqrt{2/\pi}$)을 보정하는 상수고, 결과적으로 $RV - BV$가 점프 크기가 돼. 다만 10분봉 정도의 N에서는 희석이 불완전해.

이 부분이 이해됐으면 4번 realized semibeta로 넘어가자.
