# 15. 레짐의 정의 (What Is a Regime?)

14번이 레짐을 **"어떻게"** 찾느냐였다면, 15번은 **"무엇을"** 레짐이라고 부르느냐야. 같은 JM이라도 변동성 피처를 넣으면 변동성 레짐이, 상관관계 피처를 넣으면 상관 레짐이 나와. 방법보다 이 정의가 연구 결과를 더 크게 좌우해. 순서는 이래.

1. 레짐의 형식적 정의와 세 가지 조건
2. 무엇이 달라지는가 — 레짐의 종류
3. 왜 변동성 레짐은 잘 잡히고 추세 레짐은 안 잡히나
4. 어느 수준에서 정의하나 (시장/섹터/종목)
5. 시간 척도와 레짐 개수
6. 좋은 레짐 정의의 평가 기준
7. 순환성과 선택 편향
8. 네 프로젝트의 레짐 정의안과 코드

---

## 0단계: 레짐의 형식적 정의

통계적으로 레짐은 **"데이터를 만들어내는 규칙(데이터 생성 과정)이 대략 일정하게 유지되는 구간"**이야.

$$
y_t \sim F_{S_t}, \qquad F_k \neq F_j \quad (k \neq j)
$$

레짐 $S_t = k$일 때 데이터는 분포 $F_k$에서 나오고, 레짐마다 분포가 달라. 여기서 "분포"는 수익률 분포일 수도 있고(8번 HMM), 회귀 관계일 수도 있어(9번 Markov switching).

어떤 구간을 "레짐"이라고 부를 자격이 있으려면 세 가지 조건을 만족해야 해.

**조건 ① 구별성(distinct):** 레짐끼리 분포가 **의미 있게** 달라야 해. 변동성이 1.0%와 1.05%인 두 "레짐"은 레짐이 아니라 노이즈야.

**조건 ② 지속성(persistent):** 한 번 들어가면 **일정 기간 머물러야** 해. 8번에서 유도한 기대 지속기간을 떠올려봐.

$$
E[\text{지속기간}] = \frac{1}{1 - p_{kk}}
$$

$p_{kk} = 0.5$면 평균 2일이야. 이건 레짐이 아니라 매일 동전을 던지는 것과 비슷해.

**조건 ③ 식별 가능성(identifiable):** 데이터로 **어느 레짐인지 알아낼 수 있어야** 해. 특히 실시간으로. 사후적으로만 알 수 있는 레짐은 예측에 쓸 수 없어.

## 1단계: 무엇이 달라지는가 — 레짐의 종류

$F_k$의 **어떤 측면**이 레짐마다 다른지에 따라 레짐의 종류가 갈려.

**(a) 변동성 레짐:** 분산 $\sigma_k^2$가 달라. 피처는 RV, VKOSPI, 하방편차 등이야. **가장 잘 식별되고 가장 많이 쓰여.**

**(b) 추세 레짐 (강세장/약세장):** 평균 $\mu_k$가 달라. 규칙으로 정의하는 경우가 많은데, 대표적인 게 고점 대비 하락률(drawdown)이야.

$$
DD_t = 1 - \frac{P_t}{\max_{s \leq t} P_s}
$$

$DD_t \geq 20\%$면 약세장으로 보는 식이야. 관행적인 기준이지 이론적 근거가 있는 건 아니야.

**(c) 상관 레짐:** 자산 간 상관관계가 달라. 위기 때 모든 종목이 같이 움직이는 현상을 잡아. 대표 지표로 **흡수비율(absorption ratio, Kritzman et al. 2011)**이 있어.

$$
AR_t = \frac{\sum_{i=1}^{m}\sigma^2_{PC_i}}{\sum_{j=1}^{N}\sigma^2_{\text{종목}_j}}
$$

N개 종목 수익률의 전체 분산 중 **상위 m개 주성분이 설명하는 비율**이야. 이 값이 높으면 종목들이 소수의 공통 요인에 휩쓸려 같이 움직이는 상태라는 뜻이야. 4번 semibeta에서 말한 "하락장에 동조화가 강해진다"를 레짐으로 정의하는 방법이지.

**(d) 유동성 레짐:** 거래가 얼마나 쉬운가가 달라. 대표 지표로 **Amihud 비유동성**이 있어.

$$
ILLIQ = \frac{1}{D}\sum_{d=1}^{D}\frac{|r_d|}{\text{거래대금}_d}
$$

거래대금 1원당 가격이 얼마나 움직이나야. 값이 클수록 조금만 거래해도 가격이 크게 흔들리는 비유동 상태지.

**(e) 거시 레짐:** 성장·인플레이션·금리 국면이야. 한국은 원/달러 환율과 외국인 수급이 특히 중요한 변수야.

**(f) 관계 레짐:** 변수 사이의 **관계** $\beta_k$가 달라. 9번 Markov switching 회귀가 찾는 레짐이야.

**(g) 제도적 레짐:** 규칙 자체가 달라. 공매도 금지, 가격제한폭 변경 같은 거야.

## 2단계: 왜 변동성 레짐은 잘 잡히고 추세 레짐은 안 잡히나 (유도)

8번에서 "HMM은 사실상 변동성 레짐을 찾는다"고 했지. 그 이유를 숫자로 보여줄게. 일별 변동성 σ = 1.2%인 시장을 생각하자.

**평균 차이를 확인하는 데 필요한 일수.** 어떤 레짐의 일평균 수익률이 0.1%라는 걸 t-값 2로 확인하려면:

$$
t = \frac{\hat\mu}{\sigma/\sqrt{n}} = 2 \quad\Rightarrow\quad n = \left(\frac{2\sigma}{\mu}\right)^2 = \left(\frac{2 \times 1.2}{0.1}\right)^2 = \mathbf{576\text{일}}
$$

약 **2.3년**이야. 그 레짐이 2.3년 동안 한 번도 안 바뀌고 유지돼야 평균을 겨우 확인할 수 있어.

**분산 차이를 확인하는 데 필요한 일수.** 두 레짐의 변동성이 3배 차이(분산 9배)라고 하자. 정규분포에서 n개 관측치로 추정한 로그분산의 표준오차는 대략 $\sqrt{2/n}$이야(7번에서 $\text{Var}(z^2) = 2$였던 것과 같은 원리). 두 레짐에서 각각 n일씩 관측하면 차이의 표준오차는 $\sqrt{4/n} = 2/\sqrt{n}$이니:

$$
t = \frac{\ln 9}{2/\sqrt{n}} = \frac{2.20\sqrt{n}}{2} = 2 \quad\Rightarrow\quad \sqrt{n} = 1.82 \quad\Rightarrow\quad n \approx \mathbf{4\text{일}}
$$

| 무엇을 식별하나 | 필요한 일수 |
|---|---|
| 평균 차이 (0.1%/일) | **약 576일** |
| 분산 차이 (3배) | **약 4일** |

**100배 이상 차이가 나.** 이게 금융 데이터의 근본적인 성질이야. 일별 수익률에서 **평균은 노이즈에 묻혀 있고, 분산은 뚜렷하게 드러나.** 그래서 통계적 방법에 수익률을 넣으면 거의 무조건 변동성으로 레짐이 갈려. "강세장/약세장 레짐을 찾았다"는 주장은 대개 "고변동/저변동 레짐을 찾았고, 우연히 고변동 쪽 평균이 낮았다"에 가까워.

**네 연구에 대한 함의:** 추세 레짐을 원한다면 통계적 방법보다 **규칙(drawdown 등)**으로 정의하는 게 정직해. 통계적 방법으로는 **변동성 레짐을 목표로 하는 게** 현실적이야.

## 3단계: 어느 수준에서 정의하나

**시장 레짐:** KOSPI 전체의 상태야. 모든 종목에 공통으로 적용돼.
- 장점: 데이터가 깨끗하고(지수는 개별 종목보다 노이즈가 작음), 경제적 해석이 쉽고, 종목 분석과 분리돼서 순환성을 피할 수 있어.
- 단점: 개별 종목의 특수한 상황(예: 이 종목만 실적 쇼크 국면)을 못 잡아.

**섹터 레짐:** 반도체, 2차전지, 바이오처럼 업종별 상태야. 한국은 테마별 쏠림이 심해서 의미가 있을 수 있어.

**종목 레짐:** 종목마다 따로 레짐을 추정해.
- 장점: 가장 세밀해.
- 단점: 노이즈가 커서 추정이 불안정하고, 종목 수백 개에 모형을 따로 맞추면 관리가 어렵고, **종목 레짐과 종목 수익률 사이의 순환성**이 가장 심해.

**2단계 구조도 가능해:** 시장 레짐을 큰 틀로 두고, 그 안에서 종목의 상태(예: 종목 RV가 자기 과거 대비 높은가)를 변수로 쓰는 방식이야. 이건 레짐이라기보다 **"시장 레짐 × 종목 특성" 상호작용**이지만, 실무에서는 이게 가장 다루기 쉬워.

## 4단계: 시간 척도와 레짐 개수

**시간 척도는 예측 기간과 맞아야 해.** 레짐이 예측에 쓸모 있으려면 **레짐 지속기간이 예측 기간보다 충분히 길어야** 해.

- 내일 수익률을 예측하는데 레짐 평균 지속기간이 30일 → 오늘 레짐이 내일도 유지될 가능성이 높으니 쓸모 있어.
- 다음 달 수익률을 예측하는데 레짐 평균 지속기간이 5일 → 오늘 레짐은 한 달 뒤와 거의 무관해.

지속기간 공식으로 확인할 수 있어. 레짐 유지 확률이 $p$일 때 h일 후에도 같은 레짐일 확률은 대략 $p^h$야(도중에 한 번도 안 빠져나가는 경우만 센 하한값). $p = 0.97$이면 1일 후 97%, 20일 후 $0.97^{20} \approx 54\%$야. 월간 예측에 쓰기엔 약하지.

**레짐 개수 K.**

- **K = 2** (평온/혼란): 가장 흔하고 안정적이야.
- **K = 3** (평온/보통/위기): 위기 레짐이 따로 분리돼서 해석이 풍부해지지만, 위기 레짐의 표본이 적어져.

**표본 수를 반드시 계산해봐.** 10년 일별 데이터(약 2,500일)에서 위기 레짐이 전체의 10%라면 약 250일이야. 8단계에서 말한 상호작용 회귀를 하기에 충분하지만, 위기 에피소드 자체는 3~4번뿐일 수 있어. **"관측일 수"와 "독립적인 에피소드 수"는 다르다**는 걸 기억해야 해. 250일이 3개 에피소드에서 나왔다면, 실제 정보량은 250개가 아니라 3개에 가까워.

## 5단계: 좋은 레짐 정의인지 평가하는 기준

레짐을 정의했으면 아래 기준으로 **진단**해야 해. 이 진단표를 논문이나 보고서에 넣으면 설득력이 크게 올라가.

| 기준 | 무엇을 확인하나 | 측정 방법 |
|---|---|---|
| 구별성 | 레짐 간 통계가 확연히 다른가 | 레짐별 평균·변동성·상관, 차이 검정 |
| 지속성 | 충분히 오래 머무는가 | 평균 지속기간, 연간 전환 횟수 |
| 경제적 해석 | 알려진 사건과 맞는가 | 2008 금융위기, 2011 유럽위기, 2020년 3월, 2022년 긴축기와의 일치 |
| 실시간 식별성 | 당시에도 알 수 있었나 | 필터링 레짐과 사후(스무딩) 레짐의 일치율, 감지 지연 일수 |
| 안정성 | 재추정해도 비슷한가 | 롤링 재추정 시 과거 레짐 라벨이 얼마나 바뀌나 |
| 비순환성 | 결과 변수와 독립적인가 | 레짐 피처와 예측 대상 변수의 겹침 여부 |

**실시간 식별성은 이렇게 측정해:**

$$
\text{일치율} = \frac{1}{T}\sum_{t=1}^{T}\mathbf{1}\{\hat{S}_t^{\text{실시간}} = \hat{S}_t^{\text{사후}}\}
$$

일치율이 낮다면, 그 레짐은 **사후적으로만 보이는 레짐**이야. 백테스트에서 좋아 보여도 실전에서는 쓸 수 없어.

## 6단계: 순환성과 선택 편향 (가장 중요한 부분)

지금까지 여러 번 "순환성"을 경고했는데, 여기서 **왜 결과가 왜곡되는지** 정확히 보여줄게.

회귀 $y = \beta x + \varepsilon$을 생각하자. 레짐을 정의한다는 건 결국 **데이터의 일부를 골라내는 것**이야. 이때 무엇을 기준으로 고르느냐에 따라 결과가 달라져.

**x를 기준으로 고르면 → 기울기가 편향되지 않아.**
예를 들어 "x > 0인 날만" 골라서 회귀하면, 관계가 선형인 한 β 추정치는 여전히 편향이 없어. 오차 ε은 x와 무관하니까, x로 골라도 ε의 분포가 안 바뀌어.

**y를 기준으로 고르면 → 기울기가 편향돼.**
"y가 크게 떨어진 날만" 골라서 회귀하면 문제가 생겨. y가 작다는 건 "x가 작았거나, ε이 작았거나" 둘 중 하나야. 그래서 **x가 큰데도 y가 작았던 날**은 ε이 극단적으로 작은 날만 뽑혀. 골라낸 표본 안에서 x와 ε 사이에 **인위적인 음의 상관**이 생기고, β가 0 쪽으로 쪼그라들어(절단 편향, truncation bias).

**네 연구에서 이게 왜 중요한가:**

- 레짐을 **시장 수익률**로 정의(강세/약세)하고 개별 종목 수익률 관계를 검정한다고 해보자. 개별 종목 수익률은 시장 수익률과 강하게 상관돼 있으니, 이건 사실상 **y를 기준으로 고르는 것**에 가까워. 레짐별로 계수가 다르게 나와도 그게 진짜 차이인지 선택 편향인지 구분할 수 없어.
- 레짐을 **시장 변동성**으로 정의하면 상대적으로 안전해. 변동성은 수익률의 **크기**이지 **방향**이 아니니까, 방향을 예측하는 회귀에서는 y를 기준으로 고르는 문제가 훨씬 약해.
- 다만 변동성 예측(HAR)을 할 때는 반대야. 레짐을 시장 RV로 정의하고 종목 RV를 예측하면 다시 y 기준 선택에 가까워져. 이때는 **레짐을 전날까지의 정보로만** 정의해서 적어도 예측 대상 날의 정보가 레짐에 들어가지 않게 해야 해.

**원칙을 한 문장으로 정리하면:** 레짐을 정의하는 데 쓴 정보가 **예측 대상(y)이나 그와 강하게 상관된 변수를 담고 있으면 안 된다.**

## 7단계: 네 프로젝트의 레짐 정의안

지금까지의 논의를 종합한 제안이야. 분석을 시작하기 **전에** 이걸 확정하고 문서로 남겨둬. 결과를 보고 정의를 바꾸면 12번의 선택 편향이 생기니까.

| 우선순위 | 레짐 | 정의 | 선택 이유 |
|---|---|---|---|

그리고 5단계의 진단표를 주 레짐에 대해 반드시 계산해서 보고해.

## 계산 코드: 레짐 진단

```python
import numpy as np
import pandas as pd

def regime_diagnostics(regime: pd.Series, ret: pd.Series, regime_expost: pd.Series = None):
    """
    regime: 실시간(필터링) 레짐 라벨 (0/1), 날짜 인덱스
    ret: 시장 일별 수익률
    regime_expost: 사후(스무딩) 레짐 라벨 (선택)
    """
    df = pd.DataFrame({'s': regime, 'r': ret}).dropna()

    # 구별성: 레짐별 통계
    stats = df.groupby('s')['r'].agg(
        mean='mean', vol='std', n='count',
        skew=lambda x: x.skew(), kurt=lambda x: x.kurt())
    stats['vol_ann'] = stats['vol'] * np.sqrt(252)

    # 지속성: 에피소드 길이
    episode_id = (df['s'] != df['s'].shift()).cumsum()
    episodes = df.groupby(episode_id).agg(s=('s', 'first'), length=('s', 'size'))
    persistence = episodes.groupby('s')['length'].agg(
        avg_duration='mean', n_episodes='count')
    switches_per_year = (df['s'] != df['s'].shift()).sum() / (len(df) / 252)

    out = {'stats': stats, 'persistence': persistence,
           'switches_per_year': switches_per_year}

    # 실시간 식별성: 실시간 vs 사후 일치율
    if regime_expost is not None:
        aligned = pd.concat([regime, regime_expost], axis=1).dropna()
        out['realtime_agreement'] = (aligned.iloc[:, 0] == aligned.iloc[:, 1]).mean()

    return out

# 경제적 해석: 알려진 사건 구간에서 고변동 레짐 비율
events = {'GFC': ('2008-09-01', '2009-03-31'),
          'EU crisis': ('2011-08-01', '2011-12-31'),
          'COVID': ('2020-02-20', '2020-04-30'),
          'Tightening': ('2022-01-01', '2022-10-31')}
for name, (a, b) in events.items():
    print(name, regime.loc[a:b].mean().round(2))   # 1에 가까워야 해석이 맞음
```

**한 줄 요약:** 레짐은 "데이터 생성 과정이 대략 일정한 구간"이고, **구별성·지속성·식별 가능성**을 갖춰야 해. 일별 수익률에서 분산 차이는 며칠이면 식별되지만 평균 차이는 수년이 걸리기 때문에, 통계적 레짐은 현실적으로 **변동성 레짐**이야. 가장 중요한 원칙은 **레짐을 정의하는 정보가 예측 대상과 겹치면 안 된다**는 거야. 그렇지 않으면 선택 편향으로 레짐별 차이가 인위적으로 만들어져.
