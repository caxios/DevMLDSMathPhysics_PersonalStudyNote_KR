# 17. Rank IC (순위 정보계수)

3번 RSJ나 4번 semibeta 같은 시그널을 만들었다면, 그다음 질문은 **"이 시그널이 미래 수익률을 얼마나 잘 맞히나?"**야. Rank IC는 그걸 재는 업계 표준 지표야. 순서는 이래.

1. 횡단면 예측이 뭔가
2. IC의 정의 (피어슨 상관)
3. Rank IC의 정의 (스피어만 상관)와 숫자 예시
4. 왜 순위를 쓰나 (이상치 예시)
5. IC의 시계열: 평균, ICIR, t-통계량
6. IC 크기의 감각: 우연이면 얼마나 나오나
7. 적극적 운용의 기본법칙
8. IC 감소(decay)와 중립화
9. 레짐별 IC 비교 (네 프로젝트의 핵심 검정)
10. 함정과 코드

---

## 0단계: 횡단면 예측이란

예측에는 두 종류가 있어.

- **시계열 예측:** "삼성전자가 **내일 오를까 내릴까?**" 한 종목의 시간에 따른 움직임을 맞히는 거야. 처음 대화에서 네가 원래 하려던 게 이거였어.
- **횡단면(cross-sectional) 예측:** "오늘 기준으로 800개 종목 중 **어느 종목이 다른 종목보다 더 오를까?**" 같은 시점의 종목들 사이의 **상대적 순서**를 맞히는 거야.

3번의 Bollerslev-Li-Zhao, 1번의 Amaya et al. 결과가 전부 횡단면 결과였지. 퀀트 운용에서 롱숏 전략이 주로 이쪽이야. 시장 전체가 오르든 내리든, "잘 될 종목을 사고 안 될 종목을 팔면" 수익이 나니까.

Rank IC는 **횡단면 예측력**을 재는 지표야.

---

## 1단계: IC (Information Coefficient)

매 시점 t마다, 모든 종목에 대해 두 개의 숫자가 있어.

- $x_{i,t}$: t시점에 계산한 종목 i의 **시그널** (예: 이번 주 RSJ)
- $y_{i,t+1}$: 종목 i의 **다음 기간 수익률** (예: 다음 주 수익률)

**IC는 이 둘의 횡단면 상관계수**야. 종목들을 가로질러 계산해.

$$
IC_t = \text{Corr}_i\big(x_{i,t},\ y_{i,t+1}\big) = \frac{\sum_i (x_{i,t} - \bar{x}_t)(y_{i,t+1} - \bar{y}_{t+1})}{\sqrt{\sum_i (x_{i,t} - \bar{x}_t)^2}\sqrt{\sum_i (y_{i,t+1} - \bar{y}_{t+1})^2}}
$$

5번에서 배운 공분산을 두 변수의 표준편차로 나눈 **피어슨 상관계수**야. 모멘트 표준화와 같은 원리로 −1~+1 사이 값이 나와.

- $IC = +1$: 시그널 순서대로 수익률이 완벽하게 정렬됨
- $IC = 0$: 시그널과 수익률이 무관
- $IC = -1$: 시그널과 **반대** 순서로 완벽하게 정렬됨

**음수 IC도 똑같이 유용해.** 부호를 뒤집어서 쓰면 되니까. 3번에서 RSJ가 높은 종목이 이후 부진했지. 그건 RSJ의 IC가 음수라는 뜻이고, "RSJ가 낮은 종목을 사라"는 시그널로 쓰면 돼.

---

## 2단계: Rank IC — 값 대신 순위로

Rank IC는 시그널과 수익률을 **각각 순위로 바꾼 다음** 상관계수를 구해.

$$
\text{Rank } IC_t = \text{Corr}_i\big(\text{rank}(x_{i,t}),\ \text{rank}(y_{i,t+1})\big)
$$

통계학에서는 이걸 **스피어만 순위상관계수(Spearman's ρ)**라고 불러. 동점이 없을 때는 간단한 공식으로도 계산돼.

$$
\rho = 1 - \frac{6\sum_i d_i^2}{n(n^2 - 1)}, \qquad d_i = \text{rank}(x_i) - \text{rank}(y_i)
$$

$d_i$는 각 종목의 **시그널 순위와 수익률 순위의 차이**야. 순위가 완벽히 같으면 모든 $d_i = 0$이라 ρ = 1이 되고, 순위가 어긋날수록 $\sum d_i^2$가 커져서 ρ가 작아져.

### 숫자 예시

5개 종목의 이번 주 RSJ와 다음 주 수익률이야.

| 종목 | RSJ (시그널) | 다음 주 수익률 | 시그널 순위 | 수익률 순위 | $d$ | $d^2$ |
|---|---|---|---|---|---|---|
| A | +0.6 | −1.2% | 5 | 1 | 4 | 16 |
| B | +0.2 | +0.5% | 4 | 3 | 1 | 1 |
| C | −0.1 | −0.3% | 3 | 2 | 1 | 1 |
| D | −0.4 | +1.8% | 2 | 5 | −3 | 9 |
| E | −0.7 | +0.9% | 1 | 4 | −3 | 9 |

(순위는 작은 값부터 1위)

$$
\rho = 1 - \frac{6 \times 36}{5 \times (25 - 1)} = 1 - \frac{216}{120} = \mathbf{-0.8}
$$

Rank IC = −0.8이야. **RSJ가 높을수록 다음 주 수익률 순위가 낮다**는 강한 음의 관계지. 3번의 실증 결과와 같은 방향이야. (실제 데이터에서 −0.8은 나올 수 없는 수준이야. 6단계에서 현실적인 크기를 다룰게.)

## 3단계: 왜 굳이 순위로 바꾸나

**이유 1: 이상치에 강해.**

위 예시에서 D의 수익률이 +1.8%가 아니라 **+15%**(호재 공시로 급등)였다고 해보자. D는 여전히 수익률 1위라 **순위는 전혀 안 바뀌어.**

| | 원래 (D = +1.8%) | 이상치 (D = +15%) |
|---|---|---|
| 피어슨 IC | −0.78 | **−0.44** |
| Rank IC | −0.80 | **−0.80** |

피어슨 IC는 종목 하나의 급등으로 **절반 가까이 약해졌어.** 1번에서 3제곱이 극단값에 민감했던 것처럼, 피어슨 상관도 제곱 기반이라 극단값 하나에 휘둘려. 한국 시장, 특히 코스닥에는 상한가와 테마 급등이 흔해서 이 문제가 심각해.

**이유 2: 시그널의 단위가 상관없어.**

RSJ는 −1~+1이고, semibeta는 0~3 정도고, IVOL은 0.01~0.1 정도야. 순위로 바꾸면 전부 1~N이 되니까 **서로 다른 시그널을 같은 척도로 비교**할 수 있어.

**이유 3: 실제 운용 방식과 맞아.**

롱숏 전략은 보통 "시그널 상위 10%를 사고 하위 10%를 판다"처럼 **순위**로 종목을 골라. 시그널 값이 0.61이냐 0.62냐는 중요하지 않고, 순서가 중요해. 그러니 평가도 순서로 하는 게 자연스러워.

**대가:** 순위는 **"얼마나"**의 정보를 버려. 극단값 종목에서 수익이 크게 나는 시그널이라면 Rank IC가 그 강점을 과소평가할 수 있어. 그래서 보통 Rank IC와 분위 포트폴리오 수익률(8단계)을 함께 보고해.

## 4단계: IC의 시계열 — 한 번의 IC는 의미가 없다

IC는 **매 시점마다** 하나씩 나와. 10년 동안 매주 계산하면 약 520개의 IC 값이 생겨.

$$
IC_1,\ IC_2,\ \dots,\ IC_T
$$

이 시계열로 세 가지를 계산해.

**평균 IC:** 시그널의 평균적인 예측력

$$
\overline{IC} = \frac{1}{T}\sum_{t=1}^{T} IC_t
$$

**ICIR (IC Information Ratio):** 예측력이 얼마나 **꾸준한가**

$$
ICIR = \frac{\overline{IC}}{\text{sd}(IC_t)}
$$

평균 IC가 같아도, 매주 0.03씩 꾸준한 시그널과 어떤 주는 +0.3이고 어떤 주는 −0.25인 시그널은 완전히 달라. ICIR은 샤프 비율과 같은 구조야. 평균을 변동성으로 나눈 거지.

**t-통계량:** 평균 IC가 0과 통계적으로 다른가

$$
t = \frac{\overline{IC}}{\text{sd}(IC_t)/\sqrt{T}} = ICIR \times \sqrt{T}
$$

**숫자 예시:** 주간 평균 IC 0.03, 표준편차 0.10, 10년(T = 520주)

- ICIR = 0.03 / 0.10 = 0.30 (주간)
- t = 0.30 × √520 ≈ **6.8**
- 연율화 ICIR ≈ 0.30 × √52 ≈ 2.2

평균 IC 0.03이 작아 보여도, **꾸준하고 표본이 길면** 통계적으로 매우 강한 결과야.

## 5단계: IC 크기의 감각 — 우연이면 얼마나 나오나

시그널에 예측력이 **전혀 없어도** 매주 IC는 0이 아니라 우연히 흔들려. 얼마나 흔들릴까?

종목 수가 N개일 때, 두 변수가 서로 무관하면 표본 상관계수의 표준오차는 대략 이래.

$$
\text{SE}(IC_t) \approx \frac{1}{\sqrt{N - 1}}
$$

| 종목 수 N | 우연에 의한 IC 표준오차 |
|---|---|
| 30 | 0.19 |
| 200 (KOSPI200) | 0.07 |
| 800 | 0.035 |
| 2,000 (전 종목) | 0.022 |

**어떤 한 주의 IC가 0.05라는 건, 800종목 기준으로 우연히도 충분히 나올 수 있는 값이야.** 그래서 한 기간의 IC로는 아무것도 판단할 수 없고, 반드시 **여러 기간의 평균**으로 봐야 해.

**현실적인 IC 크기:** 실무에서 주간·월간 횡단면 시그널의 평균 Rank IC는 대략 **0.02~0.05면 쓸 만하고, 0.05를 넘으면 강한 편**으로 봐. 0.1 이상이 나오면 기뻐하기 전에 **look-ahead나 데이터 오류를 먼저 의심**해야 해. 16번 체크리스트를 다시 돌려봐.

(현실에서 매주 IC의 표준편차가 위 표의 값보다 큰 이유는, 종목들이 같은 날 같은 시장 충격을 받아서 서로 독립이 아니기 때문이야.)

## 6단계: 적극적 운용의 기본법칙 — 작은 IC가 왜 중요한가

Grinold(1989)의 **적극적 운용의 기본법칙(Fundamental Law of Active Management)**은 IC와 전략 성과를 이렇게 연결해.

$$
IR \approx IC \times \sqrt{BR}
$$

- $IR$: 정보비율 (벤치마크 대비 초과수익 ÷ 추적오차, 샤프 비율과 비슷한 개념)
- $BR$ (breadth): 1년 동안 하는 **독립적인 베팅의 수**

**직관:** 한 번의 예측은 약해도(IC 작음), **많은 종목에 대해 자주** 예측하면 운이 평균으로 상쇄되고 실력이 드러나. 동전 던지기에서 앞면 확률이 51%인 동전도, 한 번 던지면 거의 운이지만 만 번 던지면 확실히 이기는 것과 같아.

그래서 퀀트는 IC 0.03짜리 시그널로도 전략을 만들어. 수백 종목에 매주 적용하니까.

**하지만 과신하면 안 돼.** 이 공식은 베팅들이 **서로 독립**이라고 가정하는데, 실제로는 종목들이 같은 시장 충격을 공유해서 독립적인 베팅 수가 종목 수보다 훨씬 적어. 거래비용, 공매도 제약도 반영하지 않아. **방향을 이해하는 직관**으로만 쓰고, 실제 성과는 백테스트로 확인해야 해.

## 7단계: IC 감소(Decay)와 중립화

**IC decay: 시그널이 얼마나 오래 유효한가**

같은 시그널로 여러 예측 기간의 IC를 계산해봐.

$$
IC^{(h)}_t = \text{RankCorr}_i\big(x_{i,t},\ r_{i,\,t+1 \to t+h}\big), \quad h = 1, 5, 10, 20\text{일}
$$

| 예측 기간 h | Rank IC (가상의 예) |
|---|---|
| 1일 | 0.040 |
| 5일 | 0.030 |
| 10일 | 0.015 |
| 20일 | 0.005 |

이런 식으로 빨리 사라지는 시그널은 **자주 리밸런싱**해야 하고 그만큼 **거래비용**이 커. 장중 데이터로 만든 시그널(RSJ, RSkew)은 대체로 빨리 감소하는 편이야. 반대로 IVOL처럼 느리게 감소하는 시그널은 월간 리밸런싱으로도 충분해.

**주의:** h일 수익률을 매일 계산하면 인접한 날들의 예측 기간이 겹쳐서(12번의 누수 문제와 같은 구조) IC 시계열에 자기상관이 생겨. t-통계량을 계산할 때 **Newey-West 표준오차**를 쓰거나, **겹치지 않는 기간**(예: 5일 수익률이면 5일마다)만 써야 해.

**중립화: 새로운 정보인가, 알려진 효과의 재포장인가**

3번과 5번에서 경고했지. RSJ가 높은 종목은 대개 이번 주에 이미 많이 오른 종목이고, IVOL이 높은 종목은 대개 소형주야. 그러면 RSJ의 IC가 사실은 **단기 반전 효과의 IC**일 수 있어.

해결책은 **시그널에서 알려진 요인을 빼고** IC를 계산하는 거야. 매 시점 횡단면 회귀를 돌려.

$$
\text{rank}(x_{i,t}) = a_t + b_t\,\text{rank}(\text{전주수익률}_{i,t}) + c_t\,\text{rank}(\text{시가총액}_{i,t}) + \text{업종더미} + u_{i,t}
$$

잔차 $u_{i,t}$는 "전주 수익률, 규모, 업종으로 설명되지 **않는** 순수한 RSJ 성분"이야. 이 잔차의 Rank IC가 여전히 유의하면, 그때 비로소 "RSJ에 **새로운** 정보가 있다"고 말할 수 있어.

## 8단계: Rank IC와 분위 포트폴리오의 관계

| | Rank IC | 분위(5분위, 10분위) 롱숏 스프레드 |
|---|---|---|
| 보는 범위 | **전체** 종목의 순서 | **양 끝** 종목만 |
| 단위 | 상관계수 (무차원) | 수익률 (%) |
| 강점 | 안정적, 시그널 간 비교 쉬움 | 실제 돈으로 해석 가능 |
| 약점 | 수익 크기를 모름 | 양 끝 몇 종목에 결과가 좌우될 수 있음 |

둘은 **보완 관계**라서 함께 보고해. 흔한 불일치 사례도 알아둬.

- **IC는 양수인데 롱숏 수익이 없음:** 순서는 전반적으로 맞는데 양 끝에서 안 맞는 경우야.
- **롱숏 수익은 큰데 IC는 작음:** 극단 종목 몇 개에서만 효과가 나는 경우야. 한국에서는 이게 **소형 급등주 몇 개의 효과**일 가능성이 있어서 5번에서 말한 필터링이 중요해.
- **둘 다 좋은데 실전에서 안 됨:** 거래비용이나 공매도 제약이야. 특히 숏 쪽 종목이 공매도 불가능한 소형주라면 롱숏 스프레드는 달성 불가능한 숫자야.

---

## 9단계: 레짐별 IC 비교 — 네 프로젝트의 핵심 검정

지금까지의 모든 개념이 여기서 합쳐져. 네 원래 질문 "레짐별로 시그널이 다른가?"를 횡단면 버전으로 쓰면 이렇게 돼.

**"평온 레짐과 혼란 레짐에서 RSJ의 Rank IC가 다른가?"**

주간 IC 시계열 $IC_t$와 16번의 **online** 레짐 더미 $D_{t}$(t시점까지의 정보로 판단한 혼란 레짐 여부)를 가지고 이 회귀를 돌려.

$$
IC_t = a + b\,D_t + e_t
$$

- $a$: 평온 레짐의 평균 IC
- $a + b$: 혼란 레짐의 평균 IC
- **$b$: 두 레짐의 IC 차이** → $H_0: b = 0$ 검정이 곧 연구 질문의 답이야.

14번에서 본 상호작용항 δ와 같은 발상인데, IC 시계열 수준에서 하는 거라 훨씬 단순하고 해석이 명확해. 표준오차는 Newey-West로 계산해.

**주의할 점 세 가지를 복습하면:**

- $D_t$는 반드시 **online 레짐**이어야 해(16번). 스무딩 레짐을 쓰면 레짐 판단에 예측 대상 기간의 수익률이 들어가서, 15번의 "y 기준 선택 편향"이 생겨.
- 혼란 레짐 주가 50주라면 그 평균 IC의 표준오차가 커. 15번에서 말했듯 **독립적인 에피소드 수**가 적다는 점을 보고해야 해.
- 공매도 금지 더미도 같은 방식으로 넣을 수 있어: $IC_t = a + b\,D^{\text{금지}}_t + e_t$. 5번의 Stambaugh-Yu-Yuan 논리대로라면, IVOL이나 RSJ 같은 "고평가 관련" 시그널의 IC가 금지 기간에 **더 강해졌을** 거라고 예측할 수 있어.

## 10단계: 함정

**동점(ties).** 수익률이 정확히 0인 종목(거래정지, 거래 없음), 상한가에 묶인 종목들은 같은 값을 가져. 순위를 매길 때 **평균 순위**를 줘야 해(pandas `rank(method='average')`가 기본값이야). 동점이 너무 많으면 IC가 왜곡되니 거래정지 종목은 제외해.

**생존편향.** 상장폐지된 종목을 데이터에서 빼면, **망한 종목의 마지막 수익률**이 사라져. 하필 IVOL이나 RSJ 같은 시그널은 부실 종목을 골라내는 경향이 있어서, 상폐 종목을 빼면 IC가 체계적으로 왜곡돼. 상폐 직전까지의 수익률을 포함해야 해.

**투자 가능 종목 집합(universe).** 관리종목, 투자경고 종목, 초저가주를 포함하느냐에 따라 IC가 크게 달라져. universe 정의를 명확히 하고, 결과가 universe에 민감한지 확인해.

**시그널과 수익률 시점.** 16번 6단계와 똑같아. 금요일 장 마감까지의 10분봉으로 만든 주간 RSJ는 **다음 주 월요일 이후** 수익률과 비교해야 해. 금요일 종가 수익률이 예측 대상에 들어가면 안 돼.
