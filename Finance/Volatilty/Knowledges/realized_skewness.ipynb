1. Realized Skewness (실현 왜도)

아래 5단계로 쌓아 올릴게. 수익률 → 왜도라는 개념 → 실현변동성 → 실현왜도 공식 → 해석과 한계 순서야.

---

## 1단계: 장중 수익률 정의

하루(t일)의 장중 가격을 10분 간격으로 N번 관측한다고 하자. 한국 정규장은 09:00~15:30(390분)이니 N ≈ 39야.

i번째 10분봉의 로그수익률은 다음과 같아.

$$
r_{t,i} = \ln P_{t,i} - \ln P_{t,i-1}, \quad i = 1, \dots, N
$$

로그수익률을 쓰는 이유는 **더하면 하루 수익률이 되기 때문**이야.

$$
r_t = \sum_{i=1}^{N} r_{t,i}
$$

이제 하루 하나의 수익률 $r_t$ 대신, 그 하루 **안에서 39개의 수익률이 어떻게 분포했는지**를 볼 수 있게 됐어. realized 계열 지표는 전부 이 39개 조각을 가지고 만드는 거야.

---

## 2단계: 왜도(skewness)가 원래 뭔지

확률변수 X의 왜도는 **3차 표준화 모멘트**야.

$$
\text{Skew}(X) = \frac{E[(X-\mu)^3]}{\sigma^3}
$$

- 분자의 **3제곱**은 부호를 살려둬. 평균보다 크게 튄 값은 양수로, 크게 떨어진 값은 음수로 크게 기여해. 게다가 3제곱이라 **큰 값일수록 압도적으로** 기여해.
- 분모의 $\sigma^3$는 단위를 없애기 위한 정규화야. 변동성이 큰 주식과 작은 주식을 같은 척도로 비교할 수 있게 해줘.

해석은 이렇게 해.

| 왜도 | 분포 모양 | 직관 |
|---|---|---|
| > 0 | 오른쪽 꼬리가 김 | 대부분 조금씩 잃다가 가끔 크게 오름 (복권형) |
| < 0 | 왼쪽 꼬리가 김 | 대부분 조금씩 벌다가 가끔 크게 폭락 |
| = 0 | 좌우 대칭 | 정규분포 |

---

## 3단계: 실현분산(Realized Variance)

실현왜도의 분모에 들어가니까 먼저 짚고 갈게. 하루의 실현분산은 장중 수익률 제곱의 합이야.

$$
RV_t = \sum_{i=1}^{N} r_{t,i}^2
$$

평균을 빼지 않는 이유는, 10분 단위 기대수익률은 사실상 0이라 빼나 안 빼나 차이가 무시할 만하기 때문이야. 이론적으로는 N → ∞일 때 $RV_t$가 그날의 **진짜 분산**(quadratic variation)에 수렴해. 이게 ABDL과 Barndorff-Nielsen-Shephard의 핵심 결과야.

---

## 4단계: 실현왜도 공식

Amaya, Christoffersen, Jacobs, Vasquez(2015)의 정의는 다음과 같아.

$$
RSkew_t = \frac{\sqrt{N} \, \sum_{i=1}^{N} r_{t,i}^3}{RV_t^{3/2}}
$$

구성요소를 하나씩 뜯어보자.

- **분자 $\sum r_{t,i}^3$**: 2단계의 $E[(X-\mu)^3]$에 해당해. 장중 수익률 세제곱의 합이야.
- **분모 $RV_t^{3/2}$**: 2단계의 $\sigma^3$에 해당해. $RV = \sigma^2$이니 $RV^{3/2} = \sigma^3$이야.
- **$\sqrt{N}$은 왜 곱할까?** 여기가 제일 헷갈리는 부분이니 유도해볼게.

하루 변동성을 σ라 하면, 10분봉 하나의 크기는 대략 $\sigma/\sqrt{N}$ 수준이야(분산이 N개 조각에 나눠지니까). 10분봉 수익률 분포의 왜도를 s라고 하면:

$$
\sum_{i=1}^{N} r_{t,i}^3 \approx N \cdot s \cdot \left(\frac{\sigma}{\sqrt{N}}\right)^3 = \frac{s\,\sigma^3}{\sqrt{N}}
$$

$$
\frac{\sum r_{t,i}^3}{RV_t^{3/2}} \approx \frac{s\,\sigma^3/\sqrt{N}}{\sigma^3} = \frac{s}{\sqrt{N}}
$$

즉 $\sqrt{N}$을 안 곱하면 **N이 달라질 때 값이 바뀌어**. 5분봉과 10분봉 결과를 비교할 수 없게 되는 거지. $\sqrt{N}$을 곱해야 $RSkew_t \approx s$가 되어서, 샘플링 빈도와 무관하게 **장중 수익률 분포의 왜도** 그 자체를 측정하게 돼.

## 5단계: 숫자 예시

설명을 단순하게 하려고 N = 5라고 하자. 두 종목의 하루를 비교할 거야(단위 %).

**종목 A (조금씩 오르다 막판 급락):** $r = [0.2,\ 0.1,\ 0.1,\ 0.2,\ -1.0]$

- $RV = 0.04 + 0.01 + 0.01 + 0.04 + 1.00 = 1.10$
- $\sum r^3 = 0.008 + 0.001 + 0.001 + 0.008 - 1.000 = -0.982$
- $RV^{3/2} = 1.10^{1.5} \approx 1.154$
- $RSkew = \sqrt{5} \times (-0.982 / 1.154) \approx 2.236 \times (-0.851) \approx \mathbf{-1.90}$

**종목 B (조금씩 내리다 막판 급등):** $r = [-0.2,\ -0.1,\ -0.1,\ -0.2,\ +1.0]$

- RV는 똑같이 1.10이야. 제곱하면 부호가 사라지니까.
- 세제곱합은 부호만 뒤집혀 +0.982
- $RSkew \approx \mathbf{+1.90}$

**핵심 포인트:** 두 종목의 **변동성(RV)은 완전히 같아.** 변동성만 보면 두 날을 구분할 수 없어. 그런데 실현왜도는 "그 변동성이 **어느 방향의 큰 움직임**에서 나왔는가"를 잡아내. 네가 원래 하려던 "변동성으로 방향을 본다"는 아이디어의 한계를 보완하는 게 바로 이 지점이야.

## 6단계: 실증 결과 — 이게 왜 중요한가

Amaya et al.(2015)의 핵심 발견은 이거야. **실현왜도가 높은 주식은 다음 주 수익률이 낮다.** 미국 주식에서 매주 RSkew 하위 10% 종목을 사고 상위 10% 종목을 팔면, 주당 약 0.19%의 유의미한 초과수익이 났어.

직관적으로는 이상하지. 급등한 흔적이 있는 주식이 오히려 이후 부진하다는 거니까. 주된 해석은 **복권 선호(lottery preference)**야. 투자자들이 양(+)의 왜도를 가진 주식을 복권처럼 과대평가해 비싸게 사고, 그래서 이후 기대수익률이 낮아진다는 거지.

**주의:** 이건 **횡단면(cross-section)** 결과야. "RSkew가 높은 종목이 낮은 종목보다 덜 오른다"는 **상대 비교**이지, "RSkew가 높으면 내일 이 종목이 하락한다"는 **절대 방향 예측**이 아니야. 이 둘을 혼동하면 연구 설계가 틀어져.

## 7단계: 네 프로젝트에 쓸 때의 함정

**N = 39는 작아.** 3제곱은 큰 값 하나에 극도로 민감해서, 10분봉 하나가 튀면 값이 거의 전부 그 봉으로 결정돼. 이론상 $|RSkew_t|$의 최댓값은 $\sqrt{N}$이고(봉 하나만 움직인 경우), N = 39면 약 6.2야. 일별 값은 노이즈가 너무 커서 보통 **주 단위로 평균**해서 써.

$$
RSkew_{week} = \frac{1}{5}\sum_{d=1}^{5} RSkew_d
$$

**첫 봉 문제.** 09:00~09:10 봉에는 시가 동시호가와 밤사이 정보가 섞여 있어. 이 봉을 포함하면 왜도가 그 봉에 지배돼. 오버나이트 수익률(전일 종가 → 당일 시가)은 제외하고 장중만 쓰는 게 표준이야.

**가격제한폭.** 상·하한가에 닿으면 이후 봉의 수익률이 0으로 잘려서 왜도가 왜곡돼. 해당 종목·일자는 제외하거나 따로 표시해야 해.

**저가주 호가단위.** 1,000원짜리 주식은 1틱 움직임이 곧 큰 %변동이라 수익률이 이산적으로 튀어. 가격이나 유동성 필터가 필요해.

**다른 변수와 겹침.** RSkew는 그날 수익률 자체나 단기 반전 효과와 상관이 있어. 회귀할 때 전주 수익률을 통제변수로 넣어야 "진짜 왜도 효과"인지 구분돼.

Qustion : 그러니까 realized라고 하는 건 실제 데이터를 갖다가 계산한다는 거야? 즉, historical 데이터를 쓴다는 거야?

반은 맞고 반은 틀려. 과거 데이터를 쓰는 건 맞지만, "realized"의 핵심은 **historical과는 다른 개념**이야.

## "Realized"의 정확한 의미

"realized"는 **사후적(ex-post)으로, 모형 없이, 그 기간 안의 고빈도 데이터로 그 기간 자체의 값을 직접 측정한다**는 뜻이야. 대비되는 개념들과 나란히 놓으면 차이가 분명해져.

| 구분 | 사용 데이터 | 측정 대상 | 시점 |
|---|---|---|---|
| **Realized** | **그날 안의** 장중 수익률 (10분봉 39개) | **그날 하루**의 변동성 | 사후 |
| Historical | **과거 여러 날의** 일별 수익률 (예: 20일) | 그 구간 **평균적인** 변동성 | 사후 |
| Model-based (GARCH 등) | 과거 일별 수익률 + 모형 가정 | 모형이 추정한 조건부 변동성 | 사전 추정 |
| Implied | 옵션 가격 | 시장이 **예상하는** 미래 변동성 | 사전 |

## Historical과의 차이 (여기가 핵심)

**Historical volatility:** 과거 20일의 일별 수익률로 표준편차 하나를 계산해.

$$
\hat\sigma_{hist} = \sqrt{\frac{1}{19}\sum_{d=t-19}^{t} (r_d - \bar r)^2}
$$

여기에는 **"20일 동안 변동성이 일정했다"**는 가정이 숨어 있어. 20일 중 하루가 폭락장이었어도 그 하루의 변동성은 알 수 없고, 20일치에 평균으로 뭉개져.

**Realized variance:** t일 **하루 안의** 39개 봉만으로 그날 하나의 값을 계산해.

$$
RV_t = \sum_{i=1}^{39} r_{t,i}^2
$$

그래서 **매일 하나씩** 서로 다른 변동성 값이 나와. 월요일은 조용했고 화요일은 난리였다는 게 숫자로 구분되는 거야.

## 왜 굳이 "realized"라는 이름을 붙였나

원래 변동성은 **관측 불가능한 잠재변수**였어. 주가를 봐도 "오늘의 진짜 σ"는 보이지 않고, GARCH 같은 모형으로 **추정**만 할 수 있었지.

ABDL과 Barndorff-Nielsen-Shephard의 공헌은 **장중 데이터를 충분히 잘게 쪼개면(N → ∞) $RV_t$가 그날의 진짜 분산에 수렴한다**는 걸 보인 거야. 즉 변동성을 "추정할 대상"에서 **"관측 가능한 데이터"로 바꿔놓은 것**이지. "이미 실현된(realized) 값을 측정한다"는 뜻이 그래서 붙었어.

실무적으로는 이게 큰 차이를 만들어. RV를 **종속변수**로 놓고 HAR 회귀를 돌리거나, RV를 **설명변수**로 놓고 다음 날 수익률을 예측할 수 있게 돼. 변동성을 일반 시계열처럼 다룰 수 있게 된 거야.

## 네 질문에 대한 정리

- **"실제 데이터로 계산한다"** → 맞아. 모형 없이(model-free) 데이터에서 바로 계산해.
- **"historical 데이터를 쓴다"** → 과거 데이터라는 의미에서는 맞지만, 용어로서의 historical volatility와는 달라. Historical은 **여러 날을 묶어 하나의 값**을 내고, realized는 **하루 안을 쪼개 그날의 값**을 내.

그래서 realized skewness도 "과거 20일 일별 수익률의 왜도"가 아니라 **"t일 하루 동안 39개 봉 수익률 분포의 왜도"**라는 점을 기억하면 돼.

이 구분이 됐으면 2번 realized semivariance로 넘어가자.
