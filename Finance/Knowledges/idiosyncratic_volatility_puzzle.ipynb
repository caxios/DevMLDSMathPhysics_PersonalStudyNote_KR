# 5. 고유변동성 퍼즐 (Idiosyncratic Volatility Puzzle)

순서는 변동성 분해 → 분산투자로 사라지는 위험 → 이론의 예측 → 측정 방법 → 퍼즐의 발견 → 설명 후보들 → 함정 → 코드야. 이 개념은 앞의 베타 개념 위에 그대로 쌓여.

---

## 0단계: 종목 수익률을 두 부분으로 나누기

앞 개념에서 베타는 "시장과 같이 움직이는 정도"라고 했지. 그럼 **시장으로 설명되지 않는 나머지**가 있을 거야. 회귀식으로 쓰면:

$$
r_i = \alpha_i + \beta_i\, r_m + \varepsilon_i
$$

- $\beta_i r_m$: 시장 때문에 움직인 부분 → **체계적(systematic) 부분**
- $\varepsilon_i$: 시장과 무관하게 이 종목만의 이유(실적, 공시, 경영진 이슈 등)로 움직인 부분 → **고유(idiosyncratic) 부분**

"idiosyncratic"은 "그 개체에만 고유한"이라는 뜻이야.

---

## 1단계: 분산도 두 부분으로 깔끔하게 쪼개진다 (유도)

양변의 분산을 구하자. α는 상수라 분산이 0이야.

$$
\text{Var}(r_i) = \text{Var}(\beta_i r_m + \varepsilon_i)
$$

두 변수 합의 분산 공식은 $\text{Var}(X+Y) = \text{Var}(X) + \text{Var}(Y) + 2\,\text{Cov}(X,Y)$야.

$$
\text{Var}(r_i) = \beta_i^2\,\text{Var}(r_m) + \text{Var}(\varepsilon_i) + 2\beta_i\,\text{Cov}(r_m, \varepsilon_i)
$$

여기서 핵심은, **OLS 회귀는 잔차가 설명변수와 상관이 없도록 계수를 고른다**는 거야. 즉 $\text{Cov}(r_m, \varepsilon_i) = 0$이 **구조적으로** 보장돼. 마지막 항이 사라지니까:

$$
\boxed{\underbrace{\sigma_i^2}_{\text{총위험}} = \underbrace{\beta_i^2\,\sigma_m^2}_{\text{체계적 위험}} + \underbrace{\sigma_{\varepsilon,i}^2}_{\text{고유위험}}}
$$

**고유변동성(IVOL)**은 이 잔차의 표준편차 $\sigma_{\varepsilon,i}$야.

---

## 2단계: 고유위험은 분산투자로 사라진다 (유도)

종목 n개를 똑같은 비중($1/n$)으로 담은 포트폴리오를 생각하자. 계산을 단순하게 하려고 모든 종목의 베타가 β, 고유분산이 $\sigma_\varepsilon^2$로 같고, **종목끼리의 잔차는 서로 독립**이라고 가정할게. 고유 사건은 종목마다 따로 일어나니까 자연스러운 가정이야.

포트폴리오 수익률은:

$$
r_p = \frac{1}{n}\sum_{i=1}^{n} r_i = \alpha + \beta\, r_m + \frac{1}{n}\sum_{i=1}^{n}\varepsilon_i
$$

잔차 평균 부분의 분산을 구하면 (독립이라 공분산 항이 모두 0):

$$
\text{Var}\left(\frac{1}{n}\sum_{i=1}^{n}\varepsilon_i\right) = \frac{1}{n^2}\sum_{i=1}^{n}\sigma_\varepsilon^2 = \frac{1}{n^2}\cdot n\,\sigma_\varepsilon^2 = \frac{\sigma_\varepsilon^2}{n}
$$

따라서:

$$
\text{Var}(r_p) = \beta^2\sigma_m^2 + \frac{\sigma_\varepsilon^2}{n}
$$

**n이 커지면 두 번째 항은 0으로 가고, 첫 번째 항은 그대로 남아.** 한 종목의 악재는 다른 종목의 호재와 상쇄되지만, 시장 전체의 하락은 아무리 많이 담아도 피할 수 없다는 뜻이지.

---

## 3단계: 이론은 무엇을 예측하나

**CAPM의 예측: 고유변동성은 가격이 매겨지지 않는다.** 고유위험은 공짜로(분산투자만 하면) 없앨 수 있어. 없앨 수 있는 위험을 굳이 떠안은 사람에게 시장이 보상을 줄 이유가 없지. 그래서 기대수익률은 오직 베타(체계적 위험)로만 결정되고, IVOL과 미래 수익률의 관계는 **0**이어야 해.

**Merton(1987)의 예측: 고유변동성이 높을수록 기대수익률도 높다.** 현실의 투자자들은 정보 비용 등의 이유로 분산투자를 충분히 하지 못해. 그러면 고유위험을 실제로 떠안게 되고, 그 대가를 요구하니까 IVOL과 기대수익률의 관계는 **양(+)**이어야 해.

정리하면 이론은 **0 아니면 양수**를 예측해. 음수를 예측하는 주류 이론은 없었어.

---

## 4단계: IVOL을 실제로 측정하는 방법

Ang, Hodrick, Xing, Zhang(2006, Journal of Finance)의 방식이 표준이야. 매월 말, 지난 한 달(약 21 거래일)의 **일별** 수익률로 Fama-French 3팩터 회귀를 돌려.

$$
r_{i,d} - r_{f,d} = \alpha_i + \beta_i^{MKT}\,MKT_d + \beta_i^{SMB}\,SMB_d + \beta_i^{HML}\,HML_d + \varepsilon_{i,d}
$$

- $MKT$: 시장 초과수익률 (0단계의 $r_m$ 역할)
- $SMB$ (Small Minus Big): 소형주 수익률 − 대형주 수익률
- $HML$ (High Minus Low): 가치주(장부가/시가 높음) − 성장주 수익률

시장 하나만 쓰는 대신 규모와 가치 요인까지 걸러낸 **더 엄격한 잔차**를 구하는 거야. 그다음 이렇게 정의해.

$$
IVOL_{i,t} = \sqrt{\frac{1}{D-1}\sum_{d=1}^{D}\hat\varepsilon_{i,d}^2}
$$

(여기서는 여러 날을 묶었으니 앞에서 배운 구분으로 보면 **historical** 방식의 측정이야.)

---

## 5단계: 퍼즐의 발견

AHXZ(2006)는 매월 미국 주식들을 IVOL 기준으로 5분위로 나누고 **다음 달 수익률**을 봤어.

결과는 이론과 정반대였어. **IVOL이 가장 높은 5분위 종목들이 가장 낮은 1분위보다 다음 달 수익률이 크게 낮았고**, Fama-French 3팩터 알파 기준으로 그 차이가 월 약 −1.06%였어. 연율로 치면 10%가 넘는 차이야. 이후 Ang et al.(2009)은 여러 선진국 시장에서도 같은 패턴을 보고했어.

그래서 "퍼즐"이라고 불러. 이론이 예측한 0도 양수도 아닌, **음수**가 나왔으니까.

---

## 6단계: 숫자 예시

어떤 종목의 일별 수치가 이렇다고 하자.

- 시장 일변동성 $\sigma_m = 1\%$
- 베타 $\beta = 1.2$
- 종목 총 일변동성 $\sigma_i = 2.5\%$

1단계 공식에 넣으면:

| 구성 | 계산 | 값 |
|---|---|---|
| 총분산 | $2.5^2$ | 6.25 |
| 체계적 분산 | $1.2^2 \times 1^2$ | 1.44 (23%) |
| 고유분산 | $6.25 - 1.44$ | 4.81 (77%) |
| **IVOL (일별)** | $\sqrt{4.81}$ | **2.19%** |
| IVOL (연율) | $2.19 \times \sqrt{252}$ | ≈ 34.8% |

**주목할 점:** 개별 종목 변동의 **대부분(77%)이 고유 부분**이야. 이 비율이 체계적 부분이 설명하는 비율, 즉 회귀의 $R^2$(여기서는 23%)의 나머지야. 개별 주식에서는 이게 전형적인 수준이야.

분산투자 효과(2단계)를 이 숫자로 보면:

| 종목 수 n | 고유분산 $4.81/n$ | 포트폴리오 총분산 |
|---|---|---|
| 1 | 4.81 | 6.25 |
| 10 | 0.48 | 1.92 |
| 100 | 0.05 | 1.49 |

10종목만 담아도 고유위험이 거의 사라지고, 체계적 분산 1.44에 수렴해.

## 7단계: 왜 이런 일이 생기나 — 설명 후보들

아직 **합의된 정답은 없어.** 주요 설명 후보는 네 갈래야.

**(1) 복권 선호 (가장 유력한 설명 중 하나)**

IVOL이 높은 종목은 가끔 크게 튀어. 투자자들이 이런 "대박 가능성"을 복권처럼 좋아해서 비싸게 사고, 그래서 이후 수익률이 낮아진다는 거야. Bali, Cakici, Whitelaw(2011)는 **지난달 일별 최대 수익률(MAX)**이라는 단순한 지표를 만들었는데, 이걸 통제하면 IVOL 효과가 크게 약해지거나 사라진다고 보고했어.

앞에서 배운 개념과 연결되는 지점이야. RSkew나 RSJ가 높은 종목이 이후 부진하다는 결과와 **같은 계열의 설명**이야. 결국 "3차 모멘트(왜도)에 대한 선호"가 2차 모멘트(IVOL) 효과의 정체일 수 있다는 거지.

**(2) 단기 반전과 측정 방법의 문제**

- Fu(2009): 지난달 IVOL이 아니라 **다음 달의 기대 IVOL**(EGARCH로 추정)을 쓰면 관계가 **양(+)**으로 뒤집힌다고 주장했어. 지난달 IVOL이 높은 종목은 대개 지난달 급등한 소형주라, 퍼즐의 상당 부분이 단기 반전 효과라는 거야.
- Bali & Cakici(2008): 동일가중이냐 시가총액가중이냐, 소형·비유동 종목을 빼느냐에 따라 결과가 크게 흔들린다고 보고했어.

**(3) 차익거래의 비대칭 (네 프로젝트에 가장 중요)**

Stambaugh, Yu, Yuan(2015)의 논리는 이래.

- IVOL이 높은 종목은 차익거래가 **위험**해. 가격이 이상해 보여도 베팅했다가 크게 물릴 수 있으니까.
- 그래서 IVOL이 높은 종목일수록 **잘못된 가격이 오래 유지**돼.
- 그런데 **저평가**는 사면 되니 쉽게 교정되지만, **고평가**는 공매도를 해야 해서 교정이 훨씬 어려워.
- 결과적으로 고IVOL 종목 중에는 교정되지 못한 **고평가 종목이 많이 남고**, 이 종목들이 나중에 가격이 떨어지면서 음의 관계가 생긴다는 거야.

실제로 이들은 **고평가 종목 안에서는 IVOL-수익률 관계가 음(−)**, **저평가 종목 안에서는 양(+)**이라고 보고했어. 둘을 합치면 공매도 제약 때문에 음의 효과가 이긴다는 거지.

**(4) 개인투자자**

Han & Kumar(2013) 등은 개인투자자 거래 비중이 높은 종목에서 IVOL 효과가 집중된다고 보고했어. 개인이 복권형 종목을 선호한다는 (1)과 이어지는 이야기야.

---

## 8단계: 네 프로젝트에 쓸 때의 함정

**IVOL은 총변동성과 거의 같은 변수야.** 6단계에서 봤듯 개별 종목 변동의 대부분이 고유 부분이라, IVOL과 총변동성의 상관이 매우 높아. 그래서 "고유변동성 퍼즐"과 "저변동성 이상현상(low-volatility anomaly)"은 사실상 같은 현상을 다른 이름으로 부르는 경우가 많아. 네 연구에서 "변동성이 미래 수익률에 주는 시그널"이라고 할 때, 이게 **새로운 발견인지 IVOL 퍼즐의 재발견인지** 반드시 구분해야 해.

**앞에서 배운 지표들과 서로 겹쳐.** IVOL, MAX, RSkew, RSJ, 전월 수익률(반전)은 모두 서로 상관이 높아. 이 중 하나로 "효과를 발견했다"고 하려면 **나머지를 전부 통제변수로 넣고도 살아남는지** 보여줘야 해. 안 그러면 심사자나 면접관이 가장 먼저 지적할 부분이야.

**한국 시장에서는 소형주에 결과가 휘둘려.** 코스닥의 저가·비유동 종목이 결과를 좌우할 가능성이 높아. Bali & Cakici의 지적처럼 **시가총액가중 결과**, **가격·유동성 필터 적용 결과**를 동일가중 결과와 반드시 같이 보고해야 해.

**한국 팩터 데이터 문제.** 미국은 Kenneth French 웹사이트에서 팩터를 받으면 되지만, 한국은 SMB·HML을 직접 구축해야 할 수도 있어. 처음엔 **시장 팩터 하나만 쓴 CAPM 잔차**로 시작해도 괜찮아. IVOL 순위는 어떤 모형을 쓰든 크게 달라지지 않는 경우가 많아.

---

## 계산 코드

```python
import numpy as np
import pandas as pd

def ivol_capm(g: pd.DataFrame) -> float:
    """g: 한 종목의 한 달치 일별 [exret, mkt_exret]"""
    g = g.dropna()
    if len(g) < 15:                       # 최소 관측치 필터
        return np.nan
    X = np.column_stack([np.ones(len(g)), g['mkt_exret'].values])
    y = g['exret'].values
    coef, *_ = np.linalg.lstsq(X, y, rcond=None)
    resid = y - X @ coef
    return resid.std(ddof=2)              # 추정한 계수 2개만큼 자유도 차감

daily['month'] = daily['date'].dt.to_period('M')
ivol = (daily.groupby(['ticker', 'month'])
             .apply(ivol_capm)
             .rename('IVOL')
             .reset_index())

# 다음 달 수익률과 결합 후 5분위 포트폴리오
ivol['q'] = ivol.groupby('month')['IVOL'].transform(
    lambda x: pd.qcut(x, 5, labels=False, duplicates='drop'))
```

Fama-French 3팩터로 확장하려면 `X`에 SMB, HML 열만 추가하고 `ddof=4`로 바꾸면 돼.

## 네 프로젝트와의 연결

7단계의 Stambaugh-Yu-Yuan 논리는 네 프로젝트에 **바로 쓸 수 있는 가설**을 줘.

> 공매도가 막히면 고평가가 더 교정되지 않으니, **2023년 11월~2025년 3월 공매도 전면 금지 기간에는 IVOL 퍼즐(음의 관계)이 더 강해졌어야 한다.**

이건 기존 이론에서 직접 도출되는 예측이고, 한국 시장이 준 자연실험으로 검정할 수 있어. 결과가 예측대로 나오든 안 나오든 그 자체로 의미 있는 발견이야. "통계적으로 추정한 레짐"보다 이런 **제도적 레짐**이 훨씬 설득력 있다고 앞에서 말한 이유가 이거야. 이게 확정된 결과는 아니고, 네가 검정해야 할 가설이라는 점은 분명히 해둘게.

---

**한 줄 요약:** 종목 변동성은 시장 때문에 생기는 부분(β²σ²ₘ)과 종목 고유의 부분(σ²ₑ)으로 정확히 나뉘고, 후자는 분산투자로 사라지니 이론상 보상받지 못하거나(CAPM) 오히려 보상받아야(Merton) 해. 그런데 실제로는 **고유변동성이 높은 종목의 미래 수익률이 낮아서** 퍼즐이라고 불러. 유력한 설명은 복권 선호, 단기 반전, 그리고 공매도 제약으로 인한 고평가 지속이야.

다음 6번 **volatility-managed portfolio**는 방향을 바꿔서, 변동성을 **종목 선택**이 아니라 **투자 비중 조절**에 쓰는 이야기야. 준비되면 말해줘.
