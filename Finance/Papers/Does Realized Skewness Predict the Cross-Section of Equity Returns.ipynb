Does Realized Skewness Predict the Cross-Section of Equity Returns?

Amaya, Christoffersen, Jacobs, Vasquez

제시된 논문 **"Does Realized Skewness Predict the Cross-Section of Equity Returns?"** (Diego Amaya, Kris Jacobs, Peter Christoffersen, Aurelio Vasquez, 2015)에 대한 종합 요약 및 주요 수식 분석입니다.

---

### 1. 논문 개요 및 핵심 결론

* **연구 배경 및 목적**: 기존의 실현 변동성(Realized Volatility) 문헌을 확장하여, 5분 단위의 장중(Intraday) 고빈도 데이터를 통해 개별 주식의 주간 실현 변동성, 실현 왜도, 실현 첨도를 측정하고, 이 실현 적률들이 차주 주식 수익률의 횡단면(Cross-section)을 예측할 수 있는지 분석합니다.


* **분석 데이터**: 1993년 1월부터 2013년 12월까지 21년간 TAQ(Trade and Quote) 데이터베이스에 수록된 모든 상장 주식(300만 개 이상의 firm-week 관측치)을 활용했습니다.


* **핵심 결론**:
1. **실현 왜도(Realized Skewness)의 강력한 음(-)의 횡단면 예측력**: 당주의 실현 왜도와 차주의 주식 수익률 사이에는 통계적으로 매우 유의미한 음(-)의 관계가 존재합니다. 실현 왜도가 가장 낮은 10분위 포트폴리오를 매수하고 가장 높은 10분위 포트폴리오를 매도하는 롱-숏(Long-Short) 전략은 가치평중(Value-weighted) 기준 주당 19 bps (연율화 약 10%, $t$ -통계량 3.70), 동일가중(Equal-weighted) 기준 주당 34 bps ($t$ -통계량 8.86)의 초과수익을 기록했습니다. 이 성과는 Carhart 4요인 알파로도 설명되지 않습니다.


2. **실현 첨도(Realized Kurtosis)의 양(+)의 관계**: 실현 첨도는 차주 수익률과 대체로 양(+)의 관계를 보이나, 설정에 따라 통계적 유의성이 유지되지 않는 등 실현 왜도에 비해 덜 강건합니다.


3. **실현 변동성(Realized Volatility)의 무의미성**: 주간 단위 횡단면 분석에서 실현 변동성은 차주 주식 수익률에 대해 통계적으로 유의한 예측력을 보이지 못했습니다.

### 2. 핵심 이론 및 주요 수식 설명

#### (1) 장중 고빈도 데이터를 활용한 일별 및 주별 실현 적률 정의

거래일 $t$ 의 $i$ 번째 5분 로그 수익률을 $r_{t,i}$ 라 정의합니다 ($N = 78$, 6.5시간 거래 기준).

$$r_{t,i} = p_{t,\frac{i}{N}} - p_{t,\frac{i-1}{N}}$$

여기서 $p$ 는 로그 가격입니다.

* **일별 실현 분산 (Realized Daily Variance)**:



$$RDVar_{t} = \sum_{i=1}^{N} r_{t,i}^{2}$$

고빈도 관측 환경에서 5분 수익률의 평균(drift)은 분산 대비 크기가 매우 작으므로 0으로 가정합니다.

* **일별 실현 왜도 (Realized Daily Skewness)**:



$$RDSkew_{t} = \frac{\sqrt{N} \sum_{i=1}^{N} r_{t,i}^{3}}{RDVar_{t}^{3/2}}$$

분자는 장중 3차 적률 합이며, 분모는 일별 실현 분산의 $3/2$ 승(실현 표준편차의 세제곱)으로 표준화한 형태입니다. $\sqrt{N}$ 스케일링을 통해 일별 왜도의 척도와 일치하도록 보정합니다.

* **일별 실현 첨도 (Realized Daily Kurtosis)**:



$$RDKurt_{t} = \frac{N \sum_{i=1}^{N} r_{t,i}^{4}}{RDVar_{t}^{2}}$$

장중 4차 적률 합을 실현 분산의 제곱으로 나누어 표준화하고 $N$ 을 곱해 일별 첨도 척도로 환산합니다.

* **주간 집계 (Weekly Realized Moments)**:
자산가격결정 분석이 주간(수요일~차주 화요일) 단위로 진행되므로, 5영업일의 일별 측정치를 집계합니다.



$$RVol_{t} = \left( \frac{252}{5} \sum_{i=0}^{4} RDVar_{t-i} \right)^{1/2}$$

$$RSkew_{t} = \frac{1}{5} \sum_{i=0}^{4} RDSkew_{t-i}$$

$$RKurt_{t} = \frac{1}{5} \sum_{i=0}^{4} RDKurt_{t-i}$$

여기서 $RVol_{t}$ 는 연율화된 주간 실현 변동성입니다.

#### (2) 연속 시간 점프-확산 모형에서의 극한 특성 (Limiting Properties)

기저 자산의 로그 가격 $p_{T}$ 가 확률적 변동성과 점프(Jump)를 포함하는 세미마틴게일(Semi-martingale) 과정을 따른다고 가정합니다.

$$p_{T} = \int_{0}^{T} \mu_{s} ds + \int_{0}^{T} \sigma_{s} dW_{s} + J_{T}$$

여기서 $\mu_{s}$ 는 표류율, $\sigma_{s}$ 는 변동성 과정, $W_{s}$ 는 표준 브라운 운동, $J_{T}$ 는 순수 점프 과정입니다.

샘플링 빈도 $N \to \infty$ ($dt = T/N \to 0$)일 때 각 표본 적률의 확률수렴(convergence in probability, $\xrightarrow{p}$) 극한은 다음과 같습니다:

1. **2차 실현 적률 (Realized Variance)**:



$$RM(2) \equiv \sum_{i=1}^{N} \left( p_{T\frac{i}{N}} - p_{T\frac{i-1}{N}} \right)^{2} \xrightarrow{p} \int_{0}^{T} \sigma_{s}^{2} ds + \sum_{0 < s \le T} (\Delta p_{s})^{2}$$

* **수식 의미**: 극한에서 연속적 확산에 의한 적분 분산(Integrated Variance)과 불연속 점프 크기의 제곱합($\sum (\Delta p_s)^2$)이 **모두** 남습니다.



2. **3차 실현 적률 (Realized Third Moment)**:



$$RM(3) \equiv \sum_{i=1}^{N} \left( p_{T\frac{i}{N}} - p_{T\frac{i-1}{N}} \right)^{3} \xrightarrow{p} \sum_{0 < s \le T} (\Delta p_{s})^{3}$$

3. **4차 실현 적률 (Realized Fourth Moment)**:



$$RM(4) \equiv \sum_{i=1}^{N} \left( p_{T\frac{i}{N}} - p_{T\frac{i-1}{N}} \right)^{4} \xrightarrow{p} \sum_{0 < s \le T} (\Delta p_{s})^{4}$$

* **수식에 기반한 직관적 설명**:
* 브라운 운동 증분의 크기는 확률적으로 $dW_s \sim O_p(\sqrt{dt}) = O_p(N^{-1/2})$ 의 차수를 가집니다.


* 3차 항의 합을 계산하면 연속 확산 성분은 $\sum_{i=1}^{N} (N^{-1/2})^3 = N \cdot O_p(N^{-3/2}) = O_p(N^{-1/2}) \xrightarrow{p} 0$ 이 되어 극한에서 소멸합니다.


* 반면, 유한한 횟수로 발생하는 불연속 점프 $\Delta p_s = p_s - p_{s-}$ 는 표본 분할 수 $N$ 에 비례하여 줄어들지 않으므로 극한에서도 그대로 살아남습니다.


* 따라서 고빈도 장중 데이터로 계산한 실현 왜도 $RM(3)$ 과 실현 첨도 $RM(4)$ 는 확산 성분(예: 레버리지 효과로 인한 연속적 비대칭성이나 변동성의 변동성)을 제외하고, **오직 점프(Jump) 성분의 크기와 방향만을 추출**합니다. 이는 일별 저빈도 데이터로 추정한 역사적 왜도(Historical Skewness)와 정보의 성격이 수학적으로 완전히 다름을 의미합니다.

#### (3) 아핀 점프-확산 모형(Affine Jump-Diffusion Model)에서의 이론적 기댓값

Heston 형태의 확률적 변동성과 복합 포아송 점프를 따르는 구체적인 모형을 설정합니다:

$$dp_{t} = \left( \mu - \frac{1}{2}V_{t} - \bar{\mu}_{J}\lambda \right) dt + \sqrt{V_{t}} dW_{t}^{(1)} + J dN_{t}$$

$$dV_{t} = \kappa(\theta - V_{t}) dt + \sigma \sqrt{V_{t}} dW_{t}^{(2)}$$

여기서 $Corr(dW_t^{(1)}, dW_t^{(2)}) = \rho$ 이고, 점프 도착 강도는 $\lambda$, 점프 크기 $J \sim N(\mu_J, \sigma_J^2)$ 입니다.

적률생성함수(MGF)의 테일러 전개를 통해 구한 실현 적률 극한의 기댓값은 다음과 같습니다:

$$IM(2) \equiv \lim_{N \to \infty} E_0[RM(2)] = \theta T - (\theta - V_0)\frac{1 - e^{-\kappa T}}{\kappa} + \lambda (\mu_J^2 + \sigma_J^2) T$$

$$IM(3) \equiv \lim_{N \to \infty} E_0[RM(3)] = \lambda (\mu_J^3 + 3\mu_J \sigma_J^2) T$$

$$IM(4) \equiv \lim_{N \to \infty} E_0[RM(4)] = \lambda (\mu_J^4 + 6\mu_J^2 \sigma_J^2 + 3\sigma_J^4) T$$

* **수식의 의미**:
* $IM(2)$ 에는 확산 모수($\theta, \kappa, V_0$)와 점프 모수($\lambda, \mu_J, \sigma_J$)가 모두 포함되어 있습니다.


* 반면 $IM(3)$ 과 $IM(4)$ 의 수식에는 확산 모수 및 브라운 운동 간의 상관계수 $\rho$ (레버리지 효과), 변동성의 변동성 $\sigma$ 가 전혀 포함되지 않으며, **오직 점프 강도 $\lambda$ 와 점프 분포 모수($\mu_J, \sigma_J$)로만 결정**됩니다.


* $IM(3)$ 의 부호는 평균 점프 크기 $\mu_J$ 에 의해 결정되므로, 실현 왜도는 가격 상승 점프와 하락 점프 간의 비대칭성을 정확하게 포착합니다.

#### (4) 파마-맥베스(Fama-MacBeth) 횡단면 회귀 모형

매주 $t$ 마다 개별 주식 $i$ 에 대해 다음 횡단면 회귀식을 추정합니다:

$$r_{i,t+1} = \gamma_{0,t} + \gamma_{1,t} RVol_{i,t} + \gamma_{2,t} RSkew_{i,t} + \gamma_{3,t} RKurt_{i,t} + \mathbf{\gamma}_{4,t}' \mathbf{Z}_{i,t} + \epsilon_{i,t+1}$$

여기서 $r_{i,t+1}$ 은 차주 수익률(bps 단위)이며, $\mathbf{Z}_{i,t}$ 는 기업 규모($\log(Size)$), 장부가-시가 비율($\log(BE/ME)$), 시장 베타($\beta$), 1주 지연 수익률(Lagged Return), 역사적 왜도, 고유 변동성(Idiosyncratic Volatility), 공왜도(Coskewness), 전월/전주 최대 및 최소 수익률, 비유동성(Amihud Illiquidity), 장중 거래 횟수 등 기존에 알려진 가격 결정 요인 벡터입니다.

### 4. 경제학적 함의 및 추가 분석

1. **단기 반전 효과(Short-term Reversal)와의 차별성**:
* 실현 왜도는 전주 지연 수익률(Lagged Return)과 약 $22.3\%$ 의 양의 상관관계를 가집니다.


* 그러나 지연 수익률과 실현 왜도를 결합한 포트폴리오를 구성했을 때, 실현 왜도 기반 전략은 지연 수익률 전략에 비해 표준편차가 작고 꼬리 위험(Value-at-Risk, Expected Shortfall)이 현저히 낮아 더 높은 연율화 샤프 지수(Sharpe Ratio)를 제공합니다.


* 지연 수익률은 확산과 점프를 모두 포함하는 반면, 실현 왜도는 불연속 점프 충격만을 순수하게 반영하기 때문에 독립적인 정보를 제공합니다.




2. **고유 변동성 퍼즐(Idiosyncratic Volatility Puzzle)의 규명**:
* Ang et al.(2006)이 발견한 "고유 변동성이 높은 주식이 낮은 수익률을 기록하는 퍼즐"에 대해, 논문은 이중 정렬(Double Sort)을 통해 해답을 제시합니다.


* 왜도가 낮은(음의 왜도) 주식 집단에서는 변동성이 커질수록 투자자들이 높은 위험 프리미엄(높은 미래 수익률)을 요구합니다.


* 반면, 왜도가 높은(양의 왜도, 복권형 주식) 주식 집단에서는 투자자들이 극단적인 양의 점프 가능성을 선호(Skewness Preference)하여 높은 변동성을 감수하고도 낮은 기대수익률을 기꺼이 받아들이기 때문에 음의 관계가 나타납니다.


* 즉, 실현 왜도를 통제하면 고유 변동성과 기대수익률 간의 상충 관계가 합리적으로 설명될 수 있습니다.
