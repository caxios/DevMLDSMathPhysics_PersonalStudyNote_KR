Deep FinResearch Bench: Evaluating AI’s Ability to Conduct Professional Financial Investment Research
link : https://arxiv.org/pdf/2604.21006#page=12.78

제시된 논문(arXiv:2604.21006v1)은 JPMorgan Chase AI Research 팀이 발표한 연구로, 최신 AI 심층 리서치(Deep Research, DR) 에이전트가 전문 금융 애널리스트 수준의 주식 투자 리서치를 수행할 수 있는지를 실증적으로 평가한 논문입니다.

이 논문의 핵심 내용을 배경, 평가 프레임워크, 세부 평가 지표, 실험 결과 및 한계점 순으로 상세히 정리해 드립니다.

---

### 1. 연구 배경 및 동기 (Introduction & Background)

* **배경** : 대규모 언어 모델(LLM)과 검색·계획·추론을 결합한 심층 리서치(Deep Research, DR) 에이전트가 등장하면서 복잡한 금융 투자 리서치 업무를 자동화할 수 있는지에 대한 관심이 급증했습니다.


* **기존 벤치마크의 한계** : 기존의 AI 평가 벤치마크는 일반적이거나 단답형 Q&A 위주이며, 실제 증권사나 투자은행(IB)의 주식 리서치(Equity Research) 실무 프로세스(다단계 정보 수집, 재무 예측, 가치평가 모델링, 투자 논거 제시 등)를 충분히 반영하지 못했습니다.


* **목적** : 실제 금융기관의 전문 애널리스트가 작성한 리서치 보고서와 최신 상용 DR 에이전트가 작성한 보고서를 정량적·정성적·신뢰성 측면에서 포괄적으로 비교·평가하는 표준화된 프레임워크인 **Deep FinResearch Bench** 를 구축했습니다.



---

### 2. 평가 프레임워크 (Deep FinResearch Bench Framework)

이 연구의 벤치마킹 파이프라인은 크게 세 단계로 구성됩니다:

1. **전문 애널리스트 보고서 수집 (Professional Report Collection)**

* Yahoo Finance 구독을 통해 실제 금융기관 2곳(Firm A, Firm B)에서 발간한 사전 실적 발표(pre-earnings) 주식 리서치 보고서 100건을 수집했습니다.


* S&P 500에 속한 25개 기업(정보기술, 금융, 헬스케어 섹터)의 2025 회계연도 1분기(Q1) 및 2분기(Q2) 대상 보고서로 구성되었습니다.




2. **AI 보고서 생성 (AI Report Generation)**

* 평가 대상 에이전트: 최상위 상용 DR 에이전트 4종인 **OpenAI DR** (GPT-o3 기반), **Gemini DR** (Gemini 2.5 Pro 기반), **Grok DR** (Grok 4 기반), **Perplexity DR** (Sonar Reasoning 기반)을 평가했습니다.


* 공정한 비교를 위해 전문 보고서에서 세부 루브릭(Rubric)을 추출하여 프롬프트에 반영하였으며, 보고서 발간 시점 이후의 정보 유출을 막기 위해 엄격한 **지식 컷오프(Knowledge Cut-off)** 제약을 적용했습니다.




3. **종합 평가 파이프라인 (Evaluation Pipeline)**

* 정성적 분석, 정량적 정확도, 검증 가능성 및 출처 신뢰도의 3대 축을 기준으로 평가합니다.





---

### 3. 세부 평가 지표 (Evaluation Metrics)

#### (1) 정성적 분석 지표 (Qualitative Analysis)

보고서의 질을 전문가 기준에 맞춰 4점 척도(1=Poor, 2=Fair, 3=Good, 4=Excellent)로 평가합니다. 대규모 평가를 위해 도메인 전문가 평가와 일치율이 가장 높은 GPT-5를 **LLM-as-a-Judge** 로 채택했습니다.

* **포괄성 (Comprehensiveness)** : 필수 섹션 포함 여부, 섹터별 핵심 지표(SaaS의 ARR/NRR, 은행의 NII/NIM 등) 반영 여부, 내용의 비례적 깊이.


* **논리적 일관성 (Coherence)** : 섹션 간 인과관계 연결, 보고서 전반의 서술·도표·가정·날짜 간 모순 부재 여부.


* **가정의 질 (Assumption Quality)** : 미래 실적 추정 및 밸류에이션의 핵심 가정들이 명시적인지, 과거 데이터나 벤치마크로 뒷받침되는지, 시나리오 분석(Bull/Bear 케이스)을 통해 스트레스 테스트를 거쳤는지 여부.


* **분석적 깊이 (Analytical Depth)** : 단순 현상 서술을 넘어 결과가 발생한 근본적 인과 메커니즘을 설명하는지, 투자자에게 실행 가능한 통찰을 주는지 여부.



#### (2) 정량적 정확도 지표 (Quantitative Analysis)

* **재무 예측 정확도 (Financial Forecast Accuracy)** : 향후 1~3년의 매출(Revenue), EBITDA, 영업이익, 순이익, FCF, EPS 예측 오차를 대칭 평균 절대 백분율 오차(SMAPE)로 측정합니다.



$$SMAPE = \frac{100\%}{n} \sum_{i=1}^{n} \frac{\vert{}A_i - F_i\vert{}}{\frac{\vert{}A_i\vert{} + \vert{}F_i\vert{}}{2}}$$


(여기서 $A_i$ 는 실제 실적값, $F_i$ 는 예측값, $n$ 은 기업 수)

* **주식 밸류에이션 정확도 (Stock Valuation Accuracy)** : 3개월 및 6개월 시계($h$)에서 적정가치 추정치(FVE)의 적합성을 평가합니다.


* **Hit Rate** : 목표 주가 도달 여부.


* **MAE** : 추정 적정가치와 실현 주가 간의 백분율 오차.


* **Mean Bias** : 체계적인 낙관/보수 편향을 확인하는 부호화된 오차.




* **투자 추천 정확도 (Investment Recommendation Accuracy)** :
* **방향성 정확도 (Directional Accuracy, DA)** : BUY/SELL 추천이 실제 초과 수익 방향과 일치했는지의 비율.


* **부호화 추천 손실 (Signed Recommendation Loss, SRL)** : 시장 벤치마크 대비 초과 수익률($e_i$)의 크기를 가중치로 부여하여 잘못된 추천 신호에 페널티를 부과하는 손실 함수.





$$SRL_i = \vert{}e_i\vert{} \cdot \mathbf{1}[\mathrm{sign}(e_i) \ne s_i] + \lambda \vert{}e_i\vert{} \cdot \mathbf{1}[s_i = 0 \wedge \vert{}e_i\vert{} > \tau]$$

#### (3) 검증 가능성 및 신뢰성 지표 (Verifiability & Credibility)

보고서 문장을 숫자형 클레임(Numeric Claims)과 서술형 클레임(Descriptive Claims)으로 분해한 뒤 웹 검색 검증기로 사실 여부를 확인합니다.

* **사실성 비율 ($F(R)$)** : 제공된 인용 출처를 통해 사실로 확인된 클레임의 비율.


* **환각 비율 ($H(R)$)** : 인용 출처에 해당 내용이 전혀 존재하지 않는 비율.


* **미검증 비율 ($NV(R)$)** : 링크 오류, 동적 웹페이지, 비공개 등으로 외부 검증이 불가능한 비율.


* **출처 신뢰도 (Credibility)** : SEC 공시(sec.gov), 로이터, 블룸버그, 기업 IR 등 공인된 고신뢰 도메인($\mathcal{D}_{hi}$)과 소셜 미디어(x.com), 블로그 등 저신뢰 도메인($\mathcal{D}_{lo}$)의 인용 비율 분석.

### 4. 실험 결과 및 주요 발견 (Findings)

#### (1) 정성적 평가 결과

* **전문가와의 격차** : 전문 애널리스트(Firm A: 종합 2.84점)가 모든 DR 에이전트(2.022.31점)보다 확연히 높은 점수를 기록했습니다. 특히 분석적 깊이(Firm A 3.73점 vs AI 2.403.00점)와 논리적 일관성(Firm A 3.00점 vs AI 1.80~2.20점)에서 격차가 매우 컸습니다.


* **AI 에이전트 간 성능** : DR 에이전트 중에서는 **Gemini DR** (종합 2.31점)이 가장 우수한 정성적 품질을 보였으며, 가정의 명시성 측면에서는 인간 애널리스트보다 소폭 높은 점수를 받기도 했습니다.



#### (2) 정량적 예측 및 밸류에이션 결과

* **재무제표 예측** : 전문 애널리스트가 핵심 손익계산서 항목(매출, EBITDA, 영업이익, 순이익)에서 DR 에이전트를 압도하는 정밀도(가장 낮은 SMAPE)를 보였습니다. AI 중에서는 Grok과 Gemini가 상대적으로 양호했습니다.


* **목표 주가 및 투자 추천** : 단기~~중기(3~~6개월) 주가 방향성이나 추천 손실(SRL) 측면에서는 AI 에이전트(OpenAI DR, Gemini DR)가 전문 애널리스트와의 격차를 크게 좁히거나 유사한 성과를 보였습니다.



#### (3) 검증 가능성 및 환각(Hallucination)

* **심각한 환각 문제** : DR 에이전트가 생성한 주장 중 약 10~20%(Grok의 경우 34% 이상)가 인용 출처로 증명되지 않는 환각이거나 출처 불명으로 나타났습니다.


* **구체적 오류 사례** :
* 실재하지 않는 웹 링크를 생성하는 경우(존재하지 않는 PR Newswire URL 인용).


* 실제 인용 페이지의 내용과 다른 날짜나 수치를 기재하는 경우(예: 규제 기관의 조사 종결 날짜 왜곡).




* **출처 신뢰도 편차** : OpenAI DR(고신뢰 비중 88.2%)과 Perplexity DR(74.6%)은 공시 자료 및 메이저 뉴스 등 공인된 출처를 주로 참조했으나, Grok DR과 Gemini DR은 상대적으로 저신뢰 소셜 웹페이지 인용 비중이 높았습니다.



#### (4) 단순 검색형 LLM vs DR 에이전트 비교

* 계획-검색-추론 루프를 수행하는 DR 에이전트가 단순 웹 검색 결합형 LLM(예: GPT-4o + Search)보다 정성적 보고서 완성도 및 재무 지표 예측에서 월등히 뛰어난 성능을 입증했습니다.

### 5. AI 리서치 에이전트의 구조적 결함 (Failure Modes)

논문은 사례 분석을 통해 현재 DR 에이전트가 전문 애널리스트를 대체하기 어려운 핵심 이유를 다음과 같이 정리했습니다:

1. **상향식(Bottom-up) 모델링 부재** : 전문 애널리스트는 제품군/서비스 세그먼트 단위(예: iPhone 판매 대수, ASP 등)부터 분기별로 차곡차곡 쌓아 올리는 모델을 구축하는 반면, DR 에이전트는 회사 전체의 연간 총액 단위로만 거칠게 추정합니다.


2. **섹션 간 수치 모순(Coherence 결여)** : 본문 서술에서 가정한 연평균 성장률(CAGR)과 뒤쪽 재무제표 테이블에 기재된 수치 간의 계산이 맞지 않는 등, 섹션별 병렬 생성으로 인한 내부 불일치가 자주 발생했습니다.


3. **시나리오 및 민감도 분석(Sensitivity Analysis) 결여** : 전문가는 규제 리스크나 거시경제 변화에 따른 Bull/Bear 시나리오와 밸류에이션 범위를 제시하지만, DR 에이전트는 단일 목표가만 제시하는 블랙박스식 추정에 머물렀습니다.


4. **인과 추론 대신 단순 서술 요약** : 경영진 가이던스의 신뢰성을 비판적으로 검토하거나 촉매제(Catalyst)를 깊이 있게 분석하지 못하고, 공개된 웹 뉴스를 단순 요약·재구성하는 수준에 그쳤습니다.



---

### 6. 결론 및 시사점 (Conclusion)

* **결론** : 현재의 최신 범용 심층 리서치(DR) 에이전트는 금융 리서치 보고서의 외형적 구조와 수치적 형태는 그럴듯하게 흉내 낼 수 있으나, **세부 정량 예측의 정확도, 심층적 인과 추론, 시나리오 분석, 주장 검증 가능성** 전반에서 전문 금융 분석가에 크게 미치지 못합니다.


* **의의** : 이 연구는 상용 AI가 금융 전문직 업무를 당장 대체하기 어렵다는 점을 객관적으로 입증하였으며, 향후 금융에 특화된 심층 연구 에이전트 아키텍처(엄격한 회계 제약 조건, 세그먼트 기반 상향식 모델링, 고신뢰 공시 데이터 기반 팩트체킹 등) 개발을 위한 표준 평가 기준을 제공합니다.
